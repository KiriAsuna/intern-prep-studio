## 1. dict: storing and retrieving by name

A dict stores "key → value" pairs, like a named list in R:

```r
info <- list(name = "Ann", age = 28)
info$age
```

```python
info = {"name": "Ann", "age": 28}
info["age"]
```

Create it with curly braces, put a colon between each key and value, and retrieve a value with square brackets and the key.

````python
info = {"name": "Ann", "age": 28}
print(info["name"])
print(info["age"])

info["age"] = 29            # modify
info["city"] = "Seattle"    # if the key doesn't exist, this adds it
print(info)
print(len(info))
````

## 2. When a key doesn't exist

Accessing a key that doesn't exist raises a `KeyError`. Two ways to avoid it:

- `"Eve" in scores`: check whether the key is there first;
- `scores.get("Eve", 0)`: if the key is missing, return the default you supply (0 here).

`.get(key, default)` is used constantly; the counting pattern below depends on it.

````python
scores = {"Ann": 90, "Bob": 82}
print("Ann" in scores)
print("Eve" in scores)
print(scores.get("Ann", 0))
print(scores.get("Eve", 0))
````

## 3. Looping over a dict

`.items()` gives you the key and the value at once.

````python
scores = {"Ann": 90, "Bob": 82, "Cy": 70}
for name, s in scores.items():
    print(name, s)

print(list(scores.keys()))
print(list(scores.values()))
````

## 4. Pattern 4: counting with a dict

In R, a single `table(x)` counts how many times each value appears. The basic Python version is:

```python
counts = {}
for x in xs:
    counts[x] = counts.get(x, 0) + 1
```

Read it word by word: `counts.get(x, 0)` is "how many times x has appeared so far, or 0 if never seen"; add 1 and store it back.

````python
votes = ["A", "B", "A", "C", "A", "B"]
counts = {}
for v in votes:
    counts[v] = counts.get(v, 0) + 1
    print(v, "->", counts)    # watch it change at each step
````

## 5. set: a collection with no duplicates

A set only cares about "is it there or not," not about order or counts.

- `set(xs)`: removes duplicates, like R's `unique()`, but **order is not guaranteed**;
- `s.add(x)`: adds an element (lists use `append`, sets use `add`);
- `x in s`: membership test, **very fast**; nearly instant no matter how much data.

````python
xs = [3, 1, 3, 2, 1, 5]
s = set(xs)
print(s)
print(len(s))        # how many distinct values

seen = set()
seen.add(3)
seen.add(3)          # adding a duplicate has no effect
seen.add(7)
print(seen)
print(3 in seen, 4 in seen)
````

## 6. tuple: a list you can't change

Use parentheses: `point = (3, 4)`. Indexing works like a list (`point[0]`), but it can't be modified after creation.

You've already seen one: `return min(xs), max(xs)` in 01d returns a tuple.

## 7. Choosing among the four containers

| Need | Use |
|---|---|
| An ordered sequence of data | `list` |
| Look up values by name, counting | `dict` |
| Remove duplicates, check "seen it before?" | `set` |
| A few fixed values bundled together | `tuple` |

If you can say in an interview "I'm using a set here because membership checks are fast," you're already ahead of many candidates.
