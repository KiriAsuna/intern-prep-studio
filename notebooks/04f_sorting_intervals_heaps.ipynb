## 1. Custom sorting

`sorted(xs, key=func)`: sorts by the function's return value. `key` is often a `lambda` (a small throwaway function).

Multiple sort criteria: have `key` return a **tuple**, and Python compares the items in order. To make a numeric item descending, negate it.

````python
people = [("Ann", 28), ("Bob", 35), ("Cy", 28), ("Di", 41)]

print(sorted(people, key=lambda p: p[1]))               # by age, ascending
print(sorted(people, key=lambda p: p[1], reverse=True)) # by age, descending
print(sorted(people, key=lambda p: (-p[1], p[0])))      # age descending, then name ascending for ties

words = ["banana", "fig", "apple", "kiwi"]
print(sorted(words, key=lambda w: (len(w), w)))         # by length first, then alphabetically
````

```python
def merge_intervals(intervals):
    out = []
    for start, end in sorted(intervals):
        if out and start <= out[-1][1]:
            out[-1][1] = max(out[-1][1], end)
        else:
            out.append([start, end])
    return out
```

## 3. Heaps: the minimum on demand

A **heap** is a structure that can quickly hand you the **minimum**: both push and pop are O(log n). Python uses the `heapq` module, which operates directly on a list.

- `heapq.heappush(h, x)`: push;
- `heapq.heappop(h)`: remove and return the smallest;
- `h[0]`: peek at the smallest;
- `heapq.nlargest(k, xs)` / `heapq.nsmallest(k, xs)`: get the k largest / smallest directly.

`heapq` only provides a min-heap. When you need a max-heap, push the negated numbers.

When to use one: you need to **repeatedly** take the current smallest / largest element while new elements keep arriving. Typical problems: "Kth largest," "Top K," "merge several sorted lists," "minimum number of meeting rooms."

````python
import heapq

h = []
for x in [5, 1, 8, 3]:
    heapq.heappush(h, x)
print(h[0])                                  # the smallest: 1
print(heapq.heappop(h), heapq.heappop(h))    # 1 3
print(heapq.nlargest(2, [5, 1, 8, 3]))       # [8, 5]
````

**The Kth largest element**: maintain a min-heap of size k that always holds "the k largest numbers seen so far." The top of the heap (the smallest of them) is the kth largest. Push each new number, and whenever the heap grows beyond k, pop the smallest.

Complexity O(n log k), better than the O(n log n) of a full sort, and well suited to data that keeps streaming in.
