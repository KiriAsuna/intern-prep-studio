```python
raw = pd.DataFrame({
    "id": [1, 2, 3, 4, 5, 6],
    "segment": [" A", "a", "B", "b ", "B", "A"],
    "date": ["2026-01-01", "2026-01-03", "2026-01-02", "2026-01-04", "bad", "2026-01-05"],
    "revenue": ["10", "20", None, "50", "40", "n/a"],
    "active": [1, 1, 0, 1, 1, 1],
})
```

### Problem 1 · Missing values per column

Number of missing values in each column of `users`.

```python
users.isna().sum()
```

**Result:**

|   |   |
|---|---|
| user_id | 0 |
| name | 0 |
| country | 1 |
| channel | 0 |
| signup_date | 0 |

> R: `colSums(is.na(df))`. Run this first whenever you get a new dataset.

### Problem 2 · Find rows with missing values

Users whose country is missing.

```python
users[users["country"].isna()]
```

**Result:**

|   | user_id | name | country | channel | signup_date |
|---|---|---|---|---|---|

### Problem 3 · Drop rows missing a given column

Drop users with a missing country. How many rows are left?

```python
len(users.dropna(subset=["country"]))
```

**Result:**

```text
9
```

> With no arguments, `dropna()` drops any row with a missing value in any column; `subset=` restricts the check to the listed columns.

### Problem 4 · Fill missing values

Fill missing countries with `"Unknown"`.

```python
u = users.copy()
u["country"] = u["country"].fillna("Unknown")
u["country"].value_counts()
```

**Result:**

| country | count |
|---|---|
| US | 4 |
| CN | 3 |
| CA | 2 |
| Unknown | 1 |

*4 rows × 1 column*

### Problem 5 · String to number

Convert `revenue` in `raw` to numbers, turning anything that cannot be converted into a missing value.

```python
pd.to_numeric(raw["revenue"], errors="coerce")
```

**Result:**

|   | revenue |
|---|---|
| 0 | 10.0 |
| 1 | 20.0 |
| 2 | NaN |
| 3 | 50.0 |
| 4 | 40.0 |
| 5 | NaN |

*6 rows × 1 column*

> `errors="coerce"`: bad values become missing instead of raising an error. Both `"n/a"` and the original None become NaN.

### Problem 6 · Missing values are skipped in statistics

The mean of `revenue` after converting it to numbers; and what is the mean if missing values count as 0?

```python
rev = pd.to_numeric(raw["revenue"], errors="coerce")
print(rev.mean())
print(rev.fillna(0).mean())
```

**Result:**

```text
30.0
20.0
```

> pandas' `mean()` skips missing values by default (as if R's `na.rm = TRUE` were always on). 120 / 4 = 30, 120 / 6 = 20.

### Problem 7 · String to date, and find bad values

Convert `date` in `raw` to dates and find the rows that fail.

```python
d = pd.to_datetime(raw["date"], errors="coerce")
raw[d.isna()]
```

**Result:**

|   | id | segment | date | revenue | active |
|---|---|---|---|---|---|
| 4 | 5 | B | bad | 40 | 1 |

*1 row × 5 columns*

### Problem 8 · Strip whitespace, normalize case

Clean up `segment` and see which distinct values remain.

```python
print(raw["segment"].unique().tolist())
print(raw["segment"].str.strip().str.upper().unique().tolist())
```

**Result:**

```text
[' A', 'a', 'B', 'b ', 'A']
['A', 'B']
```

> Column-wide string operations go through `.str`.

### Problem 9 · Clean the whole table in one go

Uppercase `segment`, convert the types of `revenue` and `date`, and drop rows with a bad date.

```python
clean = raw.copy()
clean["segment"] = clean["segment"].str.strip().str.upper()
clean["revenue"] = pd.to_numeric(clean["revenue"], errors="coerce")
clean["date"] = pd.to_datetime(clean["date"], errors="coerce")
clean = clean.dropna(subset=["date"])
clean
```

**Result:**

|   | id | segment | date | revenue | active |
|---|---|---|---|---|---|
| 0 | 1 | A | 2026-01-01 | 10.0 | 1 |
| 1 | 2 | A | 2026-01-03 | 20.0 | 1 |
| 2 | 3 | B | 2026-01-02 | NaN | 0 |
| 3 | 4 | B | 2026-01-04 | 50.0 | 1 |
| 5 | 6 | A | 2026-01-05 | NaN | 1 |

*5 rows × 5 columns*

### Problem 10 · Aggregate after cleaning

Total revenue per segment (using `clean` from the previous problem).

```python
clean.groupby("segment")["revenue"].sum().reset_index()
```

|   | segment | revenue |

### Problem 14 · Extract month and weekday

Convert the order date to a date type and extract the month (as a number) and the weekday.

```python
o = orders.copy()
o["order_date"] = pd.to_datetime(o["order_date"])
o["month"] = o["order_date"].dt.month
o["weekday"] = o["order_date"].dt.day_name()
o[["order_id", "order_date", "month", "weekday"]].head(5)
```

**Result:**

|   | order_id | order_date | month | weekday |
|---|---|---|---|---|

| 3 | 1004 | 2026-01-03 | 1 | Saturday |
| 4 | 1005 | 2026-01-04 | 1 | Sunday |

*5 rows × 4 columns*

> Use `.dt` to get the parts of a date column, like lubridate's `month()` and `wday()`.

### Problem 15 · Monthly summary

Number of orders and total amount per month (with months written like `"2026-01"`).

```python
o = orders.copy()
o["month"] = pd.to_datetime(o["order_date"]).dt.strftime("%Y-%m")
o.groupby("month").agg(n=("order_id", "count"), total=("amount", "sum")).reset_index()
```

**Result:**

|   | month | n | total |
|---|---|---|---|
| 0 | 2026-01 | 13 | 535.0 |
| 1 | 2026-02 | 3 | 185.0 |

*2 rows × 3 columns*

### Problem 16 · Days between two dates

For each order, the number of days since the user signed up.

```python
m = orders.merge(users[["user_id", "signup_date"]], on="user_id")
m["days"] = (pd.to_datetime(m["order_date"]) - pd.to_datetime(m["signup_date"])).dt.days
m[["order_id", "user_id", "signup_date", "order_date", "days"]].head(6)
```

**Result:**

|   | order_id | user_id | signup_date | order_date | days |
|---|---|---|---|---|---|
| 0 | 1001 | 1 | 2026-01-01 | 2026-01-02 | 1 |
| 1 | 1002 | 1 | 2026-01-01 | 2026-01-05 | 4 |

| 3 | 1004 | 2 | 2026-01-01 | 2026-01-03 | 2 |
| 4 | 1005 | 2 | 2026-01-01 | 2026-01-04 | 3 |
| 5 | 1006 | 3 | 2026-01-02 | 2026-01-02 | 0 |

*6 rows × 5 columns*

> Subtracting two dates gives a time delta; `.dt.days` extracts the number of days.

### Problem 17 · Add days to a date

The date 7 days after each user signed up.

```python
u = users[["user_id", "signup_date"]].copy()
u["signup_date"] = pd.to_datetime(u["signup_date"])
u["day7"] = u["signup_date"] + pd.Timedelta(days=7)
u.head(4)
```

**Result:**

|   | user_id | signup_date | day7 |
|---|---|---|---|
| 0 | 1 | 2026-01-01 | 2026-01-08 |

### Problem 18 · Binning: pd.cut

Bin order amounts into three buckets, `<20`, `20-39` and `>=40`, and count the orders in each.

```python
o = orders.copy()
o["bucket"] = pd.cut(o["amount"], bins=[0, 20, 40, np.inf], labels=["<20", "20-39", ">=40"], right=False)
o.groupby("bucket")["order_id"].count().reset_index().rename(columns={"order_id": "n"})
```

**Result:**

|   | bucket | n |
|---|---|---|
| 0 | <20 | 3 |
| 1 | 20-39 | 4 |
| 2 | >=40 | 9 |

*3 rows × 2 columns*

> `right=False` makes the intervals closed on the left and open on the right, so 20 falls in `20-39`.

### Problem 19 · Lookup-table replacement: map

Replace the order status codes with readable labels.

```python
o = orders.copy()
o["status_en"] = o["status"].map({"completed": "Completed", "refunded": "Refunded"})
o[["order_id", "status", "status_en"]].head(4)
```

**Result:**

|   | order_id | status | status_en |
|---|---|---|---|
| 0 | 1001 | completed | Completed |
| 1 | 1002 | completed | Completed |
| 2 | 1003 | refunded | Refunded |
| 3 | 1004 | completed | Completed |

*4 rows × 3 columns*

> Values not found in the dictionary become missing.

### Problem 20 · Custom transformation: apply + lambda

Format the amount as text like `"$30"`.

```python
o = orders.copy()
o["label"] = o["amount"].apply(lambda a: f"${a:.0f}")
o[["order_id", "amount", "label"]].head(4)
```

**Result:**

|   | order_id | amount | label |
|---|---|---|---|
| 0 | 1001 | 30.0 | $30 |
| 1 | 1002 | 50.0 | $50 |
| 2 | 1003 | 20.0 | $20 |
| 3 | 1004 | 15.0 | $15 |

*4 rows × 3 columns*

> `lambda a: ...` is a small throwaway function, like R's `function(a) ...`. Avoid `apply` when a vectorized column operation will do; it is slow.

## Recap

| Task | Code |
|---|---|
| Check missing values | `df.isna().sum()` |
| Drop / fill missing values | `dropna(subset=[...])` / `fillna(value)` |
| Convert types | `pd.to_numeric(s, errors="coerce")`, `pd.to_datetime(s, errors="coerce")` |
| Strings | `s.str.strip()`, `.str.upper()`, `.str.contains()`, `.str.replace()`, `.str.len()` |
| Dates | `s.dt.month`, `.dt.day_name()`, `.dt.strftime("%Y-%m")`, `(s2 - s1).dt.days`, `+ pd.Timedelta(days=7)` |
| Bin / replace | `pd.cut`, `s.map(dict)`, `s.apply(lambda x: ...)` |
