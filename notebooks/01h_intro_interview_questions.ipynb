```python
def most_common(xs):
    counts = {}
    for x in xs:
        counts[x] = counts.get(x, 0) + 1

    best = None
    best_count = 0
    for x, c in counts.items():
        if c > best_count:
            best = x
            best_count = c
    return best

print(most_common([3, 1, 3, 2, 1, 3]))
```

## Fast vs. slow: an intuition for complexity

Interviewers often ask, "What's the time complexity of this solution?" It means: **if the data grows n-fold, roughly how many times longer does it run?**

| Code | Notation | Data grows 10×, time grows |
|---|---|---|
| One pass over a list | O(n) | 10× |
| Two nested loops, looking at every pair | O(n²) | 100× |
| Sorting | O(n log n) | a bit over 10× |
| Looking up a value in a dict / set | O(1) | unchanged |

The key point: **`x in some_list` scans from start to end, so it's O(n); `x in some_set` or `x in some_dict` is O(1).**

That's why Problem 2 uses `seen = set()` instead of a list. Run the code below to feel the difference:

````python
import time

n = 20000
as_list = list(range(n))
as_set = set(as_list)

t0 = time.time()
for i in range(2000):
    (n - 1) in as_list       # scans from start to end every time
t1 = time.time()
for i in range(2000):
    (n - 1) in as_set        # direct hit
t2 = time.time()

print(f"list: {t1 - t0:.4f} s")
print(f"set : {t2 - t1:.4f} s")
````

## Problem 4: Two Sum

The most famous interview question. Given a list of integers `nums` and a target `target`, find the **indices of two numbers** whose sum equals `target`.

`two_sum([2, 7, 11, 15], 9)` → `(0, 1)`, because `nums[0] + nums[1] = 2 + 7 = 9`.

### Step 1: the most direct approach

Try every pair. Two nested loops, O(n²). The code is already written; run it:

````python
def two_sum_slow(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return i, j
    return None

print(two_sum_slow([2, 7, 11, 15], 9))
print(two_sum_slow([3, 2, 4], 6))
````

### Step 2: making it faster

Look at it from another angle: when I reach the number `x`, the other number I need is `need = target - x`. The question becomes "**has `need` appeared before, and at which position?**"

"Has it appeared, and where" is exactly what a dict is good at. Use a dict `seen` to record "number → index":

Take `nums = [2, 7, 11, 15]`, `target = 9` as an example:

| i | x | need = 9 - x | Is need in seen? | Action |
|---|---|---|---|---|
| 0 | 2 | 7 | No | Record `seen[2] = 0` |
| 1 | 7 | 2 | Yes, at index 0 | Return `(0, 1)` |

Just one pass, with an O(1) dict lookup at each step: O(n) in total.

```python
def two_sum(nums, target):
    seen = {}                      # number -> index
    for i, x in enumerate(nums):
        need = target - x
        if need in seen:
            return seen[need], i
        seen[x] = i
    return None

print(two_sum([2, 7, 11, 15], 9))
```

### Step 3: explain it in your own words

In an interview, after writing the code you also have to explain it. Try answering out loud (no need to write):

**Why is the dict solution O(n) while the nested loops are O(n²)?**

<details>
<summary><b>Sample answer</b></summary>

The nested loops check every pair of elements; there are about n²/2 pairs, so it's O(n²). The dict solution visits each element only once, and each time uses an O(1) lookup to check whether the number it needs has appeared before, so the total time is O(n). The trade-off is O(n) extra space for the dict.

</details>
