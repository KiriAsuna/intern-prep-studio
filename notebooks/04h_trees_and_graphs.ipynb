# 04h | Trees and graphs: DFS and BFS

There seem to be endless tree and graph problems, but almost all of them apply one of two traversals:

- **DFS (depth-first search)**: follow one path all the way down, then back up. Written with **recursion**;
- **BFS (breadth-first search)**: expand outward one layer at a time. Written with a **queue**; it finds the **minimum number of steps**.

## 2. DFS on trees: recursion

Trees are a natural fit for recursion because **each child is itself a tree**. The pattern:

```python
def f(node):
    if node is None:          # base case: empty tree
        return some_value
    left = f(node.left)       # answer for the left subtree
    right = f(node.right)     # answer for the right subtree
    return combine(node.val, left, right)   # build the current answer
```

Example: the maximum depth of a tree = 1 (this level) + the larger of the left and right subtree depths. An empty tree has depth 0.

**Traversal orders.** There are three ways to visit every node of a tree, depending on "when you process the node itself":

- Preorder: node → left → right;
- **Inorder: left → node → right**;
- Postorder: left → right → node.

The tree above is a **binary search tree** (everything on the left is smaller than the node, everything on the right is larger). An inorder traversal of a binary search tree yields the values in ascending order.

## 3. BFS on trees: a queue

Use BFS for "output level by level" or "shortest distance". The template:

```python
from collections import deque

q = deque([start])
while q:
    node = q.popleft()        # take out the earliest item added
    process(node)
    # add node's neighbors (children) to the queue
```

Because a queue is first-in, first-out, nodes closer to the start are always processed first, so the search moves outward layer by layer.

When you need to know "which level am I on", record the queue length `n = len(q)` at the start of each round and process exactly n nodes; those n nodes form one level.

```python
                q.append(node.right)

    return out
```

## 4. Graphs

A graph is "nodes + edges". A tree is a special kind of graph. The usual representation is an **adjacency list**: a dict whose keys are nodes and whose values are lists of neighbors.

```python
graph = {"A": ["B", "C"], "B": ["A", "D"], ...}
```

Traversing a graph differs from traversing a tree in just one way: **a graph may contain cycles, so you must track which nodes you have already visited**, or you will go around a cycle forever. Use a `visited` set.

BFS for the minimum number of steps: store both "the node" and "how many steps it took to get here" in the queue.

````python
from collections import deque
````

```python
    visited = {start}
    q = deque([(start, 0)])               # (node, steps taken so far)
    while q:
        node, steps = q.popleft()
        if node == goal:
            return steps
        for nb in graph[node]:
            if nb not in visited:
                visited.add(nb)
                q.append((nb, steps + 1))
    return -1
```

## 5. Grids are graphs too

2D grid problems (islands, mazes, flood fill) are very common in OAs. Treat each cell as a node; the cells directly above, below, left, and right are its neighbors.

**Number of islands:** in the grid, 1 is land and 0 is water. Land cells connected up/down/left/right form one island. How many islands are there?

Approach: scan every cell. When you hit land you **haven't visited yet**, add 1 to the island count, then run DFS from it and "sink" all land connected to it (set it to 0) so it never gets counted again.

The idiomatic way to loop over the four directions:

```python
for dr, dc in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
    nr, nc = r + dr, c + dc
```

````python
def num_islands(grid):
    if not grid:
        return 0
    rows, cols = len(grid), len(grid[0])

    def sink(r, c):
        # set (r, c) and all land connected to it to 0
        if r < 0 or r >= rows or c < 0 or c >= cols or grid[r][c] == 0:
            return
        grid[r][c] = 0
        for dr, dc in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
            sink(r + dr, c + dc)

    count = 0
    # write your code below: scan every cell; when you find a 1, count it and call sink

    return count
````

```python
def num_islands(grid):
    if not grid:
        return 0
    rows, cols = len(grid), len(grid[0])

    def sink(r, c):
        # set (r, c) and all land connected to it to 0
        if r < 0 or r >= rows or c < 0 or c >= cols or grid[r][c] == 0:
            return
        grid[r][c] = 0
        for dr, dc in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
            sink(r + dr, c + dc)

    count = 0
    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == 1:
                count += 1
                sink(r, c)
    return count
```

`sink` is defined inside `num_islands`, so it can use the outer `grid`, `rows`, and `cols` directly. Defining a function inside a function is common in problems like this.

## Summary

- DFS on trees: recursion, with the empty node as the base case;
- BFS: `deque`, take items out with `popleft`, `append` the neighbors; for level-by-level processing, read `len(q)` first;
- Graphs need `visited` to avoid going in circles;
- Grid = graph; loop over the four directions with `(dr, dc)`;
- **Use BFS for minimum number of steps**.
