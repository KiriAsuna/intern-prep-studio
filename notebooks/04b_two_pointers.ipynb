# 04b | Two Pointers

A "pointer" here is just an index. The two-pointer technique moves two indices over the same array (or over two arrays), reducing two nested loops to one.

Two common forms:

- **Opposite ends**: one starts at the left, one at the right, and they move toward the middle. Used for palindromes and finding a pair in a sorted array;
- **Same direction**: one fast, one slow, both moving left to right. Used for in-place removal and merging two sorted arrays.

About 50 minutes.

## 1. Opposite-end pointers

Reverse a list in place: keep a left and a right index, swap, and step each one toward the middle until they meet.

`a, b = b, a` is how Python swaps two variables.

````python
def reverse_in_place(xs):
    left, right = 0, len(xs) - 1
    while left < right:
        xs[left], xs[right] = xs[right], xs[left]
        left += 1
        right -= 1

data = [1, 2, 3, 4, 5]
reverse_in_place(data)
print(data)
````

### Finding a pair in a sorted array

Two Sum in 01h used a dict. If the array is **already sorted**, you can do it without extra space:

- Put one pointer at each end and look at the sum of the two numbers;
- Sum too small → move the left pointer right (makes the sum larger);
- Sum too large → move the right pointer left (makes the sum smaller).

Each step rules out one element, so it's O(n). **Whenever you see the word "sorted," think two pointers or binary search.**

### One pointer per array

Merging two sorted arrays: at each step compare the elements the two pointers point to, put the smaller one into the result, and advance that pointer. This is the core step of merge sort and is often asked on its own.

The next two problems use two small tools that haven't been formally introduced yet:

- `out.extend(other_list)`: appends the elements of another list **one by one** to the end (`append` adds the whole thing as a single element);
- `abs(x)`: absolute value, same as in R.

````python
out = [1, 2]
out.append([3, 4])
print(out)          # [1, 2, [3, 4]]  the whole list became one element
out = [1, 2]
out.extend([3, 4])
print(out)          # [1, 2, 3, 4]
print(abs(-7), abs(3))
````

```python
def merge_sorted(a, b):
    i, j = 0, 0
    out = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            out.append(a[i])
            i += 1
        else:
            out.append(b[j])
            j += 1
    out.extend(a[i:])
    out.extend(b[j:])
    return out
```

```python
    out = [0] * n
    left, right = 0, n - 1
    for pos in range(n - 1, -1, -1):
        if abs(nums[left]) > abs(nums[right]):
            out[pos] = nums[left] ** 2
            left += 1
```
