# 01b | Lists

**After this lesson you can:** create a collection of data, pull out any slice of it, append and modify elements, and compute the sum and mean.

A list is the counterpart of R's vector `c(...)` and is the most commonly used container in Python. About 35 minutes.

## 1. Creating a list

Use square brackets with commas between items. `len()` gives the length.

````python
xs = [10, 20, 30, 40, 50]
print(xs)
print(len(xs))
````

## 2. Accessing elements: indexing starts at 0 ⚠️

This is the first habit R users need to change.

| You want | R | Python |
|---|---|---|
| The first | `xs[1]` | `xs[0]` |
| The second | `xs[2]` | `xs[1]` |
| The last | `xs[length(xs)]` | `xs[-1]` |

A negative index in R means "drop this element"; in Python it means **"count from the end."**

````python
xs = [10, 20, 30, 40, 50]
print(xs[0])    # first: 10
print(xs[1])    # second: 20
print(xs[-1])   # last: 50
print(xs[-2])   # second to last: 40
````

## 3. Slicing: the right end is excluded ⚠️

`xs[a:b]` takes indices `a` through `b-1`, **not including `b`**.

```
element  10   20   30   40   50
index     0    1    2    3    4
```

- `xs[1:3]` → indices 1, 2 → `[20, 30]` (in R you'd write `xs[2:3]`)
- `xs[:2]` → the first two
- `xs[2:]` → from index 2 to the end

An easy rule to remember: the length of `xs[a:b]` is exactly `b - a`.

````python
xs = [10, 20, 30, 40, 50]
print(xs[1:3])
print(xs[:2])
print(xs[2:])
````

## 4. Modifying and appending

Syntax like `xs.append(60)` (variable + dot + name + parentheses) is called a **method**; it means "ask `xs` to do something itself." R has almost nothing like this, but in Python it's everywhere.

⚠️ `append` **modifies `xs` in place**. You don't need to, and must not, write `xs = xs.append(60)` (that would make `xs` become `None`).

````python
xs = [10, 20, 30]
xs[0] = 99       # change the first element
xs.append(40)    # add one to the end
print(xs)
````

## 5. Common operations

| You want | R | Python |
|---|---|---|
| Sum | `sum(x)` | `sum(xs)` |
| Min / max | `min(x)` / `max(x)` | `min(xs)` / `max(xs)` |
| Mean | `mean(x)` | `sum(xs) / len(xs)` |
| Sort | `sort(x)` | `sorted(xs)` |
| Descending | `sort(x, decreasing = TRUE)` | `sorted(xs, reverse=True)` |
| Membership | `3 %in% x` | `3 in xs` |

Base Python has no `mean()`; you divide yourself.

````python
xs = [3, 1, 4, 1, 5]
print(sum(xs), min(xs), max(xs))
print(sum(xs) / len(xs))
print(sorted(xs))
print(sorted(xs, reverse=True))
print(4 in xs, 9 in xs)
````

## 6. Lists are not vectorized ⚠️

In R, `c(1, 2, 3) * 2` gives `2 4 6`. Python lists **don't** work that way:

````python
print([1, 2, 3] * 2)       # repeats the list twice; does not multiply each by 2
print([1, 2] + [3, 4])     # concatenates; does not add element-wise
````

To operate on every element, either use a loop (next lesson) or NumPy (Chapter 02, the true counterpart of R vectors).

## 7. Strings can be sliced too

A string can be thought of as a "list of characters"; indexing and slicing rules are exactly the same.

````python
s = "statistics"
print(s[0])
print(s[:4])
print(s[-3:])
````

3. Store the average score in `avg`;
4. Store the two highest scores in `top2` (a list, highest first).

<details>

## Summary

- Indexing starts at 0; `xs[-1]` is the last element;
- The slice `xs[a:b]` excludes `b`;
- `xs.append(x)` modifies in place; don't reassign it;
- Lists don't do vectorized arithmetic.
