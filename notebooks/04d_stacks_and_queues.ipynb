# 04d | Stacks and Queues

- **Stack**: last in, first out, like a pile of plates; you can only add and remove from the top.
- **Queue**: first in, first out, like a line of people.

## 1. A list as a stack, a deque as a queue

Use a plain list as a stack: `append` pushes onto the top, `pop()` removes from the top, and `stack[-1]` peeks at the top.

**Don't use a list** as a queue: `list.pop(0)` shifts every remaining element forward each time, which is O(n). Use `collections.deque`, whose `popleft()` is O(1).

````python
from collections import deque

stack = []
stack.append(1)
stack.append(2)
stack.append(3)
print(stack.pop(), stack.pop(), stack)     # 3 2 [1]

q = deque()
q.append("a")
q.append("b")
q.append("c")
print(q.popleft(), q.popleft(), q)         # a b deque(['c'])
````

## 2. Matching brackets

Check whether brackets are properly closed: for example, `"{[]}"` is valid and `"([)]"` is not.

The rule: **the most recently opened bracket must be closed first**, which is exactly "last in, first out."

- Opening bracket: push it onto the stack;
- Closing bracket: the top of the stack must be its matching opening bracket; if so, pop it, otherwise the string is invalid;
- At the end, the stack must be empty (otherwise some opening bracket was never closed).

## 3. Monotonic stack: finding the "next greater" element

Problem: for each number in an array, find **the first number to its right that is larger**; if there is none, use -1.

`[2, 1, 5, 3, 6]` → `[5, 5, 6, 6, -1]`

The direct approach is two nested loops, O(n²). A monotonic stack does it in O(n):

The stack holds **the indices of elements whose answer hasn't been found yet**. Scan left to right; for each new number:

- As long as the number at the top-of-stack index is smaller, the new number is that element's answer → pop and record it;
- Then push its own index and wait for a later number to answer it.

| New number | Action | Stack (showing values for clarity) |
|---|---|---|
| 2 | Push | 2 |
| 1 | 1 is not larger than 2; push | 2, 1 |
| 5 | Larger than 1 → 1's answer is 5; larger than 2 → 2's answer is 5; push 5 | 5 |
| 3 | Push | 5, 3 |
| 6 | 3's answer is 6; 5's answer is 6; push 6 | 6 |

Whatever is left on the stack (6) has no answer and stays -1. Each element is pushed and popped only once, so it's O(n).

    stack = []                  # stores indices: positions still waiting for an answer

        while stack and nums[stack[-1]] < x:
            j = stack.pop()
            out[j] = x
        stack.append(i)

            j = stack.pop()

        stack.append(i)

## Summary

- Stack: a list (`append` / `pop`); queue: a `deque` (`append` / `popleft`);
- "The most recent unmatched item" → stack;
- "The first larger / smaller element to the right / left" → monotonic stack.
