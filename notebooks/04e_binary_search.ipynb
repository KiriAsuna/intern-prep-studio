# 04e | Binary Search

Searching **sorted** data: look at the middle element each time and rule out half. A million numbers take at most 20 looks.

Everyone understands the idea of binary search; the hard part is the boundaries: `<` or `<=`, `mid` or `mid + 1`. Get one wrong and you have an infinite loop. This section gives you two fixed templates: **memorize them; don't derive them on the spot**.

## Template 1: find an exact value

```python
lo, hi = 0, len(a) - 1          # the search range is the closed interval [lo, hi]
while lo <= hi:                 # the range still has elements
    mid = (lo + hi) // 2
    if a[mid] == target:
        return mid
    if a[mid] < target:
        lo = mid + 1            # target is in the right half; mid is ruled out
    else:
        hi = mid - 1            # target is in the left half; mid is ruled out
return -1
```

Three key points: `hi` starts at `len - 1`; the loop condition is `<=`; and `mid` is excluded every time (`+ 1` / `- 1`).

## Template 2: find a boundary (the first position that satisfies a condition)

The more common question is: "Where is the first position with a value greater than or equal to target?" The array may contain duplicates, and target may not be present.

```python
lo, hi = 0, len(a)              # note hi is len(a): the answer may be "past the end"
while lo < hi:                  # note: <
    mid = (lo + hi) // 2
    if a[mid] < target:
        lo = mid + 1            # mid fails the condition; the answer is to the right
    else:
        hi = mid                # mid satisfies the condition and may be the answer; keep it
return lo
```

When the loop ends, `lo == hi`, and that's the answer.

This position is also called the "insertion point": inserting target there keeps the array sorted. The standard library's `bisect.bisect_left(a, target)` does exactly this; `bisect.bisect_right` returns the first position **greater than** target.

````python
import bisect

a = [1, 2, 2, 2, 5]
print(bisect.bisect_left(a, 2))    # 1: the first position >= 2
print(bisect.bisect_right(a, 2))   # 4: the first position > 2
print(bisect.bisect_left(a, 3))    # 4: 3 would be inserted here
print(bisect.bisect_left(a, 9))    # 5: larger than everything, so it goes at the end
````

## Binary search on the answer

Binary search isn't limited to arrays. Whenever the problem is "find the largest / smallest number that satisfies a condition" and the condition is **monotonic** (all smaller values satisfy it and all larger ones don't, or vice versa), you can binary search over the range of possible answers.

```python
def int_sqrt(x):
    lo, hi = 0, x
    while lo <= hi:
        mid = (lo + hi) // 2
        if mid * mid <= x:
            lo = mid + 1        # mid works; try something larger
        else:
            hi = mid - 1        # mid is too large
    return hi
```

The hidden test cases include numbers as large as 10¹²; trying values one by one from 1 would time out, which is why you need binary search.

## Summary

- Exact value: `hi = len - 1`, `while lo <= hi`, `mid ± 1`;
- Boundary: `hi = len`, `while lo < hi`, `hi = mid`; or just use `bisect_left` / `bisect_right`;
- "Find the largest / smallest value satisfying a condition" + a monotonic condition → binary search on the answer.
