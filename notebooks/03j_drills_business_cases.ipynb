### Problem 1 · Monthly sales

Total completed order amount per month.

```sql
SELECT strftime('%Y-%m', order_date) AS month,
       SUM(amount)                   AS total
FROM orders
WHERE status = 'completed'
GROUP BY month
ORDER BY month;
```

**Result:**

|   | month | total |
|---|---|---|
| 1 | 2026-01 | 480.0 |
| 2 | 2026-02 | 185.0 |

*2 rows × 2 columns*

> Year-month: MySQL `DATE_FORMAT(d, '%Y-%m')`, PostgreSQL `TO_CHAR(d, 'YYYY-MM')`.

### Problem 3 · Daily and cumulative signups

New users per signup date, and the cumulative number of users up to that day.

```sql
WITH daily AS (
  SELECT signup_date, COUNT(*) AS new_users
  FROM users
  GROUP BY signup_date
)
SELECT signup_date, new_users,
       SUM(new_users) OVER (ORDER BY signup_date) AS cum_users
FROM daily
ORDER BY signup_date;
```

**Result:**

|   | signup_date | new_users | cum_users |
|---|---|---|---|
| 1 | 2026-01-01 | 2 | 2 |
| 2 | 2026-01-02 | 1 | 3 |
| 3 | 2026-01-03 | 2 | 5 |
| 4 | 2026-01-05 | 1 | 6 |
| 5 | 2026-01-06 | 1 | 7 |
| 6 | 2026-01-08 | 1 | 8 |
| 7 | 2026-01-09 | 1 | 9 |
| 8 | 2026-01-10 | 1 | 10 |

*8 rows × 3 columns*

> Aggregate by day first, then compute the running total on the aggregated table.

### Problem 4 · ARPU by channel (non-spenders count as 0)

Average completed spend per user in each channel, counting users with no orders as 0.

```sql
WITH spend AS (
  SELECT user_id, SUM(amount) AS total
  FROM orders
  WHERE status = 'completed'
  GROUP BY user_id
)
SELECT u.channel,
       COUNT(*)                     AS n_users,
       AVG(COALESCE(s.total, 0))    AS arpu
FROM users u
LEFT JOIN spend s
  ON u.user_id = s.user_id
GROUP BY u.channel;
```

**Result:**

|   | channel | n_users | arpu |
|---|---|---|---|
| 1 | ads | 4 | 60.0 |
| 2 | organic | 4 | 65.0 |
| 3 | referral | 2 | 82.5 |

*3 rows × 3 columns*

> With `AVG(s.total)`, NULLs are skipped, so the denominator only includes users who ordered and ARPU is inflated. It's the same pitfall as forgetting `fillna(0)` in 02k.

### Problem 5 · Day-1 retention rate

The share of users who also logged in on the day after their first login.

```sql
WITH logins AS (
  SELECT DISTINCT user_id, event_date
  FROM events
  WHERE event_type = 'login'
),
first AS (
  SELECT user_id, MIN(event_date) AS first_date
  FROM logins
  GROUP BY user_id
)
SELECT COUNT(*)                          AS n_users,
       COUNT(l.user_id)                  AS n_retained,
       1.0 * COUNT(l.user_id) / COUNT(*) AS retention
FROM first f
LEFT JOIN logins l
  ON l.user_id = f.user_id
 AND l.event_date = date(f.first_date, '+1 day');
```

**Result:**

|   | n_users | n_retained | retention |
|---|---|---|---|
| 1 | 10 | 7 | 0.7 |

*1 row × 3 columns*

> Three steps: deduplicate logins → first day → left join on "first day + 1." Adding 1 day: MySQL `DATE_ADD(d, INTERVAL 1 DAY)`, PostgreSQL `d + INTERVAL '1 day'`.

### Problem 6 · Retention by first-day cohort

For each first-login date, the number of new users and how many of them were retained on day 1.

```sql
WITH logins AS (
  SELECT DISTINCT user_id, event_date
  FROM events
  WHERE event_type = 'login'
),
first AS (
  SELECT user_id, MIN(event_date) AS first_date
  FROM logins
  GROUP BY user_id
)
SELECT f.first_date,
       COUNT(*)         AS n_users,
       COUNT(l.user_id) AS n_retained
FROM first f
LEFT JOIN logins l
  ON l.user_id = f.user_id
 AND l.event_date = date(f.first_date, '+1 day')
GROUP BY f.first_date
ORDER BY f.first_date;
```

|---|---|---|---|
| 1 | 2026-01-01 | 2 | 2 |
| 2 | 2026-01-02 | 1 | 1 |
| 3 | 2026-01-03 | 2 | 1 |
| 4 | 2026-01-05 | 1 | 1 |
| 5 | 2026-01-06 | 1 | 0 |
| 6 | 2026-01-08 | 1 | 1 |

| 8 | 2026-01-10 | 1 | 0 |

*8 rows × 3 columns*

### Problem 7 · 7-day return rate

The share of users who logged in again within 7 days after their first login (days 1 through 7).

```sql
WITH logins AS (
  SELECT DISTINCT user_id, event_date
  FROM events
  WHERE event_type = 'login'
),
first AS (
  SELECT user_id, MIN(event_date) AS first_date
  FROM logins
  GROUP BY user_id
)
SELECT COUNT(DISTINCT f.user_id) AS n_users,
       COUNT(DISTINCT l.user_id) AS n_returned,
       1.0 * COUNT(DISTINCT l.user_id) / COUNT(DISTINCT f.user_id) AS return_rate
FROM first f
LEFT JOIN logins l
  ON l.user_id = f.user_id
 AND l.event_date BETWEEN date(f.first_date, '+1 day') AND date(f.first_date, '+7 day');
```

**Result:**

|   | n_users | n_returned | return_rate |
|---|---|---|---|
| 1 | 10 | 7 | 0.7 |

*1 row × 3 columns*

> The difference from day-1 retention: over 7 days a user can match several days, so after the left join it's no longer one row per user, and both numerator and denominator need `COUNT(DISTINCT ...)`. In this data the answer happens to be 7 users, the same as day-1 retention; on larger data the two usually differ.

### Problem 8 · Users at each funnel step

The number of distinct users at each of the three steps: view, add to cart, purchase.

```sql
SELECT COUNT(DISTINCT CASE WHEN event_type = 'view'     THEN user_id END) AS n_view,
       COUNT(DISTINCT CASE WHEN event_type = 'cart'     THEN user_id END) AS n_cart,
       COUNT(DISTINCT CASE WHEN event_type = 'purchase' THEN user_id END) AS n_purchase
FROM events;
```

**Result:**

|   | n_view | n_cart | n_purchase |
|---|---|---|---|
| 1 | 10 | 7 | 5 |

*1 row × 3 columns*

> With no `ELSE` in the `CASE`, non-matching rows are NULL, and `COUNT` doesn't count NULLs.

### Problem 9 · Funnel conversion rates

Building on the previous problem, compute the conversion rate of each of the two steps.

```sql
WITH f AS (
  SELECT COUNT(DISTINCT CASE WHEN event_type = 'view'     THEN user_id END) AS n_view,
         COUNT(DISTINCT CASE WHEN event_type = 'cart'     THEN user_id END) AS n_cart,
         COUNT(DISTINCT CASE WHEN event_type = 'purchase' THEN user_id END) AS n_purchase
  FROM events
)
SELECT n_view, n_cart, n_purchase,
       1.0 * n_cart / n_view     AS view_to_cart,
       1.0 * n_purchase / n_cart AS cart_to_purchase
FROM f;
```

**Result:**

|   | n_view | n_cart | n_purchase | view_to_cart | cart_to_purchase |
|---|---|---|---|---|---|
| 1 | 10 | 7 | 5 | 0.7 | 0.7143 |

*1 row × 5 columns*

> Once you have the numbers, say which step loses the most users without being asked. Here view → cart loses 30%, and cart → purchase loses about 29%.

### Problem 10 · Funnel by channel

The number of users who viewed, added to cart, and purchased, per channel.

```sql
SELECT u.channel,
       COUNT(DISTINCT CASE WHEN e.event_type = 'view'     THEN e.user_id END) AS n_view,
       COUNT(DISTINCT CASE WHEN e.event_type = 'cart'     THEN e.user_id END) AS n_cart,
       COUNT(DISTINCT CASE WHEN e.event_type = 'purchase' THEN e.user_id END) AS n_purchase
FROM events e
JOIN users u
  ON e.user_id = u.user_id
GROUP BY u.channel;
```

**Result:**

|   | channel | n_view | n_cart | n_purchase |
|---|---|---|---|---|
| 1 | ads | 4 | 3 | 2 |
| 2 | organic | 4 | 3 | 2 |
| 3 | referral | 2 | 1 | 1 |

*3 rows × 4 columns*

### Problem 11 · Users with at least 3 consecutive login days

Users who have ever logged in on at least 3 consecutive days.

```sql
WITH logins AS (
  SELECT DISTINCT user_id, event_date
  FROM events
  WHERE event_type = 'login'
),
numbered AS (
  SELECT user_id, event_date,
         ROW_NUMBER() OVER (PARTITION BY user_id ORDER BY event_date) AS rn
  FROM logins
),
grouped AS (
  SELECT user_id, date(event_date, '-' || rn || ' day') AS grp
  FROM numbered
)
SELECT DISTINCT user_id
FROM grouped
GROUP BY user_id, grp
HAVING COUNT(*) >= 3;
```

**Result:**

|   | user_id |
|---|---|
| 1 | 1 |
| 2 | 3 |
| 3 | 6 |

*3 rows × 1 column*

> Subtracting the within-group row number from the date gives consecutive days the same `grp`. Subtracting rn days: MySQL `DATE_SUB(d, INTERVAL rn DAY)`, PostgreSQL `d - rn`.

### Problem 12 · Each user's longest login streak

The maximum number of consecutive days each user logged in.

```sql
WITH logins AS (
  SELECT DISTINCT user_id, event_date
  FROM events
  WHERE event_type = 'login'
),
numbered AS (
  SELECT user_id, event_date,
         ROW_NUMBER() OVER (PARTITION BY user_id ORDER BY event_date) AS rn
  FROM logins
),
runs AS (
  SELECT user_id, date(event_date, '-' || rn || ' day') AS grp, COUNT(*) AS len
  FROM numbered
  GROUP BY user_id, grp
)
SELECT user_id, MAX(len) AS max_streak
FROM runs
GROUP BY user_id;
```

|   | user_id | max_streak |
|---|---|---|
| 1 | 1 | 3 |
| 2 | 2 | 2 |
| 3 | 3 | 4 |
| 4 | 4 | 1 |
| 5 | 5 | 2 |
| 6 | 6 | 3 |
| 7 | 7 | 1 |
| 8 | 8 | 2 |
| 9 | 9 | 2 |
| 10 | 10 | 1 |

*10 rows × 2 columns*

> The previous problem filters for streaks of length ≥ 3; this one takes each user's longest streak.

### Problem 13 · Days from signup to first order

For each channel, the average number of days from signup to first order (only users who ordered).

```sql
WITH first AS (
  SELECT user_id, MIN(order_date) AS first_order
  FROM orders
  GROUP BY user_id
)
SELECT u.channel,
       AVG(julianday(f.first_order) - julianday(u.signup_date)) AS avg_days
FROM first f
JOIN users u
  ON f.user_id = u.user_id
GROUP BY u.channel;
```

**Result:**

|   | channel | avg_days |
|---|---|---|
| 1 | ads | 1.5 |
| 2 | organic | 0.5 |
| 3 | referral | 4.0 |

*3 rows × 2 columns*

> Day difference: MySQL `DATEDIFF(b, a)`, PostgreSQL `b - a` (when both are of type date).

### Problem 14 · Repeat purchase rate

Among users with a completed order, the share with at least 2 completed orders.

```sql
WITH cnt AS (
  SELECT user_id, COUNT(*) AS n
  FROM orders
  WHERE status = 'completed'
  GROUP BY user_id
)
SELECT COUNT(*)                                    AS n_buyers,
       SUM(CASE WHEN n >= 2 THEN 1 ELSE 0 END)     AS n_repeat,
       AVG(CASE WHEN n >= 2 THEN 1.0 ELSE 0 END)   AS repeat_rate
FROM cnt;
```

**Result:**

|   | n_buyers | n_repeat | repeat_rate |
|---|---|---|---|
| 1 | 8 | 5 | 0.625 |

*1 row × 3 columns*

> Aggregate to one row per user first, then use conditional aggregation. The denominator is "users who bought," not all users; if the question doesn't say, ask.

### Problem 15 · First-order vs. repeat-order revenue

Across all completed orders, the total amount from first orders and from non-first orders.

```sql
WITH ranked AS (
  SELECT amount,
         ROW_NUMBER() OVER (PARTITION BY user_id ORDER BY order_date) AS rn
  FROM orders
  WHERE status = 'completed'
)
SELECT CASE WHEN rn = 1 THEN 'first' ELSE 'repeat' END AS order_type,
       SUM(amount) AS total
FROM ranked
GROUP BY order_type;
```

**Result:**

|   | order_type | total |
|---|---|---|
| 1 | first | 375.0 |
| 2 | repeat | 290.0 |

*2 rows × 2 columns*

> "First order" is defined among completed orders, so `WHERE` is applied before `ROW_NUMBER`. A different order of operations gives a different result, so clarify the definition first.

### Problem 16 · Users who ordered within 7 days of signup

| 5 | 5 | Eli |

### Problem 17 · Month-over-month growth

Completed sales per month and the growth rate versus the previous month (rounded to 3 decimals).

```sql
WITH monthly AS (
  SELECT strftime('%Y-%m', order_date) AS month, SUM(amount) AS total
  FROM orders
  WHERE status = 'completed'
  GROUP BY month
)
SELECT month, total,
       ROUND(1.0 * total / LAG(total) OVER (ORDER BY month) - 1, 3) AS growth
FROM monthly;
```

### Problem 18 · Top-spending user in each country

The user with the highest completed spend in each country (keep ties).

```sql
WITH spend AS (
  SELECT u.country, u.name, SUM(o.amount) AS total
  FROM users u
  JOIN orders o
    ON u.user_id = o.user_id
  WHERE o.status = 'completed'
  GROUP BY u.user_id, u.country, u.name
),
ranked AS (
  SELECT country, name, total,
         RANK() OVER (PARTITION BY country ORDER BY total DESC) AS rk
  FROM spend
)
SELECT country, name, total
FROM ranked
WHERE rk = 1;
```

**Result:**

|   | country | name | total |
|---|---|---|---|
| 1 | CA | Eli | 100.0 |
| 2 | CN | Chen | 180.0 |
| 3 | US | Hana | 140.0 |

*3 rows × 3 columns*

> Aggregate to one row per user first, then rank within groups. The two steps can't be merged into one.

### Problem 20 · Last login and churn flag

Each user's last login date, the number of days until 2026-01-15, and a churn flag (1) if it's more than 7 days.

## Syntax recap

| Goal | Syntax |
|---|---|
| By month | `strftime('%Y-%m', d)`, then `GROUP BY` |
| DAU | `COUNT(DISTINCT user_id)`, grouped by day |
| Retention | Deduplicate logins → first day → left join on "first day + N"; when multiple days can match, use `COUNT(DISTINCT)` for both numerator and denominator |
| Funnel | `COUNT(DISTINCT CASE WHEN step = ... THEN user_id END)` |
| Consecutive logins | Group by `date(d, '-' \|\| rn \|\| ' day')` |
| ARPU | Left join starting from `users`, `AVG(COALESCE(x, 0))` |
| First order / repeat | `ROW_NUMBER() ... = 1`; aggregate to one row per user before computing shares |
| Day difference | `julianday(b) - julianday(a)` |

## A fixed recipe for business case questions

1. **What is one row of the result?** (a user / a day / a channel)
2. **Which table does the denominator come from?** If every user must appear, start from `users` and left join.
3. **One CTE per step**: deduplicate → aggregate to one row per unit → join → compute the metric.
4. **Call out edge cases**: duplicate records, NULLs, ties, integer division.
