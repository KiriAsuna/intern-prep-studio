## 1. What makes a good estimator

$$\mathrm{MSE}(\hat\theta) = \mathrm{Var}(\hat\theta) + \mathrm{Bias}(\hat\theta)^2$$

- **Unbiased:** $E[\hat\theta] = \theta$;
- **Consistent:** $\hat\theta$ converges to $\theta$ as the sample size goes to infinity;
- **Efficient:** smallest variance within some class of estimators.

**🗣 Say it out loud:** Is an unbiased estimator always better than a biased one?

<details>
<summary><b>Reference answer</b></summary>

Not necessarily. Estimator quality is usually judged by MSE = variance + bias². An estimator with a little bias but much smaller variance can have a smaller MSE. Regularization (ridge, lasso), shrinkage estimators and Bayesian priors all deliberately introduce bias in exchange for lower variance.

**Short version:** No. What matters is usually mean squared error, which is variance plus squared bias. A slightly biased estimator with much lower variance can win, which is exactly what regularization does.

</details>

## 2. Why the sample variance divides by n − 1

When the sample mean $\bar x$ replaces the unknown population mean $\mu$, the sum of squared deviations from $\bar x$ is always **too small** ($\bar x$ is exactly the point that minimizes that sum of squares). Dividing by n − 1 corrects this bias and gives an unbiased estimator.

**🗣 Say it out loud:** Explain n − 1 intuitively in terms of "degrees of freedom".

<details>
<summary><b>Reference answer</b></summary>

The n deviations $x_i - \bar x$ must sum to 0. Once the first n − 1 are fixed, the last one is determined, so only n − 1 quantities are free to vary. Estimating the mean "uses up" one degree of freedom.

**Short version:** The n deviations from the sample mean must sum to zero, so only n minus one of them are free. Estimating the mean uses up one degree of freedom, and dividing by n minus one makes the variance estimator unbiased.

</details>

## 3. Standard deviation (SD) vs. standard error (SE)

- **SD:** the spread of the data itself;
- **SE:** the standard deviation of an **estimator's** sampling distribution; it measures the uncertainty of the estimate.

$$\mathrm{SE}(\bar X) = \frac{\sigma}{\sqrt n}$$

**🗣 Say it out loud:** As the sample size grows, what happens to the SD and to the SE of the mean?

<details>
<summary><b>Reference answer</b></summary>

The SD describes the spread of the population; with more data it is just estimated more precisely and does not systematically shrink. The SE describes the uncertainty of the estimated mean and shrinks like $1/\sqrt n$: quadruple the sample size and the SE halves, and so does the width of the confidence interval.

**Short version:** The SD describes spread in the data and doesn't shrink with more data. The standard error describes uncertainty in the estimate and shrinks like one over root n, so quadrupling the sample halves it.

</details>

## 4. Confidence intervals

The correct interpretation of a frequentist 95% CI: **if we sampled repeatedly and built an interval the same way each time, in the long run 95% of those intervals would contain the true parameter.**

The interval is random, not the parameter.

**🗣 Say it out loud:** What is wrong with "there is a 95% probability that the parameter lies in this interval"?

<details>
<summary><b>Reference answer</b></summary>

In the frequentist framework the parameter is a fixed unknown constant with no probability distribution. Once the data are observed, this particular interval is fixed too: it either contains the parameter or it doesn't. The 95% describes the coverage of the **procedure that builds the interval** under repeated sampling, not a probability for one specific interval. The statement in the question is really the interpretation of a Bayesian credible interval.

You don't need to be pedantic in an interview, but you should be able to say "95% refers to the long-run coverage of the method".

**Short version:** In the frequentist view the parameter is fixed and the interval is random. 95% is the long-run coverage of the procedure, not the probability that this particular interval contains the parameter. That probability statement belongs to a Bayesian credible interval.

</details>

## 5. Hypothesis testing

| | H₀ true | H₀ false |
|---|---|---|
| Reject H₀ | Type I error (α, false positive) | Correct (power = 1 − β) |
| Fail to reject H₀ | Correct | Type II error (β, false negative) |

**🗣 Say it out loud:** What is a p-value? What is it **not**?

<details>
<summary><b>Reference answer</b></summary>

**It is:** the probability, assuming H₀ and the model assumptions hold, of observing a test statistic as extreme as or more extreme than the one we got.

**It is not:** (1) the probability that H₀ is true, $P(H_0 \mid \text{data})$; (2) "the probability that the result is due to chance"; (3) a measure of the size or importance of an effect; (4) p > 0.05 does not mean "no effect", only insufficient evidence.

**Short version:** It's the probability, assuming the null is true, of seeing data at least as extreme as what we observed. It is not the probability that the null is true, and it says nothing about the size or importance of the effect.

</details>

**🗣 Say it out loud:** What determines statistical power?

<details>
<summary><b>Reference answer</b></summary>

(1) Larger effect size → higher power; (2) larger sample size → higher power; (3) less noise (variance) in the data → higher power; (4) larger significance level α → higher power (at the cost of more Type I errors). Also, a one-sided test has more power than a two-sided one, and variance-reduction methods (paired designs, covariate adjustment, CUPED) are equivalent to increasing the sample size.

**Short version:** Effect size, sample size, noise level, and the significance level. Bigger effects, more data, less variance, or a looser alpha all raise power. Variance reduction is effectively free sample size.

</details>

## 6. Where t, χ² and F come from

Let $X_i \sim N(\mu, \sigma^2)$ be independent:

- $\dfrac{\bar X - \mu}{\sigma/\sqrt n} \sim N(0, 1)$;
- $\dfrac{(n-1)S^2}{\sigma^2} \sim \chi^2_{n-1}$, independent of $\bar X$;
- standard normal ÷ $\sqrt{\chi^2/\mathrm{df}}$ → **t distribution** (appears when S replaces the unknown σ);
- ratio of two independent $\chi^2/\mathrm{df}$ → **F distribution** (comparing two variances, the overall test in regression).

**🗣 Say it out loud:** When do you use a t-test and when a z-test?

<details>
<summary><b>Reference answer</b></summary>

Use t when the population variance is unknown and estimated by the sample variance; with many degrees of freedom t is close to the standard normal and the two are practically the same (z is perfectly fine in an A/B test with thousands of users). Tests of proportions usually use z (the variance is determined by the proportion itself). With small samples the t-test also needs roughly normal data; with unequal variances in the two groups use Welch's t-test, which is also the default choice in practice.

**Short version:** Use t when the variance is estimated from the sample, which matters for small n; with large samples t and z coincide. For proportions we use z. With unequal variances, Welch's t-test is the safe default.

</details>

## 7. Bootstrap and permutation tests

**🗣 Say it out loud:** What is the essential difference between the bootstrap and a permutation test?

<details>
<summary><b>Reference answer</b></summary>

**Different purpose, different resampling.**

Bootstrap: resample from the sample **with replacement**, simulating "how would the estimator vary if we drew a new sample from the population"; used to get standard errors and confidence intervals.

Permutation test: **shuffle the group labels** (without replacement), simulating "what would the statistic's distribution be if H₀ were true and labels were unrelated to outcomes"; used to get a p-value.

One answers "how uncertain is the estimate", the other "how extreme is this result under the null hypothesis".

**Short version:** Bootstrap resamples with replacement to approximate the sampling distribution of an estimator, giving standard errors and confidence intervals. A permutation test shuffles labels to build the null distribution, giving a p-value.

</details>

## 8. Multiple testing

**🗣 Say it out loud:** You run 20 independent tests, none of which has a real effect, each at α = 0.05. What is the probability of at least one false positive? How do you deal with it?

<details>
<summary><b>Reference answer</b></summary>

$1 - 0.95^{20} \approx 64\%$.

Remedies: **Bonferroni** changes each test's threshold to α/m and controls the family-wise error rate (FWER, the probability of at least one false positive); it is conservative. **Benjamini–Hochberg** controls the false discovery rate (FDR, the expected share of false positives among rejected hypotheses); it has more power and suits exploratory analyses with many tests. In experiments, the more fundamental fix is to **pre-specify one primary metric** and treat the others as exploratory.

**Short version:** About 64%. Bonferroni controls the family-wise error rate by testing each at alpha over m; Benjamini–Hochberg controls the false discovery rate and is less conservative. In experiments, the cleaner fix is to pre-register one primary metric.

</details>

## 9. Maximum likelihood estimation (MLE)

The likelihood treats the joint density of the data as a function of the **parameter**: $L(\theta; x) = p(x \mid \theta)$. The MLE is the $\theta$ that maximizes it.

**🗣 Say it out loud:** Derive the MLE of p from Bernoulli observations $x_1, \dots, x_n$.

<details>
<summary><b>Reference answer</b></summary>

Log-likelihood:

$$\ell(p) = \sum_i \big[x_i \log p + (1 - x_i)\log(1 - p)\big]$$

Differentiate and set to 0: $\dfrac{\sum x_i}{p} - \dfrac{n - \sum x_i}{1 - p} = 0$, which gives $\hat p = \bar x$, the sample proportion.

**Short version:** Write the log-likelihood, differentiate with respect to p, set it to zero, and you get the sample proportion.

</details>

**🗣 Say it out loud:** What is the difference between MLE and MAP?

<details>
<summary><b>Reference answer</b></summary>

MLE maximizes the likelihood $p(x \mid \theta)$; MAP maximizes the posterior $p(\theta \mid x) \propto p(x \mid \theta)\,p(\theta)$, which adds a prior. On the log scale the prior becomes a penalty term: a Gaussian prior corresponds to L2 regularization (ridge), a Laplace prior to L1 regularization (lasso). With a uniform prior the two coincide; with lots of data the prior's influence vanishes and the two converge.

**Short version:** MLE maximizes the likelihood; MAP maximizes the posterior, which adds a prior. On the log scale the prior acts as a penalty: a Gaussian prior gives ridge and a Laplace prior gives lasso.

</details>
