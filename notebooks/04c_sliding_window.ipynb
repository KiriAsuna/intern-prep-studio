# 04c | Sliding Window

When a problem mentions a "**contiguous** subarray," a "substring," or "the longest / shortest stretch," it's almost always a sliding window. It's also a same-direction two-pointer technique: the stretch between `left` and `right` is the "window."

There are two kinds: **fixed-length windows** and **variable-length windows**.

About 50 minutes.

## 1. Fixed-length windows

"What is the maximum sum of a contiguous subarray of length k?"

Direct approach: re-add k numbers for every starting point, O(n·k).

Sliding window: when the window moves one step right, **add the element coming in and subtract the one going out**; no need to recompute. O(n).

```
a = [1, 4, 2, 10, 2],  k = 3
[1, 4, 2]            sum = 7
   [4, 2, 10]        sum = 7 + 10 - 1 = 16
      [2, 10, 2]     sum = 16 + 2 - 4 = 14
```

    if k <= 0 or k > len(nums):
        return []
    cur = sum(nums[:k])

    for i in range(k, len(nums)):

## 2. Variable-length windows

"The **longest** (or shortest) contiguous stretch that satisfies some condition." Template:

```python
left = 0
for right in range(len(xs)):
    add xs[right] to the window
    while the window violates the condition:
        remove xs[left] from the window
        left += 1
    update the answer using the current window [left, right]
```

`right` moves one step right each time to **expand** the window; as soon as the condition fails, move `left` to **shrink** the window until it holds again.

Both pointers only move right, and each element enters and leaves the window at most once, so it's O(n) even though it looks like two nested loops.

**Example: the longest substring without repeating characters.** Track the characters in the window with a set. If a new character is already in the window, keep removing characters from the left until it isn't.

Take `"abcabcbb"`: the window grows to `abc` (length 3); the next `a` is a repeat, so the left `a` is removed and the window becomes `bca`; and so on. The longest is 3.

Adding to and removing from a set: `seen.add(x)` adds, `seen.remove(x)` removes (raises an error if x isn't there).

````python
seen = set()
seen.add("a"); seen.add("b")
print(seen, "a" in seen)
seen.remove("a")
print(seen, "a" in seen)
````

def longest_unique(s):

    left = 0
    best = 0
    for right in range(len(s)):
        while s[right] in seen:
            seen.remove(s[left])
            left += 1
        seen.add(s[right])
        best = max(best, right - left + 1)
    return best

```python
def min_subarray_len(target, nums):
    left = 0
    total = 0
    best = len(nums) + 1
    for right in range(len(nums)):
        total += nums[right]
        while total >= target:
            best = min(best, right - left + 1)
            total -= nums[left]
            left += 1
    return 0 if best == len(nums) + 1 else best
```

- Fixed window: add the new, subtract the old;
- Variable window: `for right` expands, `while` shrinks; for the longest, update the answer after the `while`; for the shortest, update it inside the `while`;
