        return None
    return sum(xs) / len(xs)

## C. SQL diagnostic

Table `orders(user_id, order_date, amount)`.

1. Compute each user's total spend.
2. Keep only users whose total spend is > 100.
3. For each user, compute cumulative spend by date.

<details>
<summary><b>Answer</b></summary>

```sql
-- 1
SELECT user_id, SUM(amount) AS total
FROM orders
GROUP BY user_id;

-- 2
SELECT user_id, SUM(amount) AS total
FROM orders
GROUP BY user_id
HAVING SUM(amount) > 100;

-- 3
SELECT
  user_id,
  order_date,
  amount,
  SUM(amount) OVER (
      PARTITION BY user_id
      ORDER BY order_date
  ) AS running_total
FROM orders;
```

</details>
