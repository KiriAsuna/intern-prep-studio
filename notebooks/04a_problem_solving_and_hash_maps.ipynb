## A fixed problem-solving process

In a live interview (someone watching you code), **say each step out loud**; in an OA, run through them in your head.

1. **Restate the problem** and confirm the input and output types;
2. **Work a small example by hand**;
3. **State the most direct approach first** (even if slow) and its complexity;
4. **Look for something faster**: usually a different data structure, or a known pattern;
5. **Write the code**;
6. **Trace through it mentally with your example**;
7. **Check edge cases**: empty input, a single element, all elements equal, negative numbers;
8. **State the time and space complexity**.

Many people fail before step 5: they start coding right after reading the problem and discover halfway through that they misunderstood it.

## Complexity cheat sheet

| Operation | Complexity |
|---|---|
| One pass | O(n) |
| Two nested loops | O(n²) |
| Sorting | O(n log n) |
| Binary search | O(log n) |
| `x in set`, `d[key]`, `set.add`, `list.append` | O(1) |
| `x in list`, `list.index(x)`, `list.pop(0)`, `list.insert(0, x)` | O(n) |

**Work backward from the input size to how fast you need to be** (OA problems usually state the range of n):

| Max n (approx.) | Acceptable complexity |
|---|---|
| 1,000 | O(n²) is fine |
| 100,000 | Need O(n log n) or O(n) |
| 10⁹ | Need O(log n) or a math formula |

Hidden test cases usually include one or two very large inputs designed specifically to fail O(n²) solutions.

## Hash maps: dict and set

"Have I seen it?", "How many times does it appear?", "Where is it?": all three kinds of questions use a hash map, which turns an O(n) lookup into O(1). You already used them in 01e and 01h.

The standard library has two handy tools:

- `Counter(xs)`: gives you an "element → count" dict directly (the counting pattern from 01e);
- `defaultdict(list)`: automatically creates an empty list when you access a missing key, which is ideal for grouping.

````python
from collections import Counter, defaultdict

c = Counter("statistics")
print(c)
print(c["s"], c["z"])            # a missing key returns 0 instead of raising an error
print(c.most_common(2))          # the two most common

groups = defaultdict(list)
for word in ["apple", "avocado", "banana", "blueberry", "cherry"]:
    groups[word[0]].append(word)     # group by first letter
print(dict(groups))
````

    groups = defaultdict(list)
    for w in words:
        key = "".join(sorted(w))
        groups[key].append(w)
    return list(groups.values())
