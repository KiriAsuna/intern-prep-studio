## 1. Assignment does not copy ⚠️

First guess what the code below prints, then run it.

````python
a = [1, 2]
b = a
b.append(3)
print(a)
print(b)
````

`a` has also become `[1, 2, 3]`.

In R, after `b <- a`, changing `b` leaves `a` untouched (R copies automatically on modification). **Python does not copy.** `b = a` means "give the same list a second name, b." `a` and `b` are two names for the same object.

If you want an independent copy, you have to ask for it explicitly:

```python
b = a.copy()
```

## 2. A list passed into a function is the same list

Same idea: if a function modifies the list passed into it, the list outside changes too.

````python
def add_zero(xs):
    xs.append(0)

data = [5, 6]
add_zero(data)
print(data)      # [5, 6, 0]: the data outside was modified
````

So when writing a function, unless the problem asks you to "modify in place," it's safer to **build a new list and return it**.

## 3. Which objects behave like this

- **Mutable** (can be modified in place, so they have the problem above): `list`, `dict`, `set`
- **Immutable** (can't be modified in place, nothing to worry about): `int`, `float`, `str`, `tuple`

Numbers and strings don't have this problem because they can't be changed at all; any "modification" actually creates a new object.

## 4. Shallow copies

`copy()` only copies the outermost layer. If a list contains other lists, the inner lists are still shared. Guess the output first, then run it:

````python
a = [[1], [2]]
b = a.copy()
b[0].append(9)
print(a)
print(b)
````

<details>
<summary><b>Explanation</b></summary>

Both show `[[1, 9], [2]]`. `a.copy()` created a new outer list, but the inner `[1]` is still the same object.

To copy the inner layers too, `import copy` and use `copy.deepcopy(a)`. For interviews, knowing that "a shallow copy only copies one level" is enough.

</details>

## 5. == vs. is

- `==`: are the **values** equal;
- `is`: is it **the same object**.

For everyday comparisons, always use `==`. The one exception: to check for `None`, the convention is `x is None`.

````python
a = [1, 2]
b = [1, 2]
c = a
print(a == b)   # True: same contents
print(a is b)   # False: two different lists
print(a is c)   # True: c is a
````

## 6. Handling errors: try / except

Some operations can fail, such as converting `"abc"` to an integer. `try / except` is the equivalent of R's `tryCatch()`:

```python
try:
    code that might fail
except ErrorType:
    runs if it fails
```

````python
text = "abc"
try:
    value = int(text)
    print("Conversion succeeded", value)
except ValueError:
    print("Conversion failed: not a number")
````

Common error types:

| Error | When it occurs |
|---|---|
| `ValueError` | Inappropriate value, e.g. `int("abc")` |
| `TypeError` | Wrong type, e.g. `int(None)`, `1 + "a"` |
| `ZeroDivisionError` | Division by 0 |
| `KeyError` | The key isn't in the dict |
| `IndexError` | List index out of range |
