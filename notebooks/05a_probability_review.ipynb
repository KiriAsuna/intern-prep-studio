## 1. Conditional probability and Bayes

$$P(A \mid B) = \frac{P(A \cap B)}{P(B)}, \qquad P(A \mid B) = \frac{P(B \mid A)\,P(A)}{P(B)}$$

**🗣 Say it out loud:** A disease has a prevalence of 1%, and a test has 99% sensitivity and 95% specificity. If a person tests positive, what is the probability they actually have the disease?

<details>
<summary><b>Reference answer</b></summary>

$$P(D \mid +) = \frac{0.99 \times 0.01}{0.99 \times 0.01 + 0.05 \times 0.99} = \frac{0.0099}{0.0594} \approx 16.7\%$$

After a positive result, the probability of disease is only about one in six, because the base rate is low: there are so many healthy people that even with only a 5% false-positive rate, false positives far outnumber true positives.

**How to say it in an interview:** lead with the conclusion and the intuition ("the base rate is low, so the posterior is far below 99%"), then write out the formula.

**Sample phrasing:** About 17%. Even with a very accurate test, the low base rate means false positives from the healthy 99% outnumber true positives, so the posterior is far below the test's sensitivity.

</details>

**🗣 Say it out loud:** What is the difference between "independent" and "mutually exclusive"?

<details>
<summary><b>Reference answer</b></summary>

Mutually exclusive: the two events can't happen together, $P(A \cap B) = 0$. Independent: whether one happens doesn't change the probability of the other, $P(A \cap B) = P(A)P(B)$. Two mutually exclusive events that both have positive probability are **always dependent**: knowing A happened tells you B didn't.

**Sample phrasing:** Mutually exclusive means they can't both happen; independent means knowing one tells you nothing about the other. Two mutually exclusive events with positive probability are actually dependent.

</details>

## 2. Expectation and variance

$$E[aX + b] = aE[X] + b, \qquad \mathrm{Var}(aX + b) = a^2\,\mathrm{Var}(X), \qquad \mathrm{Var}(X) = E[X^2] - E[X]^2$$

$$\mathrm{Var}(X + Y) = \mathrm{Var}(X) + \mathrm{Var}(Y) + 2\,\mathrm{Cov}(X, Y)$$

Linearity of expectation, $E[X + Y] = E[X] + E[Y]$, **does not require independence**. It is the most useful tool for expectation problems (the indicator-variable method).

**🗣 Say it out loud:** Why is $E[g(X)] \neq g(E[X])$ in general? Give an example.

<details>
<summary><b>Reference answer</b></summary>

Take $g(x) = x^2$: $E[X^2] = \mathrm{Var}(X) + E[X]^2$, which exceeds $E[X]^2$ whenever X is not constant. More generally, for convex $g$, $E[g(X)] \ge g(E[X])$ (Jensen's inequality). In practice: "a function of the average" is not "the average of the function". For example, you can't plug an average growth rate straight into a compound-interest formula, and $E[1/X] \ne 1/E[X]$ (a common pitfall with ratio metrics).

**Sample phrasing:** Because expectation is linear but g generally isn't. With g(x) = x squared, E[X²] exceeds (E[X])² by the variance. Jensen's inequality gives the direction for convex g.

</details>

## 3. Covariance and correlation

$$\mathrm{Cov}(X, Y) = E[(X - EX)(Y - EY)], \qquad \mathrm{Corr}(X, Y) = \frac{\mathrm{Cov}(X, Y)}{\sigma_X \sigma_Y}$$

Independence ⇒ zero covariance; the converse generally fails.

**🗣 Say it out loud:** What's the difference between covariance and correlation? Construct an example that is "uncorrelated but not independent".

<details>
<summary><b>Reference answer</b></summary>

Covariance has units, and its size depends on the scale of the variables; correlation is standardized covariance, between −1 and 1, and measures only the strength of a **linear** relationship.

Example: $X \sim \mathrm{Uniform}(-1, 1)$, $Y = X^2$. By symmetry $E[X^3] = 0$, so $\mathrm{Cov}(X, Y) = E[X^3] - E[X]E[X^2] = 0$, yet Y is completely determined by X. Zero correlation only means there is no linear relationship.

**Sample phrasing:** Covariance has units; correlation is the scale-free version between −1 and 1 and only captures linear association. If X is symmetric around zero and Y equals X squared, they're uncorrelated but completely dependent.

</details>

## 4. Law of total expectation and law of total variance

$$E[Y] = E\big[E[Y \mid X]\big], \qquad \mathrm{Var}(Y) = E\big[\mathrm{Var}(Y \mid X)\big] + \mathrm{Var}\big(E[Y \mid X]\big)$$

Total variance = the average within-group variance + the variance of the group means.

**🗣 Say it out loud:** A population is split 50/50 into two groups. Group A has mean 0 and variance 1; group B has mean 4 and variance 1. What is the overall variance?

<details>
<summary><b>Reference answer</b></summary>

Within groups: $E[\mathrm{Var}(Y \mid G)] = 1$. Between groups: the group mean is 0 or 4 with probability one half each, so its variance is $\frac{(0-2)^2 + (4-2)^2}{2} = 4$. Overall variance $= 1 + 4 = 5$.

**Sample phrasing:** By the law of total variance: within-group variance 1 plus between-group variance 4, so 5.

</details>

**🗣 Say it out loud:** What is Simpson's paradox?

<details>
<summary><b>Reference answer</b></summary>

A trend that points one way within every group reverses once the groups are combined. The cause is a confounding variable that affects both the makeup of the groups and the outcome. Classic example: a drug has a higher cure rate among both mild and severe patients, but because it is given mostly to severe patients, its pooled cure rate is lower. The lesson: before comparing two groups, check whether their composition is comparable; this is also why A/B tests use random assignment.

**Sample phrasing:** A trend that holds within every subgroup reverses when the groups are pooled, because a confounder changes the mix of subgroups. It's why we compare like with like, and why randomization matters.

</details>

## 5. Common distributions: recognize the scenario

| Distribution | Scenario | Mean | Variance |
|---|---|---|---|
| Bernoulli(p) | A single 0/1 trial | p | p(1−p) |
| Binomial(n, p) | Number of successes in n independent trials | np | np(1−p) |
| Geometric(p) | Number of trials until the first success | 1/p | (1−p)/p² |
| Poisson(λ) | Number of events in a fixed time / region | λ | λ |
| Exponential(λ) | Waiting time between events in a Poisson process | 1/λ | 1/λ² |
| Uniform(a, b) | Equally likely over an interval | (a+b)/2 | (b−a)²/12 |
| Normal(μ, σ²) | Sum of many small effects; CLT | μ | σ² |

t: a standardized normal mean when the variance is unknown; χ²: a sum of squared independent standard normals; F: a ratio of two independent χ²/df.

**🗣 Say it out loud:** In a Poisson process, why are waiting times exponentially distributed? What is the most important property of the exponential distribution?

<details>
<summary><b>Reference answer</b></summary>

With event rate λ, $P(T > t) = P(\text{no events in [0, t]}) = e^{-\lambda t}$, which is exactly the survival function of Exponential(λ). The most important property is **memorylessness**: $P(T > s + t \mid T > s) = P(T > t)$; how long you've already waited doesn't affect how much longer you'll wait.

**Sample phrasing:** The chance of no event in time t is e to the minus lambda t, which is the exponential survival function. Its key property is memorylessness.

</details>

## 6. LLN and CLT

- **Law of large numbers (LLN):** the sample mean converges to the population mean. It tells you "**where it lands**".
- **Central limit theorem (CLT):** the **distribution** of the standardized sample mean approaches a normal distribution. It tells you "**how it fluctuates around there**".

$$\sqrt{n}\,(\bar X_n - \mu) \xrightarrow{d} N(0, \sigma^2)$$

Don't blur the two into "with enough data, things become normal". What becomes normal is **the sampling distribution of the sample mean**, not the data itself.

**🗣 Say it out loud:** The raw data is extremely skewed (say, income or user spending). With n = 100, can you safely rely on the CLT?

<details>
<summary><b>Reference answer</b></summary>

Not necessarily. The CLT is an asymptotic result, and how good the approximation is in a finite sample depends on the skewness and tails of the underlying distribution. Heavy-tailed data (a few "whale users" driving most of the spending) needs a much larger n, and if the variance doesn't exist, the CLT doesn't apply at all. In practice: check whether the bootstrap distribution is symmetric; winsorize or log-transform the metric; or switch to robust metrics such as the median or other quantiles.

**Sample phrasing:** Not necessarily. The CLT is asymptotic, and heavy-tailed data like revenue may need far larger samples. I'd check the bootstrap distribution, and consider winsorizing, a log transform, or a robust metric.

</details>

## 7. Quick calculations

Give the formula for each within 30 seconds.

**🗣 Say it out loud:** ① The probability that two dice sum to 7; ② the probability of at least one success in 10 Bernoulli(0.2) trials; ③ X ~ Poisson(3), P(X = 0); ④ the mean and variance of X ~ Exp(2); ⑤ you flip two coins at a time until both come up heads for the first time; how many rounds do you expect?

<details>
<summary><b>Reference answer</b></summary>

① $6/36 = 1/6$. ② $1 - 0.8^{10} \approx 0.893$. ③ $e^{-3} \approx 0.050$. ④ Mean $1/2$, variance $1/4$. ⑤ Each round succeeds with probability $1/4$; the mean of a geometric distribution is $1/p = 4$.

General tricks: for "at least once", first compute "never"; for "expected number of tries", think geometric distribution.

</details>
