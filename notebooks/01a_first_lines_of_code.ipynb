## 1. Arithmetic

Python works as a calculator out of the box. Most operators are the same as in R; only three differ:

| | R | Python |
|---|---|---|
| Power | `2^10` | `2 ** 10` |
| Floor division | `7 %/% 2` | `7 // 2` |
| Remainder | `7 %% 2` | `7 % 2` |

Anything after `#` is a comment, same as in R.

````python
print(7 / 2)    # regular division, gives 3.5
print(7 // 2)   # floor division, gives 3
print(7 % 2)    # remainder, gives 1
print(2 ** 10)  # 2 to the 10th power
````

## 2. Variables

Assign with `=` (R's `<-`).

Use lowercase letters and underscores for variable names, e.g. `sample_size`. **You can't use dots like in R**: `sample.size` means something else in Python ("the `size` attribute of `sample`") and raises an error.

````python
n = 25
total = 80
mean = total / n
print(mean)
````

## 3. Four basic types

| Type | Example | R equivalent |
|---|---|---|
| `int` integer | `25` | integer |
| `float` decimal | `3.2` | numeric |
| `str` string | `"hello"` | character |
| `bool` logical | `True` / `False` | logical |

Note that `True` / `False` are capitalized only on the first letter. "Nothing" is written `None` (similar to R's `NULL`).

`type(x)` tells you the type, like R's `class(x)`.

````python
print(type(25))
print(type(3.2))
print(type("hello"))
print(type(True))
````

## 4. Strings

Single or double quotes both work. `+` concatenates, and `len()` gives the length (R's `nchar()`).

````python
first = "Data"
second = "Science"
full = first + " " + second
print(full)
print(len(full))
````

### f-strings: putting variables into a sentence

Put an `f` before the opening quote, and you can write variables directly inside curly braces `{}`. It's like R's `paste0()` or `sprintf()`, but easier to read.

`{x:.2f}` means "round to two decimal places."

````python
n = 25
mean = 3.14159
print(f"The sample size is {n} and the mean is {mean:.2f}")
````

## 5. Type conversion

You can't add a number and a string directly; convert first with `int()`, `float()`, `str()` (R's `as.integer()`, `as.numeric()`, `as.character()`).

````python
text = "42"
print(int(text) + 1)     # convert to int, then add: 43
print(text + "1")        # string concatenation: "421"
print(float("3.5") * 2)  # 7.0
````

## 6. Reading errors

The code below is **wrong on purpose**. Click Run and see what happens.

````python
height = 170
print(hieght)
````

The error message is long, but you only need the **last line**:

```
NameError: name 'hieght' is not defined
```

It means there is no variable called `hieght` (it's misspelled). The four most common errors:

| Error | Usual cause |
|---|---|
| `NameError` | Misspelled variable name, or the code box that defines it hasn't been run |
| `SyntaxError` | Missing parenthesis, quote, or colon |
| `IndentationError` | Wrong indentation (covered in 01c) |
| `TypeError` | Wrong type, e.g. adding a number to a string |
