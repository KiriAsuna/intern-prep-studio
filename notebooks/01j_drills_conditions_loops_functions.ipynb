### Problem 1 · Binning with if / elif / else

`bmi = 26.3`: below 18.5 is `"underweight"`, 18.5 to 25 is `"normal"`, 25 and above is `"overweight"`.

```python
bmi = 26.3
if bmi < 18.5:
    label = "underweight"
elif bmi < 25:
    label = "normal"
else:
    label = "overweight"
label
```

**Result:**

```text
'overweight'
```

> `elif` is only checked when all earlier conditions failed, so the second condition doesn't need `bmi >= 18.5`.

### Problem 2 · Joining two conditions with and

`age = 30`: is it between 18 and 65 (including 18, excluding 65)?

```python
age = 30
age >= 18 and age < 65
```

**Result:**

```text
True
```

> Logical operators are the English words `and`, `or`, `not`, not R's `&`, `|`, `!`.

### Problem 3 · or and not

`day = "Sat"`: is it a weekend? Then: is it a weekday?

```python
day = "Sat"
is_weekend = day == "Sat" or day == "Sun"
print(is_weekend, not is_weekend)
```

**Result:**

```text
True False
```

### Problem 5 · Accumulate

Add up 1 through 100.

```python
total = 0
for i in range(1, 101):
    total += i
total
```

**Result:**

```text
5050
```

> Pattern 1: set up `total = 0` first, then `total += i` inside the loop.

### Problem 6 · Running product

5 factorial, `5 × 4 × 3 × 2 × 1`.

```python
prod = 1
for i in range(1, 6):
    prod = prod * i
prod
```

**Result:**

```text
120
```

> The starting value for a product is 1, not 0.

### Problem 7 · Count

How many negative numbers are in `nums = [3, -1, 4, 0, -5, 8]`?

```python
nums = [3, -1, 4, 0, -5, 8]
count = 0
for x in nums:
    if x < 0:
        count += 1
count
```

**Result:**

```text
2
```

> Pattern 2: add 1 whenever the condition holds.

### Problem 8 · Collect

Collect the even numbers in `nums` into a new list.

```python
evens = []
for x in nums:
    if x % 2 == 0:
        evens.append(x)
evens
```

**Result:**

```text
[4, 0, 8]
```

> Pattern 3: start with an empty list and `append` inside the loop. 0 is even too.

### Problem 9 · enumerate: when you need the index

All positions in `xs = [5, 3, 5, 1, 5]` where the value equals 5.

```python
xs = [5, 3, 5, 1, 5]
where = []
for i, x in enumerate(xs):
    if x == 5:
        where.append(i)
where
```

**Result:**

```text
[0, 2, 4]
```

> `enumerate` gives you the index `i` and the element `x` at once.

### Problem 10 · Hand-rolled maximum and its position

Without `max`, find the largest number in `nums` and its index.

```python
best = nums[0]
best_i = 0
for i, x in enumerate(nums):
    if x > best:
        best = x
        best_i = i
print(best, best_i)
```

**Result:**

```text
8 5
```

> Take the first element as the "current max," and replace it whenever you find something bigger. Interviews sometimes ask you to write this by hand without built-ins to see whether you can write a loop.

### Problem 11 · break: find the first one

What is the first negative number in `nums`?

```python
first_neg = None
for x in nums:
    if x < 0:
        first_neg = x
        break
first_neg
```

**Result:**

```text
-1
```

> `break` as soon as you find it; nothing after it is examined. Initialize to `None` so there's still a result if nothing is found.

### Problem 12 · while: sum of digits

The sum of the digits of 12345.

```python
n = 12345
digit_sum = 0
while n > 0:
    digit_sum += n % 10
    n = n // 10
digit_sum
```

**Result:**

```text
15
```

> `n % 10` takes the last digit and `n // 10` drops it, until n becomes 0.

### Problem 13 · Cumulative sum

The cumulative sum of `xs = [1, 2, 3, 4]`, i.e. `[1, 3, 6, 10]`.

```python
xs = [1, 2, 3, 4]
running = []
total = 0
for x in xs:
    total += x
    running.append(total)
running
```

**Result:**

```text
[1, 3, 6, 10]
```

> The accumulate and collect patterns combined: record the total after each addition. Equivalent to R's `cumsum()`.

### Problem 14 · The simplest function

Write a function `is_even(n)` that returns `True` for even numbers.

```python
def is_even(n):
    return n % 2 == 0

print(is_even(4), is_even(7))
```

**Result:**

```text
True False
```

> A comparison is already `True` / `False`, so just `return` it; no need for `if ... return True else return False`.

### Problem 15 · Default arguments

Write a function `power(x, p=2)` that squares by default.

```python
def power(x, p=2):
    return x ** p

print(power(3), power(3, 3), power(2, p=10))
```

**Result:**

```text
9 27 1024
```

### Problem 16 · Returning multiple values

Write a function `summary(xs)` that returns the minimum, maximum, and mean.

```python
def summary(xs):
    return min(xs), max(xs), sum(xs) / len(xs)

lo, hi, avg = summary([4, 8, 6])
print(lo, hi, avg)
```

**Result:**

```text
4 8 6.0
```

> It returns a tuple, which you unpack with commas.

### Problem 17 · Edge case: early return

Write a function `first_or_none(xs)`: return the first element, or `None` for an empty list.

```python
def first_or_none(xs):
    if len(xs) == 0:
        return None
    return xs[0]

print(first_or_none([7, 8]), first_or_none([]))
```

**Result:**

```text
7 None
```

> Taking `xs[0]` of an empty list raises an `IndexError`.

### Problem 18 · A loop inside a function

Write a function `count_vowels(s)` that counts the vowels a, e, i, o, u in a string.

```python
def count_vowels(s):
    count = 0
    for ch in s:
        if ch in "aeiou":
            count += 1
    return count

count_vowels("statistics")
```

**Result:**

```text
3
```

> `ch in "aeiou"` checks whether a character is one of those letters.

### Problem 19 · Primality test

Write a function `is_prime(n)`.

```python
def is_prime(n):
    if n < 2:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False
    return True

[n for n in range(1, 20) if is_prime(n)]
```

**Result:**

```text
[2, 3, 5, 7, 11, 13, 17, 19]
```

> As soon as you find a divisor, `return False` early; only if the loop finishes without finding one is it prime. Note that 1 is not prime, so handle it first.

### Problem 20 · Reverse a list without slicing

Reverse `[1, 2, 3, 4]`.

```python
def reverse_list(xs):
    out = []
    for x in xs:
        out = [x] + out
    return out

reverse_list([1, 2, 3, 4])
```

**Result:**

```text
[4, 3, 2, 1]
```

> Put each new element at the front. `[x] + out` concatenates two lists.

## Recap of this lesson's syntax

| Task | Code |
|---|---|
| Binning | `if` / `elif` / `else`, checking from strictest to loosest |
| Combine conditions | `and`, `or`, `not` |
| A run of integers | `range(1, n + 1)` is 1 through n |
| Accumulate / count / collect | `total += x` / `count += 1` / `out.append(x)` |
| Need the index | `for i, x in enumerate(xs):` |
| Stop when found | `break`; inside a function, just `return` |
| Function | `def f(x, p=2):` + **always `return`** |
| Edge cases | At the top: `if len(xs) == 0: return None` |
