### Problem 1 · Build a dict and look up a value

Build a dict storing Ann with 90 points and Bob with 82, then get Bob's score.

```python
scores = {"Ann": 90, "Bob": 82}
scores["Bob"]
```

**Result:**

```text
82
```

> The keys are strings, so they need quotes.

### Problem 2 · Modify and add

Change Bob to 88, then add Cy with 70.

```python
scores["Bob"] = 88
scores["Cy"] = 70
scores
```

**Result:**

```text
{'Ann': 90, 'Bob': 88, 'Cy': 70}
```

> If the key exists it's modified; if not, it's added. The syntax is the same.

### Problem 3 · A default when the key is missing

Get Eve's score, treating a missing one as 0.

```python
scores.get("Eve", 0)
```

**Result:**

```text
0
```

> Writing `scores["Eve"]` directly raises a `KeyError`.

### Problem 4 · Is the key there?

Are `Ann` and `Eve` in `scores`?

```python
print("Ann" in scores, "Eve" in scores)
```

**Result:**

```text
True False
```

> For a dict, `in` checks the keys.

### Problem 5 · Loop over keys and values

Print each person's name and score, one per line.

```python
for name, s in scores.items():
    print(name, s)
```

**Result:**

```text
Ann 90
Bob 88
Cy 70
```

### Problem 6 · All keys, all values

A list of all the names; the average of all the scores.

```python
names = list(scores.keys())
avg = sum(scores.values()) / len(scores)
print(names, f"{avg:.2f}")
```

**Result:**

```text
['Ann', 'Bob', 'Cy'] 82.67
```

> `len(dict)` is the number of keys.

### Problem 7 · The counting pattern

`votes = ["A", "B", "A", "C", "A", "B"]`: how many votes did each option get?

```python
votes = ["A", "B", "A", "C", "A", "B"]
counts = {}
for v in votes:
    counts[v] = counts.get(v, 0) + 1
counts
```

**Result:**

```text
{'A': 3, 'B': 2, 'C': 1}
```

> R's `table()`. Memorize this line: `counts[x] = counts.get(x, 0) + 1`.

### Problem 8 · Counting characters

How many times does each letter appear in `"banana"`?

```python
counts = {}
for ch in "banana":
    counts[ch] = counts.get(ch, 0) + 1
counts
```

**Result:**

```text
{'b': 1, 'a': 3, 'n': 2}
```

> You can walk a string character by character with `for ch in s:`.

### Problem 9 · The most frequent one

From the `banana` counts above, find the letter that appears most often.

```python
best = None
best_count = 0
for ch, c in counts.items():
    if c > best_count:
        best = ch
        best_count = c
print(best, best_count)
```

**Result:**

```text
a 3
```

> Same as the "hand-rolled maximum," except you loop over `items()`.

### Problem 10 · Sum by category

`sales = [("A", 10), ("B", 5), ("A", 7)]`: the total for each category.

```python
sales = [("A", 10), ("B", 5), ("A", 7)]
totals = {}
for cat, amount in sales:
    totals[cat] = totals.get(cat, 0) + amount
totals
```

**Result:**

```text
{'A': 17, 'B': 5}
```

> Replace `+ 1` in the counting pattern with `+ amount` and you get a sum. In R: `tapply(amount, cat, sum)`.

### Problem 11 · Group by first letter

Group `words = ["apple", "bob", "avocado", "cat", "bean"]` by first letter.

```python
words = ["apple", "bob", "avocado", "cat", "bean"]
groups = {}
for w in words:
    k = w[0]
    if k not in groups:
        groups[k] = []
    groups[k].append(w)
groups
```

**Result:**

```text
{'a': ['apple', 'avocado'], 'b': ['bob', 'bean'], 'c': ['cat']}
```

> The first time you see a key, put an empty list there, then `append` to it.

### Problem 12 · Count distinct values with a set

How many distinct ids are in `ids = [101, 102, 101, 103, 102]`?

```python
ids = [101, 102, 101, 103, 102]
len(set(ids))
```

**Result:**

```text
3
```

> R's `length(unique(ids))`.

### Problem 13 · Using a set to remember "seen"

Read `ids` in order and collect the ids that appear a second time or later (each recorded only once).

```python
seen = set()
dups = []
for x in ids:
    if x in seen and x not in dups:
        dups.append(x)
    seen.add(x)
dups
```

**Result:**

```text
[101, 102]
```

> Sets use `add`; lists use `append`.

### Problem 14 · Elements common to two lists

The elements in both `a = [1, 2, 3, 4]` and `b = [3, 4, 5]`.

```python
a = [1, 2, 3, 4]
b = [3, 4, 5]
b_set = set(b)
[x for x in a if x in b_set]
```

**Result:**

```text
[3, 4]
```

> Convert `b` to a set first so `x in b_set` is fast. For elements only in `a`, replace `in` with `not in`.

### Problem 15 · Comprehension: transform

Multiply each number in `xs = [3, -1, 4, 0, 8]` by 2.

```python
xs = [3, -1, 4, 0, 8]
[x * 2 for x in xs]
```

**Result:**

```text
[6, -2, 8, 0, 16]
```

> You can't just write `xs * 2` for a list (that repeats it twice); use a comprehension.

### Problem 16 · Comprehension: filter and transform

Square the positive numbers in `xs`.

```python
[x ** 2 for x in xs if x > 0]
```

**Result:**

```text
[9, 16, 64]
```

> Read it as: for each x in xs, if x > 0, take the square of x.

### Problem 17 · Clean up a string

Turn `raw = "  Data Science,ML ,stats "` into a lowercase list with no extra spaces.

```python
raw = "  Data Science,ML ,stats "
[p.strip().lower() for p in raw.split(",")]
```

**Result:**

```text
['data science', 'ml', 'stats']
```

> `split` to break it up → `strip` each item's leading/trailing spaces → `lower` to lowercase.

### Problem 18 · Join and replace

Join `["2026", "01", "15"]` into `2026/01/15`, then replace `/` with `-`.

```python
d = "/".join(["2026", "01", "15"])
print(d)
print(d.replace("/", "-"))
```

**Result:**

```text
2026/01/15
2026-01-15
```

> The syntax is `separator.join(list)`, the reverse order of R's `paste(x, collapse = "/")`.

### Problem 19 · Sort by length

Sort `words` by length, longest first.

```python
sorted(words, key=len, reverse=True)
```

**Result:**

```text
['avocado', 'apple', 'bean', 'bob', 'cat']
```

> `key=len` means sort by the value of `len(w)`. Words of equal length keep their original order.

### Problem 20 · zip and dict comprehensions

Pair `names = ["Ann", "Bob"]` with `ages = [28, 35]` into a dict; then build a "word → length" dict.

```python
names = ["Ann", "Bob"]
ages = [28, 35]
print(dict(zip(names, ages)))
print({w: len(w) for w in words})
```

**Result:**

```text
{'Ann': 28, 'Bob': 35}
{'apple': 5, 'bob': 3, 'avocado': 7, 'cat': 3, 'bean': 4}
```

> `zip` pairs two lists by position; a dict comprehension uses curly braces with `key: value`.

## Recap of this lesson's syntax

| Task | Code |

| Get / set | `d["k"]`, `d["k"] = v`, `d.get("k", 0)`, `"k" in d` |
