# 02a | NumPy: Real Vectors

As 01b mentioned, Python lists can't do whole-vector arithmetic the way R vectors can. **The NumPy array is the true counterpart of the R vector.** Every pandas column is built on it, too.

**After this lesson you can:** compute on, filter, and summarize whole collections of numbers without loops, and estimate probabilities by random simulation. About 40 minutes.

## 1. Creating arrays and whole-array arithmetic

The conventional import is `import numpy as np`. `import ... as ...` is like R's `library()`, except that afterwards you prefix everything with `np.`.

````python
import numpy as np

x = np.array([1.0, 2.0, 3.0, 4.0])
print(x * 2)          # multiplies every element by 2, just like R
print(x + x)          # element-wise addition
print(x ** 2)
print(np.sqrt(x))
print(x.shape, x.dtype)   # shape and element type
````

Common ways to create arrays:

| R | NumPy | Gives |
|---|---|---|
| `0:4` | `np.arange(5)` | 0 1 2 3 4 |
| `seq(0, 1, length.out = 5)` | `np.linspace(0, 1, 5)` | 0 0.25 0.5 0.75 1 |
| `rep(0, 3)` | `np.zeros(3)` | 0 0 0 |
| `rep(1, 3)` | `np.ones(3)` | 1 1 1 |

Indexing and slicing work exactly like lists (start at 0, right end excluded).

## 2. Summaries

Most statistics can be written either as `np.mean(x)` or as `x.mean()`.

⚠️ **A classic interview trap:** `x.std()` and `x.var()` divide by **n** by default, whereas R's `sd()` and `var()` divide by **n−1**. For the sample standard deviation you must write `x.std(ddof=1)`.

````python
x = np.array([2.0, 4.0, 4.0, 4.0, 5.0, 5.0, 7.0, 9.0])
print(x.sum(), x.mean(), x.min(), x.max())
print(np.median(x), np.percentile(x, 75))
print(x.std())          # divides by n: 2.0
print(x.std(ddof=1))    # divides by n-1: 2.138..., matches R's sd()
print(x.argmax())       # index of the maximum (R's which.max, but counting from 0)
print(x.cumsum())       # cumulative sum
````

## 3. Boolean filtering

Exactly like R: a comparison gives an array of True/False, which you then use as an index.

- Combine conditions with `&` (and), `|` (or), `~` (not), and **wrap each condition in parentheses**;
- You can't use `and` / `or` here; those only work on a single True/False;
- `np.where(condition, a, b)` is R's `ifelse()`.

True counts as 1 and False as 0, so `(x > 3).sum()` is "how many are greater than 3" and `(x > 3).mean()` is "the share greater than 3." This idiom is used constantly in data analysis.

````python
x = np.array([3, -1, 4, 0, 8, -5])
print(x > 0)
print(x[x > 0])
print(x[(x > 0) & (x < 5)])
print(np.where(x > 0, "pos", "non-pos"))
print((x > 0).sum(), (x > 0).mean())
````

## 4. 2-D arrays and axis

A 2-D array is the counterpart of an R matrix. Access elements with `m[row, col]`; `:` means "all."

`axis` is the most confusing parameter. Remember one sentence: **`axis=0` collapses along the rows and gives one result per column; `axis=1` gives one result per row.**

| R | NumPy |
|---|---|
| `colMeans(m)` / `apply(m, 2, mean)` | `m.mean(axis=0)` |
| `rowMeans(m)` / `apply(m, 1, mean)` | `m.mean(axis=1)` |
| `dim(m)` | `m.shape` |
| `t(m)` | `m.T` |
| `m %*% v` | `m @ v` |

````python
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print(m.shape)          # (2, 3): 2 rows, 3 columns
print(m[0, 1])          # row 1, column 2: 2
print(m[:, 0])          # column 1: [1 4]
print(m.mean(axis=0))   # mean of each column
print(m.mean(axis=1))   # mean of each row
print(m.T)
````

## 5. Random numbers and simulation

First create a random number generator; the seed is the equivalent of R's `set.seed()`:

```python
rng = np.random.default_rng(0)
```

| R | NumPy |
|---|---|
| `rnorm(5, 10, 2)` | `rng.normal(10, 2, 5)` |
| `runif(5)` | `rng.random(5)` |
| `sample(1:6, 5, replace = TRUE)` | `rng.integers(1, 7, 5)` (right end excluded) |
| `sample(x, 3)` | `rng.choice(x, 3, replace=False)` |
| `rbinom(5, 10, 0.3)` | `rng.binomial(10, 0.3, 5)` |

Watch the argument order: R puts the count first, while NumPy puts the count (`size`) last.

Probability questions in interviews can often be checked by "simulate many times and look at the proportion":

````python
rng = np.random.default_rng(0)
flips = rng.integers(0, 2, 100_000)     # 100,000 coin flips, 0 or 1
print(flips[:10])
print(flips.mean())                     # share of heads, close to 0.5
````
