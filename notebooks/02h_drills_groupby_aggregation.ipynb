### Problem 1 · Mean per group

Average salary per department.

```python
employees.groupby("dept")["salary"].mean()
```

**Result:**

| dept | salary |
|---|---|
| Eng | 121.25 |
| Mkt | 72.5 |
| Sales | 76.6667 |

*3 rows × 1 column*

> The result is a Series with the department names in the index.

### Problem 2 · Back to a regular table

Same as above, but return a table with two columns, `dept` and `salary`.

```python
employees.groupby("dept")["salary"].mean().reset_index()
```

**Result:**

|   | dept | salary |
|---|---|---|
| 0 | Eng | 121.25 |
| 1 | Mkt | 72.5 |
| 2 | Sales | 76.6667 |

*3 rows × 2 columns*

### Problem 3 · Sum per group, then sort

Total order amount per user, highest first.

```python
(
    orders.groupby("user_id")["amount"].sum()
    .reset_index()
    .sort_values("amount", ascending=False)
)
```

**Result:**

|   | user_id | amount |
|---|---|---|
| 2 | 3 | 180.0 |
| 6 | 8 | 140.0 |
| 4 | 5 | 135.0 |
| 0 | 1 | 100.0 |
| 5 | 6 | 100.0 |
| 1 | 2 | 30.0 |
| 3 | 4 | 25.0 |
| 7 | 10 | 10.0 |

*8 rows × 2 columns*

### Problem 4 · Rows per group

Number of orders per user.

```python
orders.groupby("user_id")["order_id"].count().reset_index().rename(columns={"order_id": "n_orders"})
```

### Problem 5 · Distinct values per group

On how many **distinct dates** each user logged in.

| 2 | 3 | 5 |
| 3 | 4 | 1 |
| 4 | 5 | 3 |

| 6 | 7 | 1 |
| 7 | 8 | 3 |
| 8 | 9 | 2 |
| 9 | 10 | 1 |

*10 rows × 2 columns*

> User 1 logged in twice on January 2; `nunique` counts that day once.

### Problem 6 · Grouped counts drop missing keys

Number of users per country.

```python
users.groupby("country")["user_id"].count()
```

**Result:**

| country | user_id |

| CA | 2 |
| CN | 3 |

### Problem 7 · Several statistics at once

Headcount, average salary, highest salary and lowest salary per department.

```python
(
    employees.groupby("dept")
    .agg(n=("emp_id", "count"), avg=("salary", "mean"), top=("salary", "max"), low=("salary", "min"))
    .reset_index()
)
```

**Result:**

|   | dept | n | avg | top | low |
|---|---|---|---|---|---|
| 0 | Eng | 4 | 121.25 | 150 | 95 |
| 1 | Mkt | 2 | 72.5 | 85 | 60 |
| 2 | Sales | 3 | 76.6667 | 90 | 70 |

*3 rows × 5 columns*

> Named aggregation: `new_name=("source_column", "function")`, the equivalent of dplyr's `summarise()`.

### Problem 8 · Filter, then group

Using completed orders only, the number of orders and total amount per user.

```python
(
    orders[orders["status"] == "completed"]
    .groupby("user_id")
    .agg(n_orders=("order_id", "count"), total=("amount", "sum"))
    .reset_index()
)
```

**Result:**

|   | user_id | n_orders | total |
|---|---|---|---|
| 0 | 1 | 2 | 80.0 |
| 1 | 2 | 2 | 30.0 |
| 2 | 3 | 3 | 180.0 |
| 3 | 4 | 1 | 25.0 |
| 4 | 5 | 1 | 100.0 |
| 5 | 6 | 2 | 100.0 |
| 6 | 8 | 2 | 140.0 |
| 7 | 10 | 1 | 10.0 |

*8 rows × 3 columns*

### Problem 9 · count vs. nunique

Per day, the number of **logins** and the number of **users** who logged in.

```python
(
    events[events["event_type"] == "login"]
    .groupby("event_date")
    .agg(n_logins=("user_id", "count"), n_users=("user_id", "nunique"))
    .reset_index()
    .head(5)
)
```

**Result:**

|   | event_date | n_logins | n_users |
|---|---|---|---|
| 0 | 2026-01-01 | 2 | 2 |
| 1 | 2026-01-02 | 4 | 3 |
| 2 | 2026-01-03 | 4 | 4 |
| 3 | 2026-01-04 | 2 | 2 |
| 4 | 2026-01-05 | 3 | 3 |

*5 rows × 3 columns*

> There were 4 logins on January 2 but only 3 users. For daily active users, use `nunique`.

### Problem 10 · Group by two columns

Total order amount per country and status.

**Result:**

|   | country | status | amount |

| 1 | CA | refunded | 35.0 |

### Problem 11 · Pivot table

Same as above, but spread status across the columns and fill missing combinations with 0.

```python
m = users.merge(orders, on="user_id")
m.pivot_table(index="country", columns="status", values="amount", aggfunc="sum", fill_value=0)
```

**Result:**

| country | completed | refunded |

### Problem 12 · Filter after grouping (SQL's HAVING)

### Problem 14 · transform: attach a group statistic to every row

Add a column `dept_avg` to every employee: the average salary of their department.

```python
e = employees.copy()
e["dept_avg"] = e.groupby("dept")["salary"].transform("mean")
e[["name", "dept", "salary", "dept_avg"]]
```

**Result:**

|   | name | dept | salary | dept_avg |
|---|---|---|---|---|
| 0 | Alice | Eng | 150 | 121.25 |
| 1 | Brian | Eng | 120 | 121.25 |
| 2 | Cathy | Eng | 120 | 121.25 |
| 3 | David | Eng | 95 | 121.25 |
| 4 | Emma | Sales | 90 | 76.6667 |
| 5 | Frank | Sales | 70 | 76.6667 |
| 6 | Grace | Sales | 70 | 76.6667 |
| 7 | Henry | Mkt | 85 | 72.5 |
| 8 | Irene | Mkt | 60 | 72.5 |

*9 rows × 4 columns*

> `agg` collapses each group to one row; `transform` keeps every row. The equivalent of dplyr's `group_by() %>% mutate()`.

### Problem 15 · Share within group

Each employee's salary as a share of their department's total salary.

```python
e = employees.copy()
e["share"] = e["salary"] / e.groupby("dept")["salary"].transform("sum")
e[["name", "dept", "salary", "share"]]
```

### Problem 16 · Above the group average

Employees paid more than their department's average salary.

```python
e = employees.copy()
e["dept_avg"] = e.groupby("dept")["salary"].transform("mean")
e[e["salary"] > e["dept_avg"]][["name", "dept", "salary", "dept_avg"]]
```

|   | name | dept | salary | dept_avg |

### Problem 17 · Rank within group

Each employee's salary rank within their department (ties share a rank, ranks have no gaps).

```python
e = employees.copy()
e["rk"] = e.groupby("dept")["salary"].rank(method="dense", ascending=False)
e.sort_values(["dept", "rk"])[["dept", "name", "salary", "rk"]]
```

**Result:**

|   | dept | name | salary | rk |
|---|---|---|---|---|
| 0 | Eng | Alice | 150 | 1.0 |
| 1 | Eng | Brian | 120 | 2.0 |
| 2 | Eng | Cathy | 120 | 2.0 |
| 3 | Eng | David | 95 | 3.0 |
| 7 | Mkt | Henry | 85 | 1.0 |
| 8 | Mkt | Irene | 60 | 2.0 |
| 4 | Sales | Emma | 90 | 1.0 |
| 5 | Sales | Frank | 70 | 2.0 |
| 6 | Sales | Grace | 70 | 2.0 |

*9 rows × 4 columns*

> `method="dense"` matches SQL's DENSE_RANK; `"min"` is RANK; `"first"` is ROW_NUMBER.

### Problem 18 · Top N per group (with ties)

Employees in the top 2 salary levels of each department (keep all ties).

```python
e = employees.copy()
e["rk"] = e.groupby("dept")["salary"].rank(method="dense", ascending=False)
e[e["rk"] <= 2].sort_values(["dept", "rk"])[["dept", "name", "salary"]]
```

**Result:**

|   | dept | name | salary |
|---|---|---|---|
| 0 | Eng | Alice | 150 |
| 1 | Eng | Brian | 120 |
| 2 | Eng | Cathy | 120 |
| 7 | Mkt | Henry | 85 |
| 8 | Mkt | Irene | 60 |
| 4 | Sales | Emma | 90 |
| 5 | Sales | Frank | 70 |
| 6 | Sales | Grace | 70 |

*8 rows × 3 columns*

> In Eng, Brian and Cathy tie at 120; ranking keeps both. `sort_values(...).groupby("dept").head(2)` would keep only one.

### Problem 19 · Cumulative sum within group + order number

Each user's running total of spending by date, and which order number this is for them.

```python
o = orders.sort_values(["user_id", "order_date"]).copy()
o["running_total"] = o.groupby("user_id")["amount"].cumsum()
o["nth"] = o.groupby("user_id").cumcount() + 1
o[["user_id", "order_date", "amount", "running_total", "nth"]].head(7)
```

**Result:**

|   | user_id | order_date | amount | running_total | nth |
|---|---|---|---|---|---|
| 0 | 1 | 2026-01-02 | 30.0 | 30.0 | 1 |
| 1 | 1 | 2026-01-05 | 50.0 | 80.0 | 2 |
| 2 | 1 | 2026-01-20 | 20.0 | 100.0 | 3 |
| 3 | 2 | 2026-01-03 | 15.0 | 15.0 | 1 |
| 4 | 2 | 2026-01-04 | 15.0 | 30.0 | 2 |
| 5 | 3 | 2026-01-02 | 80.0 | 80.0 | 1 |
| 6 | 3 | 2026-01-15 | 40.0 | 120.0 | 2 |

*7 rows × 5 columns*

> Always sort by time first. `cumcount()` starts at 0, hence the + 1.

### Problem 20 · Previous row: shift

For each user, the number of days between each order and the previous one.

```python
o = orders.sort_values(["user_id", "order_date"]).copy()
o["order_date"] = pd.to_datetime(o["order_date"])
o["prev_date"] = o.groupby("user_id")["order_date"].shift(1)
o["gap_days"] = (o["order_date"] - o["prev_date"]).dt.days
o[["user_id", "order_date", "prev_date", "gap_days"]].head(7)
```

**Result:**

|   | user_id | order_date | prev_date | gap_days |
|---|---|---|---|---|
| 0 | 1 | 2026-01-02 | NaT | NaN |
| 1 | 1 | 2026-01-05 | 2026-01-02 | 3.0 |
| 2 | 1 | 2026-01-20 | 2026-01-05 | 15.0 |
| 3 | 2 | 2026-01-03 | NaT | NaN |
| 4 | 2 | 2026-01-04 | 2026-01-03 | 1.0 |
| 5 | 3 | 2026-01-02 | NaT | NaN |
| 6 | 3 | 2026-01-15 | 2026-01-02 | 13.0 |

*7 rows × 4 columns*

> `shift(1)` is SQL's LAG. Each user's first order has no previous order, so the value is missing.

## Recap

| Task | Code |
|---|---|
| One number per group | `df.groupby("g")["x"].mean().reset_index()` |
| Several numbers per group | `.agg(new_name=("column", "function"))` |
| Count rows / count distinct values | `"count"` / `"nunique"` |
| Filter after grouping | Summarize into a table, then filter |
| Keep every row | `.transform("mean")` |
| Rank within group | `.rank(method="dense", ascending=False)` |
| Cumulative / previous row / row number | `.cumsum()` / `.shift(1)` / `.cumcount()`; sort first |
