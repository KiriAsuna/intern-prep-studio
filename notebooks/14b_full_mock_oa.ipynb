```python
def compress(s):
    if not s:
        return ""
    out = []
    cur, count = s[0], 1
    for ch in s[1:]:
        if ch == cur:
            count += 1
        else:
            out.append(f"{cur}{count}")
            cur, count = ch, 1
    out.append(f"{cur}{count}")
    return "".join(out)
```

What it tests: keeping track of "the current character and its count" while iterating; remembering to append the last run after the loop; empty input. Note that `aabbaa` gives `a2b2a2`, so you can't use Counter (it would merge the two runs of a).

<summary><b>Python 2 answer: Longest consecutive sequence</b></summary>

```python
def longest_consecutive(nums):
    s = set(nums)
    best = 0
    for x in s:
        if x - 1 not in s:              # x is the start of a consecutive run
            length = 1
            while x + length in s:
                length += 1
            best = max(best, length)
    return best
```

What it tests: using a set for O(1) lookups. The key is to **count only from the start of each run** (`x` is a start only if `x - 1` is not in the set), so each number is visited at most twice, for O(n) overall. Counting forward from every number is O(n²) in the worst case.

</details>

```sql
SELECT u.channel,
       COUNT(*)      AS n_orders,
       AVG(o.amount) AS avg_amount
FROM orders o
JOIN users u
  ON o.user_id = u.user_id
WHERE o.status = 'completed'
GROUP BY u.channel
ORDER BY avg_amount DESC;
```

What it tests: JOIN + WHERE + GROUP BY + ORDER BY (03b, 03c).

```sql
WITH ranked AS (
  SELECT user_id, order_id, order_date,
         ROW_NUMBER() OVER (PARTITION BY user_id ORDER BY order_date) AS rn
  FROM orders
)
SELECT user_id, order_id, order_date
FROM ranked
WHERE rn = 2;
```

What it tests: window functions + CTEs (03e). "The Nth per group" is `ROW_NUMBER() ... = N`.
