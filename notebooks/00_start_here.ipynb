## 4. R and Python: the big picture

You already know R, so most concepts don't need to be relearned; only the syntax changes. No need to memorize this table; each lesson expands on it.

| Task | R | Python |
|---|---|---|
| Assignment | `x <- 3` | `x = 3` |
| A group of numbers | `c(1, 2, 3)` | `[1, 2, 3]` |
| First element | `x[1]` | `x[0]` ⚠️ |
| Length | `length(x)` | `len(x)` |
| Logical values | `TRUE` / `FALSE` | `True` / `False` |
| Null value | `NULL` | `None` |
| And / or / not | `&` `\|` `!` | `and` `or` `not` |
| Power | `2^10` | `2 ** 10` |
| Code block | `{ ... }` | colon + indentation |
| Function | `f <- function(x) x + 1` | `def f(x): return x + 1` |
| Named list | `list(a = 1)` | `{"a": 1}` (dict) |
| Build a string | `paste0("n=", n)` | `f"n={n}"` |
| Data frame | `data.frame` / dplyr | pandas (Chapter 02) |

The three things that trip up R users most often (you'll be reminded repeatedly):

1. **Indexing starts at 0**, and slices exclude the right end;
2. **Functions need an explicit `return`**; without it they return `None`;
3. **`b = a` does not copy a list**; changing `b` changes `a` too.
