## 1. Recursion

A recursive function has two parts:

1. **Base case**: the problem is small enough to answer directly, with no further calls;
2. **Recursive step**: shrink the problem a little, hand it to the function itself, and build the answer from what comes back.

When writing recursion, don't try to unroll every level in your head. Trust one thing: **"assume the function already works for smaller inputs"**, then get the current level right.

````python
def factorial(n):
    if n <= 1:                       # base case
        return 1
    return n * factorial(n - 1)      # recursive step: n! = n × (n-1)!

print(factorial(5))
````

If you forget the base case, or the problem never gets smaller, the function calls itself forever and Python raises a `RecursionError`.

## 2. Classes: defining your own kind of data

So far you have only used Python's built-in types. A **class** lets you define your own type. You need to recognize classes in interviews because linked-list and tree problems always start by giving you a class definition.

```python
class ListNode:
    def __init__(self, val, next=None):
        self.val = val
        self.next = next
```

- `class ListNode:` defines a type called `ListNode`;
- `__init__` is a function that runs automatically when an object is created; `self` means "the object being created";
- `self.val = val` attaches an attribute named `val` to the object;
- `ListNode(3)` creates an object; `node.val` and `node.next` read and write its attributes.

Think of it like R's `list(val = 3, nxt = NULL)`, except that Python objects are passed by reference (the "aliasing" from 01g).

## 3. Linked lists

A linked list is a chain of nodes: each node stores a value and points to the next node via `next`; the last node's `next` is `None`.

```
1 → 2 → 3 → None
```

Unlike a Python list, a linked list **has no indexing**: to reach the k-th node you must start at the head (`head`) and follow `next` one step at a time.

Below we define the node class plus two helpers for converting between Python lists and linked lists (you won't have to write these in an interview):

````python
class ListNode:
    def __init__(self, val, next=None):
        self.val = val
        self.next = next

def build_list(values):
    # turn a Python list into a linked list and return the head node
    head = None
    for v in reversed(values):
        head = ListNode(v, head)
    return head

def to_list(head):
    # turn a linked list back into a Python list: the standard way to "traverse a linked list"
    out = []
    cur = head
    while cur is not None:
        out.append(cur.val)
        cur = cur.next
    return out

head = build_list([1, 2, 3])
print(head.val, head.next.val, head.next.next.val, head.next.next.next)
print(to_list(head))
````

### Reversing a linked list (the most common linked-list problem)

Turn `1 → 2 → 3 → None` into `3 → 2 → 1 → None`.

Approach: walk from head to tail and point each node's `next` at the node **before** it. You need three variables:

- `prev`: the previous node (starts as `None`);
- `cur`: the current node;
- `nxt`: save the next node first, otherwise you lose it once you change `cur.next`.

Each step does four things: save next → flip the pointer → advance `prev` → advance `cur`.

            a = a.next
        else:

            b = b.next
