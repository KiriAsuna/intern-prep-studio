### Problem 1 · Select columns and add a computed one

Each order's `order_id` and `amount`, plus a column `discounted` with the amount after a 10% discount. Show only the first 5 rows.

```sql
SELECT order_id, amount, amount * 0.9 AS discounted
FROM orders
LIMIT 5;
```

**Result:**

|   | order_id | amount | discounted |
|---|---|---|---|
| 1 | 1001 | 30.0 | 27.0 |
| 2 | 1002 | 50.0 | 45.0 |
| 3 | 1003 | 20.0 | 18.0 |
| 4 | 1004 | 15.0 | 13.5 |
| 5 | 1005 | 15.0 | 13.5 |

*5 rows × 3 columns*

> `AS` gives an alias, like `df["discounted"] = df["amount"] * 0.9` in pandas.

### Problem 2 · Multiple conditions

Completed orders with an amount between 30 and 60 (inclusive).

```sql
SELECT order_id, user_id, amount
FROM orders
WHERE status = 'completed'
  AND amount BETWEEN 30 AND 60;
```

**Result:**

|   | order_id | user_id | amount |
|---|---|---|---|
| 1 | 1001 | 1 | 30.0 |
| 2 | 1002 | 1 | 50.0 |
| 3 | 1007 | 3 | 40.0 |
| 4 | 1008 | 3 | 60.0 |
| 5 | 1012 | 6 | 45.0 |
| 6 | 1013 | 6 | 55.0 |

*6 rows × 3 columns*

> `BETWEEN` includes both ends. Equality in SQL is a single `=`, and strings use single quotes.

### Problem 3 · Matching a set of values: IN

Users from `CN` or `CA`.

```sql
SELECT user_id, name, country
FROM users
WHERE country IN ('CN', 'CA');
```

**Result:**

|   | user_id | name | country |
|---|---|---|---|
| 1 | 3 | Chen | CN |
| 2 | 5 | Eli | CA |
| 3 | 6 | Fay | CN |
| 4 | 9 | Ivan | CA |
| 5 | 10 | Jia | CN |

*5 rows × 3 columns*

> pandas `isin`; R `%in%`.

### Problem 4 · Pattern matching: LIKE

Users whose name ends with the letter `a`.

```sql
SELECT user_id, name
FROM users
WHERE name LIKE '%a';
```

**Result:**

|   | user_id | name |
|---|---|---|
| 1 | 4 | Dana |
| 2 | 8 | Hana |
| 3 | 10 | Jia |

*3 rows × 2 columns*

> `%` matches any number of characters: `'A%'` starts with A, `'%a'` ends with a, `'%an%'` contains an. SQLite's `LIKE` is case-insensitive; PostgreSQL's is case-sensitive (use `ILIKE` for case-insensitive matching).

### Problem 5 · The NULL trap

Users whose country is **not** `CN`, including those with a missing country.

```sql
SELECT user_id, name, country
FROM users
WHERE country <> 'CN'
   OR country IS NULL;
```

**Result:**

|   | user_id | name | country |
|---|---|---|---|
| 1 | 1 | Ann | US |
| 2 | 2 | Bob | US |
| 3 | 4 | Dana | US |
| 4 | 5 | Eli | CA |
| 5 | 7 | Gus | NULL |
| 6 | 8 | Hana | US |
| 7 | 9 | Ivan | CA |

*7 rows × 3 columns*

> Writing only `country <> 'CN'` drops Gus: a comparison with NULL is never true. Note this is the opposite of pandas, where `NaN != "CN"` is True.

### Problem 6 · Multiple sort keys

Sort employees by department alphabetically, then by salary from high to low within a department, then by name when salaries tie.

```sql
SELECT name, dept, salary
FROM employees
ORDER BY dept, salary DESC, name;
```

**Result:**

|   | name | dept | salary |
|---|---|---|---|
| 1 | Alice | Eng | 150 |
| 2 | Brian | Eng | 120 |
| 3 | Cathy | Eng | 120 |
| 4 | David | Eng | 95 |
| 5 | Henry | Mkt | 85 |
| 6 | Irene | Mkt | 60 |
| 7 | Emma | Sales | 90 |
| 8 | Frank | Sales | 70 |
| 9 | Grace | Sales | 70 |

*9 rows × 3 columns*

> `DESC` applies only to the one column right before it.

### Problem 7 · Top N

The 3 orders with the highest amounts (any status); for equal amounts, the smaller `order_id` comes first.

```sql
SELECT order_id, amount
FROM orders
ORDER BY amount DESC, order_id
LIMIT 3;
```

**Result:**

|   | order_id | amount |
|---|---|---|
| 1 | 1010 | 100.0 |
| 2 | 1006 | 80.0 |
| 3 | 1014 | 70.0 |

*3 rows × 2 columns*

> There are two orders of 70, so the secondary sort key is what makes the result deterministic.

### Problem 8 · Removing duplicates: DISTINCT

Which statuses appear in the orders table.

```sql
SELECT DISTINCT status
FROM orders;
```

**Result:**

|   | status |
|---|---|
| 1 | completed |
| 2 | refunded |

*2 rows × 1 column*

> `SELECT DISTINCT a, b` deduplicates (a, b) combinations, not just a.

### Problem 9 · Three ways to COUNT

The number of rows in `users`, the number of rows with a country, and the number of distinct countries.

```sql
SELECT COUNT(*)                AS n_rows,
       COUNT(country)          AS n_with_country,
       COUNT(DISTINCT country) AS n_countries
FROM users;
```

**Result:**

|   | n_rows | n_with_country | n_countries |
|---|---|---|---|
| 1 | 10 | 9 | 3 |

*1 row × 3 columns*

> `COUNT(col)` skips NULLs, and `COUNT(DISTINCT col)` does not count NULL as a value either. A common interview question.

### Problem 10 · Summarizing the whole table

For completed orders: the count, total, average, minimum, and maximum amount.

```sql
SELECT COUNT(*)    AS n,
       SUM(amount) AS total,
       AVG(amount) AS avg_amount,
       MIN(amount) AS min_amount,
       MAX(amount) AS max_amount
FROM orders
WHERE status = 'completed';
```

**Result:**

|   | n | total | avg_amount | min_amount | max_amount |
|---|---|---|---|---|---|
| 1 | 14 | 665.0 | 47.5 | 10.0 | 100.0 |

*1 row × 5 columns*

> Without `GROUP BY`, the whole table collapses into one row.

### Problem 11 · Counting with GROUP BY

The number of users in each channel.

```sql
SELECT channel, COUNT(*) AS n_users
FROM users
GROUP BY channel;
```

**Result:**

|   | channel | n_users |
|---|---|---|
| 1 | ads | 4 |
| 2 | organic | 4 |
| 3 | referral | 2 |

*3 rows × 2 columns*

> pandas: `users.groupby("channel").size()`; R: `count(users, channel)`.

### Problem 12 · Grouping by two columns

The number of users for each (country, channel) combination, in descending order of count.

```sql
SELECT country, channel, COUNT(*) AS n_users
FROM users
GROUP BY country, channel
ORDER BY n_users DESC, country, channel;
```

**Result:**

|   | country | channel | n_users |
|---|---|---|---|
| 1 | CN | ads | 2 |
| 2 | US | referral | 2 |
| 3 | NULL | organic | 1 |
| 4 | CA | ads | 1 |
| 5 | CA | organic | 1 |
| 6 | CN | organic | 1 |
| 7 | US | ads | 1 |
| 8 | US | organic | 1 |

*8 rows × 3 columns*

> NULL forms its own group in `GROUP BY`. SQLite sorts NULL first; PostgreSQL puts it last in ascending order.

### Problem 13 · Filter rows, then group

Each user's total amount of completed orders, from high to low.

```sql
SELECT user_id, SUM(amount) AS total
FROM orders
WHERE status = 'completed'
GROUP BY user_id
ORDER BY total DESC, user_id;
```

**Result:**

|   | user_id | total |
|---|---|---|
| 1 | 3 | 180.0 |
| 2 | 8 | 140.0 |
| 3 | 5 | 100.0 |
| 4 | 6 | 100.0 |
| 5 | 1 | 80.0 |
| 6 | 2 | 30.0 |
| 7 | 4 | 25.0 |
| 8 | 10 | 10.0 |

*8 rows × 2 columns*

> `ORDER BY` can use aliases defined in `SELECT` because it runs after `SELECT`. Users 5 and 6 both have 100, so adding `user_id` as a secondary sort key makes the order deterministic.

### Problem 14 · Filtering groups with HAVING

Departments with an average salary of at least 90.

```sql
SELECT dept, AVG(salary) AS avg_salary
FROM employees
GROUP BY dept
HAVING AVG(salary) >= 90;
```

**Result:**

|   | dept | avg_salary |
|---|---|---|
| 1 | Eng | 121.25 |

*1 row × 2 columns*

> Use `HAVING` to filter groups; aggregate functions are not allowed in `WHERE` (grouping hasn't happened yet when `WHERE` runs).

### Problem 15 · WHERE and HAVING together

Users with at least 2 completed orders, and how many each has.

```sql
SELECT user_id, COUNT(*) AS n_completed
FROM orders
WHERE status = 'completed'
GROUP BY user_id
HAVING COUNT(*) >= 2
ORDER BY user_id;
```

**Result:**

|   | user_id | n_completed |
|---|---|---|
| 1 | 1 | 2 |
| 2 | 2 | 2 |
| 3 | 3 | 3 |
| 4 | 6 | 2 |
| 5 | 8 | 2 |

*5 rows × 2 columns*

> Execution order: FROM → WHERE → GROUP BY → HAVING → SELECT → ORDER BY → LIMIT.

### Problem 16 · A new column with CASE WHEN

Bucket employees: salary 100 and above is `'high'`, 80 to 99 is `'mid'`, everything else is `'low'`.

```sql
SELECT name, salary,
       CASE
         WHEN salary >= 100 THEN 'high'
         WHEN salary >= 80  THEN 'mid'
         ELSE 'low'
       END AS level
FROM employees;
```

**Result:**

|   | name | salary | level |
|---|---|---|---|
| 1 | Alice | 150 | high |
| 2 | Brian | 120 | high |
| 3 | Cathy | 120 | high |
| 4 | David | 95 | mid |
| 5 | Emma | 90 | mid |
| 6 | Frank | 70 | low |
| 7 | Grace | 70 | low |
| 8 | Henry | 85 | mid |
| 9 | Irene | 60 | low |

*9 rows × 3 columns*

> The first true condition from the top wins, so the second condition doesn't need `< 100`. Equivalent to R's `case_when`.

### Problem 17 · Grouping by a CASE WHEN

How many people fall into each of the three buckets from the previous problem.

```sql
SELECT CASE
         WHEN salary >= 100 THEN 'high'
         WHEN salary >= 80  THEN 'mid'
         ELSE 'low'
       END AS level,
       COUNT(*) AS n
FROM employees
GROUP BY level;
```

**Result:**

|   | level | n |
|---|---|---|
| 1 | high | 3 |
| 2 | low | 3 |
| 3 | mid | 3 |

*3 rows × 2 columns*

> SQLite, MySQL, and PostgreSQL all allow `GROUP BY` on an alias; SQL Server does not, so you have to repeat the whole `CASE ... END`.

### Problem 18 · Conditional aggregation: counting matching rows

Each user's number of orders and how many of them were refunded.

```sql
SELECT user_id,
       COUNT(*)                                             AS n_orders,
       SUM(CASE WHEN status = 'refunded' THEN 1 ELSE 0 END) AS n_refunded
FROM orders
GROUP BY user_id;
```

**Result:**

|   | user_id | n_orders | n_refunded |
|---|---|---|---|
| 1 | 1 | 3 | 1 |
| 2 | 2 | 2 | 0 |
| 3 | 3 | 3 | 0 |
| 4 | 4 | 1 | 0 |
| 5 | 5 | 2 | 1 |
| 6 | 6 | 2 | 0 |
| 7 | 8 | 2 | 0 |
| 8 | 10 | 1 | 0 |

*8 rows × 3 columns*

> Score 1 when the condition holds and 0 otherwise, then sum. In pandas, this is `sum()` on a boolean column.

### Problem 19 · Conditional aggregation: computing a share

The share of employees in each department with a salary of 100 or more.

```sql
SELECT dept,
       AVG(CASE WHEN salary >= 100 THEN 1.0 ELSE 0 END) AS share_high
FROM employees
GROUP BY dept;
```

**Result:**

|   | dept | share_high |
|---|---|---|
| 1 | Eng | 0.75 |
| 2 | Mkt | 0.0 |
| 3 | Sales | 0.0 |

*3 rows × 2 columns*

> The `AVG` of a 0/1 column is a proportion. Writing `1.0` is safer: SQLite, MySQL, and PostgreSQL return a decimal for `AVG` of integers, but SQL Server returns an integer (which would be 0 here).

### Problem 20 · The integer division pitfall

For each user, the share of their orders that were completed.

```sql
SELECT user_id,
       1.0 * SUM(CASE WHEN status = 'completed' THEN 1 ELSE 0 END) / COUNT(*) AS completed_rate
FROM orders
GROUP BY user_id;
```

**Result:**

|   | user_id | completed_rate |
|---|---|---|
| 1 | 1 | 0.6667 |
| 2 | 2 | 1.0 |
| 3 | 3 | 1.0 |
| 4 | 4 | 1.0 |
| 5 | 5 | 0.5 |
| 6 | 6 | 1.0 |
| 7 | 8 | 1.0 |
| 8 | 10 | 1.0 |

*8 rows × 2 columns*

> Without `1.0 *`, Ann's 2 / 3 evaluates to 0 in SQLite and PostgreSQL. MySQL's `/` always does decimal division, but writing `1.0 *` never hurts.

## Syntax recap

| Goal | Syntax |
|---|---|
| Select / new column | `SELECT a, b * 2 AS c` |
| Filter rows | `WHERE ... AND / OR`, `IN (...)`, `BETWEEN x AND y`, `LIKE 'A%'` |
| Missing values | `IS NULL` / `IS NOT NULL`; `<>` never matches NULL |
| Sort / top N | `ORDER BY a, b DESC LIMIT n` |
| Count | `COUNT(*)`, `COUNT(col)`, `COUNT(DISTINCT col)` |
| Group / filter groups | `GROUP BY g HAVING <aggregate condition>` |
| Bucket | `CASE WHEN ... THEN ... ELSE ... END` |
| Conditional count / share | `SUM(CASE WHEN cond THEN 1 ELSE 0 END)` / `AVG(CASE WHEN cond THEN 1.0 ELSE 0 END)` |
