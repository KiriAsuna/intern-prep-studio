### Problem 1 · Attach the group average to every row

Show each employee alongside their department's average salary.

```sql
SELECT name, dept, salary,
       AVG(salary) OVER (PARTITION BY dept) AS dept_avg
FROM employees;
```

**Result:**

|   | name | dept | salary | dept_avg |
|---|---|---|---|---|
| 1 | Alice | Eng | 150 | 121.25 |
| 2 | Brian | Eng | 120 | 121.25 |
| 3 | Cathy | Eng | 120 | 121.25 |
| 4 | David | Eng | 95 | 121.25 |
| 5 | Henry | Mkt | 85 | 72.5 |
| 6 | Irene | Mkt | 60 | 72.5 |
| 7 | Emma | Sales | 90 | 76.6667 |
| 8 | Frank | Sales | 70 | 76.6667 |
| 9 | Grace | Sales | 70 | 76.6667 |

*9 rows × 4 columns*

> Window functions don't collapse rows; this is pandas' `groupby("dept")["salary"].transform("mean")`.

### Problem 3 · The three rankings side by side

Rank the whole company by salary from high to low, with one column for each ranking function.

```sql
SELECT name, salary,
       ROW_NUMBER() OVER (ORDER BY salary DESC) AS row_num,
       RANK()       OVER (ORDER BY salary DESC) AS rnk,
       DENSE_RANK() OVER (ORDER BY salary DESC) AS dense_rnk
FROM employees
ORDER BY salary DESC, name;
```

**Result:**

|   | name | salary | row_num | rnk | dense_rnk |
|---|---|---|---|---|---|
| 1 | Alice | 150 | 1 | 1 | 1 |
| 2 | Brian | 120 | 2 | 2 | 2 |
| 3 | Cathy | 120 | 3 | 2 | 2 |
| 4 | David | 95 | 4 | 4 | 3 |
| 5 | Emma | 90 | 5 | 5 | 4 |
| 6 | Henry | 85 | 6 | 6 | 5 |
| 7 | Frank | 70 | 7 | 7 | 6 |
| 8 | Grace | 70 | 8 | 7 | 6 |
| 9 | Irene | 60 | 9 | 9 | 7 |

*9 rows × 5 columns*

> Look at the two ties, 120/120 and 70/70: `RANK` skips numbers after a tie, `DENSE_RANK` does not. With ties, the order `ROW_NUMBER` assigns is not guaranteed; to make it deterministic, add a secondary key to `ORDER BY`.

### Problem 4 · Top 1 per group (exactly one)

The single highest-paid person in each department; on a tie, take the name that comes first alphabetically.

```sql
WITH ranked AS (
  SELECT dept, name, salary,
         ROW_NUMBER() OVER (PARTITION BY dept ORDER BY salary DESC, name) AS rn
  FROM employees
)
SELECT dept, name, salary
FROM ranked
WHERE rn = 1;
```

**Result:**

|   | dept | name | salary |
|---|---|---|---|
| 1 | Eng | Alice | 150 |
| 2 | Mkt | Henry | 85 |
| 3 | Sales | Emma | 90 |

*3 rows × 3 columns*

> Window functions can't go directly in `WHERE`; compute them in a CTE first, then filter in the outer query.

### Problem 5 · Top two salary levels per group (keep ties)

The highest and second-highest salary levels in each department, keeping everyone who ties.

```sql
WITH ranked AS (
  SELECT dept, name, salary,
         DENSE_RANK() OVER (PARTITION BY dept ORDER BY salary DESC) AS rk
  FROM employees
)
SELECT dept, name, salary
FROM ranked
WHERE rk <= 2;
```

**Result:**

|   | dept | name | salary |
|---|---|---|---|
| 1 | Eng | Alice | 150 |
| 2 | Eng | Brian | 120 |
| 3 | Eng | Cathy | 120 |
| 4 | Mkt | Henry | 85 |
| 5 | Mkt | Irene | 60 |
| 6 | Sales | Emma | 90 |
| 7 | Sales | Frank | 70 |
| 8 | Sales | Grace | 70 |

*8 rows × 3 columns*

> "Keep ties, no gaps" means `DENSE_RANK`. When you get a question like this, first ask how ties should be handled.

### Problem 6 · Nth-highest salary

The 3rd-highest salary in the company (ties count as one).

```sql
WITH ranked AS (
  SELECT DISTINCT salary,
         DENSE_RANK() OVER (ORDER BY salary DESC) AS rk
  FROM employees
)
SELECT salary AS third_salary
FROM ranked
WHERE rk = 3;
```

**Result:**

|   | third_salary |
|---|---|
| 1 | 95 |

*1 row × 1 column*

> The salary levels are 150, 120, 95, ..., so the 3rd is 95. Replace 3 with N for the general version.

### Problem 14 · LAG: the previous order's date

### Problem 18 · Order sequence number + first-order flag

For each order, which number it is for that user, plus a flag for whether it's the first.

### Problem 20 · Aggregate first, then LAG

Each month's completed order total and the previous month's total.

```sql
WITH monthly AS (
  SELECT strftime('%Y-%m', order_date) AS month, SUM(amount) AS total
  FROM orders
  WHERE status = 'completed'
  GROUP BY month
)
SELECT month, total,
       LAG(total) OVER (ORDER BY month) AS prev_total
FROM monthly;
```

**Result:**

|   | month | total | prev_total |
|---|---|---|---|
| 1 | 2026-01 | 480.0 | NULL |
| 2 | 2026-02 | 185.0 | 480.0 |

*2 rows × 3 columns*

> Window functions run after `GROUP BY`, so "aggregate by month, then take the previous month" has to be split into two steps.

## Syntax recap

| Goal | Syntax |
|---|---|
| Group stat on every row | `AVG(x) OVER (PARTITION BY g)` |
| Ranking | `ROW_NUMBER()` / `RANK()` / `DENSE_RANK()` `OVER (PARTITION BY g ORDER BY x DESC)` |
| Top N per group | Compute the rank in a CTE, then `WHERE rn <= N` outside |
| Share | `1.0 * x / SUM(x) OVER (PARTITION BY g)`; use `OVER ()` for the whole table |
| Running total | `SUM(x) OVER (PARTITION BY g ORDER BY t)` |
| Moving average | `AVG(x) OVER (ORDER BY t ROWS BETWEEN 2 PRECEDING AND CURRENT ROW)` |
| Previous / next row | `LAG(x)` / `LEAD(x) OVER (PARTITION BY g ORDER BY t)` |
| Group row count | `COUNT(*) OVER (PARTITION BY g)` |
