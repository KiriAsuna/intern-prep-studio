**1.** What are the assumptions of OLS?

<details>
<summary><b>Key points</b></summary>

Answer in layers. Unbiasedness: linearity, full rank, zero conditional mean. Gauss–Markov (BLUE): add homoskedasticity and uncorrelated errors. Exact small-sample inference: add normality. Normality is not required for unbiasedness. → 07a

</details>

**2.** Why is the zero-conditional-mean assumption the most important? What happens when it is violated?

<details>
<summary><b>Key points</b></summary>

It guarantees unbiased coefficients. When it is violated (omitted variables, reverse causality, measurement error), the coefficients are biased and inconsistent, and a larger sample doesn't help. Violations of the other assumptions usually affect only the standard errors, which robust standard errors can fix. → 07a

</details>

**3.** What problems does heteroskedasticity cause? How do you handle it?

<details>
<summary><b>Key points</b></summary>

The coefficients are still unbiased, but the classical standard errors are wrong, inference breaks down, and OLS is no longer the most efficient. Use robust (HC) standard errors, transform Y, or use WLS. → 07a

</details>

**4.** Explain the relationship among bias, variance, and MSE.

<details>
<summary><b>Key points</b></summary>

MSE = variance + bias². Unbiased isn't necessarily best; regularization trades a little bias for lower variance. → 06a

</details>

**5.** What is the difference between a standard deviation and a standard error?

<details>
<summary><b>Key points</b></summary>

The SD describes the spread of the data; the SE is the standard deviation of an estimator's sampling distribution and describes the uncertainty of the estimate. The SE of the mean = σ/√n, which shrinks as the sample size grows. → 06a

</details>

**6.** How do you interpret a 95% confidence interval?

<details>
<summary><b>Key points</b></summary>

If you repeated the sampling and built intervals the same way, in the long run 95% of the intervals would contain the true parameter. The 95% is the coverage rate of the method, not the probability that this particular interval contains the parameter. → 06a

</details>

**7.** What is a p-value?

<details>
<summary><b>Key points</b></summary>

Assuming H₀ is true, the probability of observing a result at least as extreme as the one observed. It is not the probability that H₀ is true, and it does not reflect the effect size. → 06a

</details>

**8.** How are Type I error, Type II error, and power related? What determines power?

<details>
<summary><b>Key points</b></summary>

Type I: rejecting H₀ when it is true (α). Type II: failing to reject H₀ when it is false (β). Power = 1 − β. It is determined by the effect size, sample size, variance, and α. → 06a

</details>

**9.** What do the law of large numbers and the central limit theorem each say?

<details>
<summary><b>Key points</b></summary>

LLN: the sample mean converges to the population mean. CLT: the distribution of the standardized sample mean tends to a normal distribution. What becomes normal is the sampling distribution of the mean, not the data themselves. → 05a

</details>

**10.** Explain Bayes' theorem and base rates with an example.

<details>
<summary><b>Key points</b></summary>

Posterior ∝ likelihood × prior. With 1% prevalence and a very accurate test, the probability that someone who tests positive actually has the disease is still only about 17%, because the healthy population is large and produces many false positives. → 05a

</details>

**11.** What is the difference between the bootstrap and a permutation test?

<details>
<summary><b>Key points</b></summary>

The bootstrap resamples with replacement to approximate an estimator's sampling distribution, giving SEs and CIs. A permutation test shuffles labels to construct the null distribution, giving a p-value. → 06a

</details>

**12.** How do you interpret logistic regression coefficients?

<details>
<summary><b>Key points</b></summary>

A one-unit increase in x increases the log-odds by β and multiplies the odds by e^β. It does not increase the probability by β; the effect on the probability depends on the baseline. → 08a

</details>

**13.** What is the difference between L1 and L2 regularization?

<details>
<summary><b>Key points</b></summary>

L2 shrinks all coefficients smoothly; L1 can push coefficients to exactly 0, performing variable selection. Geometrically, the L1 constraint region has corners. Standardize before using either. → 08a

</details>

**14.** How do you detect and deal with overfitting?

<details>
<summary><b>Key points</b></summary>

Training performance is far better than validation performance. Remedies: more data, a simpler model, regularization, early stopping, ensembles, feature selection; and check for leakage. → 10a

</details>

**15.** What is the difference between a random forest and gradient boosting?

<details>
<summary><b>Key points</b></summary>

RF: deep trees trained in parallel and averaged, mainly reducing variance. Boosting: shallow trees built sequentially, each fitting the residuals of the previous ones, mainly reducing bias; often more accurate but needs more tuning. → 10a

</details>

**16.** What is the difference between ROC-AUC and PR-AUC? When do you use which?

<details>
<summary><b>Key points</b></summary>

ROC-AUC is the probability that a positive is ranked above a negative, and it is insensitive to class proportions. When the positive class is rare and it is what you care about, PR-AUC is more informative. → 08a, 11a

</details>

**17.** What is data leakage? Give examples.

<details>
<summary><b>Key points</b></summary>

Training uses information not available at prediction time. Examples: a feature that is a consequence of the target; standardizing before splitting; randomly splitting a time series; the same user in both training and test sets. → 10a

</details>

**18.** How does cross-validation work? What should you watch out for?

<details>
<summary><b>Key points</b></summary>

Rotate through K folds as the validation set and average. Fit preprocessing within each fold (Pipeline); split time-ordered data by time; put all of a user's data in the same fold; report final performance on an independent test set. → 10a

</details>

**19.** How do you design an A/B test?

<details>
<summary><b>Key points</b></summary>

Goal and hypothesis → primary and guardrail metrics → randomization unit → sample size and duration (MDE, α, power) → gradual rollout → sanity checks such as SRM → estimate the effect and CI → handle multiple testing and peeking → decide in light of practical significance. → 09a

</details>

**20.** A core metric suddenly dropped. How do you investigate?

<details>
<summary><b>Key points</b></summary>

First confirm it's real (definition, logging, numerator and denominator) → locate it by time and segment (platform, region, new vs. returning users, channel, funnel) → find the cause by checking internal changes and external factors → form hypotheses, test, quantify, and recommend action. → 12

</details>
