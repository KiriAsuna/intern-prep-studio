# 12b | Drills: DS Case Questions

Covers **12**: investigating metric changes, designing metrics, interpreting experiment results and making launch decisions, evaluating without an experiment, and estimation questions. All 15 are oral questions.

Every reference answer follows the same order: **Clarify → Decompose → Hypothesize → Data / experiment → Conclude**. Say these five words at the start of your answer and the interviewer will know you have structure. The answers are key points, not the one right answer; in an interview, build on the information you're given.

**How to use this lesson:** There are no input boxes; each problem is followed directly by its reference answer. First answer **out loud** in your own words (3–5 minutes per question), then compare with the reference answer and see which step you missed.

### Problem 1 · DAU dropped 🗣

A social app's daily active users (DAU) are 5% lower this week than last week. How do you investigate?

**Reference answer:**

- **Clarify:** how is DAU defined (opened the app? any action?); which days are being compared to last week; is it a sudden drop or a day-by-day decline; is it within normal weekly fluctuation?
- **Decompose:** DAU = newly active users + returning users. Split by platform, app version, country, new vs. returning users, and acquisition channel to see whether the drop is across the board or concentrated in one group.
- **Hypothesize:** ① a logging or data-pipeline problem (some version stopped reporting); ② a release introduced a bug (the drop is only on one platform or version); ③ fewer push notifications / less ad spend; ④ holidays or seasonality; ⑤ a competitor's campaign or an external event.
- **Data:** compare against the release log, the experimentation platform, and the marketing calendar; check whether reported event volume matches server-side logs; look at the same period last year.
- **Conclude:** give the one or two most likely causes and how much of the drop each explains, with the corresponding action (fix / roll back / restore spend / judge it normal fluctuation and keep watching).

### Problem 2 · GMV up but orders down 🗣

An e-commerce platform's GMV rose 10% this month, but the number of orders fell 5%. How do you interpret this?

**Reference answer:**

- **Clarify:** does GMV include refunds and shipping; do the two numbers cover the same time window?
- **Decompose:** GMV = orders × average order value, so average order value rose about 16% (1.10 / 0.95). Then break average order value into price per item × items per order.
- **Hypothesize:** ① category mix shifted (a larger share of big-ticket items, e.g., an appliance promotion); ② price increases; ③ a higher free-shipping or discount threshold, so users add items to reach it; ④ a few large customers / corporate purchases; ⑤ users who placed low-value orders churned.
- **Data:** split average order value and order count by category, price band, new vs. returning users, and city; see which users the order decline is concentrated in; check whether the number of users also fell.
- **Conclude:** if low-value users are churning, GMV looks good in the short term but carries long-term risk and needs separate follow-up; if it's a structural promotion effect, it may fall back next month. Never report GMV alone.

### Problem 3 · Down overall, up in every group 🗣

Overall conversion fell, but when split by channel, every channel's conversion rate rose. How is that possible? How do you report it?

**Reference answer:**

- **Clarify:** did the traffic share of each channel change over this period?
- **Decompose:** overall conversion = Σ channel share × channel conversion rate. If every channel rises while the overall rate falls, the traffic mix shifted toward low-converting channels (Simpson's paradox).
- **Hypothesize:** spend increased on a low-converting channel (e.g., display ads); traffic from a high-converting channel (e.g., organic search) decreased.
- **Data:** decompose the change into a "mix effect" (change in shares) and a "within-channel effect" (change in conversion rates), and quantify each.
- **Conclude:** the product's ability to convert is actually improving; the decline comes from the traffic mix. Report both numbers together and discuss whether the acquisition strategy pays off (whether the incremental users from the low-converting channel are worth its cost).

### Problem 4 · New users surge, retention falls 🗣

New users doubled last month, but next-month retention fell from 40% to 30%. Did the product get worse?

**Reference answer:**

- **Clarify:** which users is retention computed on (the new-user cohort? everyone?); where did the new users come from?
- **Decompose:** split cohorts by acquisition channel / campaign and check whether retention of "new users from existing channels" changed.
- **Hypothesize:** ① most of the additions came from an acquisition campaign or low-intent channels, whose users naturally retain worse (a mix change); ② the surge of new users degraded service (customer support, insufficient supply); ③ a product change in the same period genuinely hurt the experience.
- **Data:** if retention of existing-channel cohorts is still 40%, it's a mix issue; if existing channels also dropped, then investigate product and experience metrics.
- **Conclude:** you can't simply say "the product got worse." Recommend reporting retention by channel and evaluating the acquisition campaign's ROI (accounting for the long-term value of low-retention users).

### Problem 5 · Success metrics for search 🗣

An e-commerce app redesigned its search feature. What metrics would you use to measure whether it succeeded?

**Reference answer:**

- **Clarify:** what is the redesign's goal (more relevant results? faster? better long-tail coverage?); is it for all users or one market?
- **Decompose:** the search funnel: start a search → see results → click → add to cart → purchase.
- **Metrics:** primary: orders per user from search (or search conversion rate). Secondary: zero-result rate, position of the first click, click-through rate, share of reformulated queries (users didn't find what they wanted), time from search to purchase.
- **Guardrails:** overall GMV, page latency, return rate (whether what users found is what they actually wanted).
- **Experiment:** a user-level A/B test run for at least one to two full weeks to cover weekends; beware of differences between searchers and non-searchers — use all users as the denominator of the primary metric to avoid selection bias.
- **Conclude:** ship only if the primary metric improves significantly and guardrails don't worsen; also look at results by query type (head vs. long-tail queries).

### Problem 6 · A delivery-experience metric for a food-delivery platform 🗣

Design a metric that measures "delivery experience" for a food-delivery platform.

**Reference answer:**

- **Clarify:** from whose perspective: customers, couriers, or restaurants? Here, focus on customers first.
- **Decompose:** experience = on time + fast + predictable + food arrives intact.
- **Metrics:** north star: on-time rate (share of orders delivered within the promised time). Secondary: average delivery time, deviation between promised and actual time, share of orders more than 15 minutes late (the tail), rate of delivery-related negative reviews.
- **How the metric could be gamed:** if you only look at on-time rate, the platform can loosen promised times to "game" it; so also track the promised time itself and the share of users who abandon orders because the promised time is too long.
- **Guardrails:** couriers' earnings per order and safety (it can't come from pushing couriers to speed), delivery cost.
- **Conclude:** use on-time rate together with promised delivery time as a pair, monitor the late tail separately, and break results down by city, time of day, and weather.

### Problem 7 · Primary metric up, guardrail down 🗣

The treatment group's conversion rate rose a significant 2%, but page load time increased by 300 ms (also significant). Do you ship?

**Reference answer:**

- **Clarify:** what were the pre-specified primary and guardrail metrics; what is the guardrail's tolerance?
- **Decompose:** the conversion gain is the direct effect of the new feature; slower loading is a long-term risk that may not yet show up in retention in a short experiment.
- **Hypothesize:** the long-term effect of 300 ms slower loading on overall experience (historical data or past latency experiments usually provide an estimate of "how much conversion drops per 100 ms of delay").
- **Data:** look at segments: did conversion actually fall for users on low-end devices or weak networks? Look at bounce rate and next-day return.
- **Conclude:** if the guardrail threshold is exceeded, don't ship directly; optimize performance first (lazy loading, caching) and rerun the experiment. If it's within tolerance and no segment is clearly harmed, ship and keep monitoring. The key is to decide by the pre-specified rules rather than picking after seeing the results.

### Problem 8 · p = 0.06 🗣

The experiment's primary metric rose 1.5%, with p = 0.06 (the pre-specified significance level was 0.05). The product manager says, "It's close enough, let's ship." How do you respond?

**Reference answer:**

- **Clarify:** did the experiment reach the pre-computed sample size; was there any peeking; what is the confidence interval?
- **Decompose:** p = 0.06 means the evidence isn't strong enough; it does not mean "no effect." The confidence interval runs roughly from slightly below 0 to about 3%, so the effect could be tiny or could be sizable.
- **Consider:** ① the cost and risk of shipping (is it reversible? maintenance cost?); ② whether any guardrail metric worsened; ③ whether this effect size is plausible compared with similar past experiments.
- **Approach:** don't relax the threshold after the fact. Options: if the sample size hasn't been reached, finish as planned; if it has, rerun with a larger sample, or make an explicit, documented "acceptable risk" decision based on costs and benefits.
- **Conclude:** walk the product manager through the confidence interval and the risk of the decision, so the decision rests on effect size and cost rather than on whether the p-value crossed 0.05.

### Problem 9 · Big jump in week one, then fading 🗣

In a new-homepage experiment, CTR rose 8% in the first week but only 1% by week three. How do you explain it? What is the final conclusion?

**Reference answer:**

- **Clarify:** is the treatment group the same set of users viewing throughout, or are new users continually entering?
- **Decompose:** split users into cohorts by when they entered the experiment and see how each cohort's effect changes over time.
- **Hypothesize:** ① novelty effect: existing users are curious about something new and revert to normal after a few days; ② seasonality or concurrent campaigns; ③ new users keep entering mid-experiment, so the user mix is changing. (The opposite pattern is called the primacy effect / change aversion: existing users need time to adapt, so the effect starts worse and improves. Here it starts high and falls, which looks more like a novelty effect.)
- **Data:** compare new users (who never saw the old version) with existing users; the effect on new users is closer to the long-term effect.
- **Conclude:** use the stabilized effect (about 1%) as the expected benefit, not the 8% from week one, and assess whether that long-term effect is worth shipping. Future experiments on changes like this should run long enough.

### Problem 10 · Opposite effects for new and existing users 🗣

The experiment shows no significant overall difference, but new users got significantly better and existing users significantly worse. How do you handle it?

**Reference answer:**

- **Clarify:** was the new/existing split planned in advance; how many segments were examined (was this significant result found by searching many segments)?
- **Decompose:** overall effect = new-user share × new-user effect + existing-user share × existing-user effect; the positive and negative cancel out.
- **Hypothesize:** the change lowered the learning curve (new users benefit) but disrupted existing users' habits (existing users need to readjust, which may be temporary).
- **Data:** if the split was post hoc, apply a multiple-comparison correction first, or confirm with a new experiment; check whether the negative effect on existing users fades over time.
- **Conclude:** if confirmed, consider launching only to new users, or keeping the old version / adding onboarding for existing users; also think long term: new users will eventually become existing users.

### Problem 11 · Experiments with network effects 🗣

A ride-hailing platform wants to test a new dispatch algorithm. What is wrong with an A/B test randomized by rider? How would you design it?

**Reference answer:**

- **Clarify:** what does the algorithm change (dispatch order? pricing?); would the two groups share the same pool of drivers?
- **The problem:** treatment and control share the driver supply. The treatment group grabs the better drivers, the control group gets worse, and the difference between groups is exaggerated; after a full launch this advantage doesn't exist (it violates the assumption that one group's treatment doesn't affect the other).
- **Design:** ① randomize by city or geographic region (cluster randomization), since regions are relatively independent; ② time-slice rotation (switchback): switch algorithms in the same city every fixed period; ③ a combination of both.
- **The cost:** fewer randomization units means higher variance, requiring a longer run or more cities; compute standard errors at the level of the randomization unit.
- **Conclude:** judge by marketplace-wide primary metrics (total completed trips, wait times, driver earnings), not just the rider side.

### Problem 12 · Already launched to everyone, no control group 🗣

A new feature launched to all users three months ago without an experiment. Now your manager asks whether it worked. How do you evaluate it?

**Reference answer:**

- **Clarify:** what is the feature's target metric; is the launch date clear; were there other changes at the same time?
- **Decompose:** you need a reference for "what would have happened without the launch."
- **Methods:** ① if it rolled out region by region / platform by platform, use difference-in-differences (DiD): compare the before-and-after change of early-launch groups with late-launch groups; ② matched comparison of feature users with similar non-users (propensity scores), while acknowledging self-selection bias; ③ time-series methods: use the pre-launch trend to forecast the "no launch" trajectory and compare it with reality; ④ regression discontinuity when there is an eligibility cutoff.
- **Check assumptions:** for DiD, check whether the groups had parallel trends before launch; run placebo tests (pretend the launch happened at a different time).
- **Conclude:** give the estimate and the assumptions it relies on, and say candidly that it is less reliable than an experiment; if possible, recommend a holdout experiment that turns the feature off for a group to confirm.

### Problem 13 · More frequent push notifications 🗣

The growth team wants to increase push notifications from 1 to 3 per day, saying the open rate will go up. How do you evaluate this?

**Reference answer:**

- **Clarify:** is "open rate" the open rate per notification, or the probability that a user opens the app each day? Is the goal short-term activity or long-term retention?
- **Decompose:** the upside: more chances to bring users back. The downside: annoyance, leading users to disable notifications or uninstall — losses that are irreversible.
- **Metrics:** primary: active days per user or 28-day retention. Secondary: open rate per notification (usually falls), opens per user. Guardrails: notification opt-out rate, uninstall rate, complaints.
- **Experiment:** a user-level A/B test, ideally with several arms (1 / 2 / 3 per day) to see the dose response; run long enough (at least several weeks) to observe fatigue and uninstalls.
- **Conclude:** judge by long-term retention, not short-term opens; segment users (highly active users may not need more notifications), and the end result may be a personalized notification frequency.

### Problem 14 · Raising the subscription price 🗣

A video subscription's monthly price is going from $15 to $18. How do you evaluate this decision?

**Reference answer:**

- **Clarify:** is the goal revenue or profit; does the increase apply to both new and existing users; how do competitors compare?
- **Decompose:** revenue = subscribers × price. With a 20% price increase, revenue rises as long as subscribers fall by less than about 17%. Also look at new-subscriber conversion, renewal rate, and the share downgrading to the free tier.
- **Hypothesize:** existing users are less price-sensitive (they have the habit), new users more so; some users will switch to an annual plan or to a competitor.
- **Data / experiment:** price experiments carry fairness and PR risks; common approaches are testing by region, testing only on new users, or analyzing reactions to past price increases and competitors' increases; look at renewal over 1–3 months, not just the first month.
- **Conclude:** estimate the price elasticity and combine it with long-term retention to compute the change in customer lifetime value; give "a range for the revenue change" and the risks (churn, word of mouth) rather than a single number.

### Problem 15 · Estimation 🗣

Estimate the number of ride-hailing trips per day in a city of 10 million people.

**Reference answer:**

- **Clarify:** ride-hailing only (excluding street-hailed taxis); weekdays or the average day?
- **Decompose:** trips = population × share of people who use ride-hailing × average trips per day for those people.
- **Hypothesize:** about 70% are adults who can hail a ride on a phone; about 30% of those use it regularly; regular users average about 3 trips per week, or about 0.4 per day.
- **Compute:** 10 million × 0.7 × 0.3 × 0.4 ≈ 840,000 trips / day. Add occasional users (e.g., 40% use it once a month: about 10 million × 0.7 × 0.4 ÷ 30 ≈ 90,000), for a total of about 900,000 trips.
- **Sanity check:** people in a city make about 2–3 trips per day, so the city has roughly 20–30 million trips per day; 900,000 is about 3%–5% of them, which is the right order of magnitude. Also compare with published figures for known large cities.
- **Conclude:** on the order of 1 million trips / day. The most uncertain input is "the share of regular users": if it doubles, the result nearly doubles; say what data could be used to calibrate it.

## Framework recap

| Question type | Core question | Points you must make |
|---|---|---|
| Metric change | Is it real? Where? Why? | Logging / definitions, numerator and denominator, segments, funnel, release log, external factors |
| Mix effects | Did every group change, or did the group shares change? | Simpson's paradox; decompose into mix effect + within-group effect |
| Metric design | What is the goal? How do you measure it? Can it be gamed? | North-star metric + decomposition + guardrails + how the metric could be gamed |
| Experiment interpretation | Are the results reliable? Which metric wins? | SRM, pre-specified primary metric, confidence interval, novelty effect, segments and multiple comparisons |
| Experiment design | Is the randomization unit right? | Network effects → randomize by region or time slice |
| Evaluation without an experiment | What is the counterfactual? | DiD, matching, regression discontinuity, time series — and state the assumptions |
| Estimation | Which factors can it be broken into? | State every assumption, compute, sanity-check, name the most uncertain assumption |

Every answer should end with **one clear recommendation** plus **the conditions it depends on**.
