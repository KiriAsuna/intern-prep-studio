### Problem 1 · Floor division and remainder

How many hours and minutes is 125 minutes?

```python
minutes = 125
hours = minutes // 60
rest = minutes % 60
print(hours, rest)
```

**Result:**

```text
2 5
```

> `//` is floor division and `%` is remainder, matching R's `%/%` and `%%`.

### Problem 2 · Powers

A principal of 1000 at 5% annual interest, compounded for 10 years: how much do you end up with (two decimal places)?

```python
amount = 1000 * (1 + 0.05) ** 10
print(f"{amount:.2f}")
```

**Result:**

```text
1628.89
```

> The power operator is `**`, not `^`.

### Problem 3 · Odd or even

Is 17 odd?

```python
17 % 2 == 1
```

**Result:**

```text
True
```

> A remainder of 1 when divided by 2 means odd. A comparison gives `True` / `False`, capitalized.

### Problem 4 · The result type of division

What types are `6 / 3` and `6 // 3`?

```python
print(type(6 / 3))
print(type(6 // 3))
```

**Result:**

```text
<class 'float'>
<class 'int'>
```

> `/` always returns a float (`2.0`); to get an integer from integer division, use `//`.

### Problem 5 · String to number

`a = "12"` and `b = "30"` are strings read from a file. Find their sum.

```python
a = "12"
b = "30"
int(a) + int(b)
```

**Result:**

```text
42
```

> Without conversion, `a + b` concatenates and gives `'1230'`.

### Problem 6 · Number to string

Use `rank = 3` to build `"No. 3"`.

```python
rank = 3
"No. " + str(rank)
```

**Result:**

```text
'No. 3'
```

> A number can't be `+`-ed to a string directly; that raises a `TypeError`.

### Problem 7 · A percentage with an f-string

Display `rate = 0.12345` as `12.3%`.

```python
rate = 0.12345
f"{rate * 100:.1f}%"
```

**Result:**

```text
'12.3%'
```

> `:.1f` means one decimal place; you can write an expression right inside the braces.

### Problem 8 · Building a sentence with an f-string

With `name = "Ann"` and `score = 91.456`, build `Ann scored 91.46`.

```python
name = "Ann"
score = 91.456
f"{name} scored {score:.2f}"
```

**Result:**

```text
'Ann scored 91.46'
```

### Problem 9 · String length, first and last character

For `s = "statistics"`: the length, the first character, and the last character.

```python
s = "statistics"
print(len(s), s[0], s[-1])
```

**Result:**

```text
10 s s
```

> `len()` matches R's `nchar()`; `s[-1]` is the last character.

### Problem 10 · String slicing

Extract the year and month from `"2026-01-15"`.

```python
date = "2026-01-15"
year = date[:4]
month = date[5:7]
print(year, month)
```

**Result:**

```text
2026 01
```

> Indices 5 and 6 are the two digits of the month; slices exclude the right end, so write `5:7`.

### Problem 11 · First and last element

The first and last elements of `xs = [10, 20, 30, 40, 50]`.

```python
xs = [10, 20, 30, 40, 50]
print(xs[0], xs[-1])
```

**Result:**

```text
10 50
```

> Indexing starts at 0. R's `xs[1]` is `xs[0]` in Python.

### Problem 12 · The first few and the last few

The first 3 and last 2 elements of `xs`.

```python
print(xs[:3])
print(xs[-2:])
```

**Result:**

```text
[10, 20, 30]
[40, 50]
```

> `xs[:3]` is indices 0, 1, 2; `xs[-2:]` runs from the second to last to the end.

### Problem 13 · Drop the first and last

Remove the first and last elements of `xs`, keeping the middle.

```python
xs[1:-1]
```

**Result:**

```text
[20, 30, 40]
```

> A right end of `-1` means "up to the last element" (not including it).

### Problem 14 · Modify one element

Change the second element of `ys = [1, 2, 3]` to 99.

```python
ys = [1, 2, 3]
ys[1] = 99
ys
```

**Result:**

```text
[1, 99, 3]
```

### Problem 15 · Append

Add a 4 to the end of `ys`.

```python
ys.append(4)
ys
```

**Result:**

```text
[1, 99, 3, 4]
```

> `append` modifies `ys` in place. Don't write `ys = ys.append(4)`; that would make `ys` become `None`.

### Problem 16 · Membership

Are `99` and `5` in `ys`?

```python
print(99 in ys, 5 in ys)
```

**Result:**

```text
True False
```

> R's `%in%`.

### Problem 17 · Sum, mean, range

For `scores = [72, 85, 90, 66, 85]`: the total, the average, and the highest minus the lowest.

```python
scores = [72, 85, 90, 66, 85]
total = sum(scores)
avg = sum(scores) / len(scores)
spread = max(scores) - min(scores)
print(total, avg, spread)
```

**Result:**

```text
398 79.6 24
```

> Base Python has no `mean()`; use `sum / len`.

### Problem 18 · Ascending and descending

Sort `scores` once from low to high and once from high to low.

```python
print(sorted(scores))
print(sorted(scores, reverse=True))
```

**Result:**

```text
[66, 72, 85, 85, 90]
[90, 85, 85, 72, 66]
```

> `sorted` returns a new list; `scores` itself is unchanged.

### Problem 19 · Median (odd count)

`scores` has 5 numbers. Find the median.

```python
s = sorted(scores)
s[len(s) // 2]
```

**Result:**

```text
85
```

> After sorting, take the exact middle: for 5 numbers the middle index is `5 // 2 = 2`. With an even count, average the two middle values.

### Problem 20 · The top k

The 3 highest scores in `scores`, highest first.

```python
sorted(scores, reverse=True)[:3]
```

**Result:**

```text
[90, 85, 85]
```

> Sort descending, then slice the first 3. Chain the two steps together.
