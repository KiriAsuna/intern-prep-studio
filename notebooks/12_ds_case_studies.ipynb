# 12 | DS Case Studies and Product Sense

Case interviews have no single right answer; what is tested is **whether you have structure**. Many people with strong statistics backgrounds lose points here: they jump straight to "build a model," or say whatever comes to mind.

Three common question types: **investigating a metric change**, **designing metrics**, and **launch decisions**. Remember one framework for each.

About 50 minutes.

> **How to use this lesson:** Answer each question **out loud** first, then open the reference answer to compare. Each answer ends with a short version you can say in an interview.

## A general rhythm for case questions

1. **Ask clarifying questions first** (30 seconds): How is the metric defined? Down relative to when? Sudden or gradual?
2. **State your framework**: "I'd look at it from these angles…" — this shows the interviewer you have structure;
3. **Go through each item**, saying clearly "what data I'd look at, and what it would mean if I saw X";
4. **Prioritize**: check the most likely and easiest-to-verify things first;
5. **Summarize and propose next steps.**

## Type 1: Investigating a metric change

"Metric X suddenly dropped by Y%. How do you investigate?" Framework:

1. **Confirm it's real:** has the metric definition, data pipeline, or logging changed; is it a reporting or definition issue; is it within normal fluctuation?
2. **Time dimension:** sudden or gradual? Does it coincide with a release, a campaign, or a holiday? What happened in the same period last year?
3. **Decompose the metric:** did the numerator or the denominator change? Look at them separately, especially for ratio metrics;
4. **Segment:** split by platform, app version, country / region, new vs. returning users, traffic source, and device, and see whether the drop is across the board or concentrated in one segment;
5. **Funnel:** which step's conversion dropped;
6. **Internal changes:** product releases, experiments, pricing, marketing spend, bugs;
7. **External factors:** competitors, seasonality, macro events, platform policies (e.g., an iOS update);
8. **Form hypotheses and test them**, quantify each factor's contribution, and recommend actions.

Rule of thumb: **first confirm it's real → then locate where → finally find why.**

**🗣 Say it out loud:** Signup conversion fell 8% last week. How do you investigate?

<details>
<summary><b>Reference answer</b></summary>

**Clarify:** the definition of conversion (signups ÷ landing-page visitors?); whether 8% is a relative or absolute drop; which week we're comparing to; whether it's a one-day plunge or the whole week running low.

**Confirm it's real:** have logging or the data pipeline changed recently; how have the absolute numerator and denominator each moved? If the denominator (visits) suddenly grew while signups stayed flat, the traffic mix probably changed (e.g., a campaign brought in lots of low-intent traffic), rather than the product getting worse.

**Locate:** split by platform, browser, country, traffic source, and new vs. returning visitors. If only Android dropped, check recent app releases; if only one channel dropped, check that channel's campaigns. Then walk each step of the signup funnel (landing page → click sign up → fill in form → verify → complete) to see which step's conversion changed.

**Find the cause:** compare against the release log, running experiments, the marketing calendar, and known incidents; externally, look at holidays and competitors.

**Conclusion:** propose the two or three most likely hypotheses with a way to test each, estimate how much of the drop each explains, and recommend action (roll back, fix a bug, adjust campaigns, or judge it normal fluctuation and keep watching).

**Short version:** I'd first clarify the metric definition and whether the drop is sudden or gradual, and verify it's real by checking logging and the numerator and denominator separately. Then I'd segment by platform, geography, traffic source, and new versus returning users, and walk the signup funnel to localize it. Finally I'd line it up against releases, experiments, and marketing changes, form hypotheses, and quantify each one's contribution.

</details>

## Type 2: Designing metrics

"How would you measure whether feature X is successful?" Framework:

1. **What is the feature's goal?** What problem does it solve for users, and what value does it create for the company;
2. **North-star metric:** the single metric that best represents the goal;
3. **Decompose:** break it into actionable drivers, e.g., revenue = traffic × conversion rate × average order value;
4. **Guardrail metrics:** things that must not get worse as a result;
5. **Metric quality:** can it be gamed, is it sensitive to changes, how soon can it be observed, does it conflict with long-term goals?

A good metric is directly tied to the goal, measurable, sensitive to product changes, hard to manipulate, and easy to understand.

**🗣 Say it out loud:** A video platform launched an "autoplay next episode" feature. How do you measure whether it's successful?

<details>
<summary><b>Reference answer</b></summary>

**Goal:** make it easier for users to keep watching, increasing engagement and long-term retention.

**Primary metric:** watch time per user, or completed episodes per user per day. But this metric can be "gamed": autoplay keeps videos playing when nobody is watching, inflating watch time. So pair it with **quality metrics**: completion rate of autoplayed episodes, exit rate within 30 seconds of playback starting, and the share of users who actively turn autoplay off.

**Long-term metrics:** 7-day / 28-day retention, subscription renewal rate.

**Guardrails:** user complaints and negative feedback, the share of users switching autoplay off in settings, bandwidth cost.

**How to test:** a user-level A/B test, run long enough to observe retention and let novelty effects wear off.

**Tradeoff to watch for:** short-term watch time goes up, but users feel interrupted and long-term retention actually falls. If the quality and long-term metrics point in a different direction from watch time, go with the long-term metrics.

**Short version:** The goal is deeper engagement and retention. I'd track watch time per user as the primary metric, but pair it with quality signals like completion rate of auto-played episodes and early exits, since autoplay can inflate watch time without real viewing. Guardrails would include opt-out rate and complaints, and I'd confirm with long-run retention in a user-level A/B test.

</details>

## Type 3: Conflicting metrics and launch decisions

**🗣 Say it out loud:** A new recommendation model increased click-through rate (CTR), but 30-day retention fell. Should you ship it?

<details>
<summary><b>Reference answer</b></summary>

**First confirm the results are reliable:** is randomization healthy (SRM check); are both results statistically significant; what are the effect sizes and confidence intervals; was retention a pre-specified metric?

**Understand why:** CTR up while retention is down is typically explained by the model pushing clickbait: content that's easy to click but disappointing once opened. Look at post-click metrics: dwell time, completion rate, bounce rate, negative feedback after clicks; also check whether the diversity of recommended content dropped.

**Decision:** CTR is a short-term proxy; retention is closer to the real business goal. If the retention drop is real and meaningful, **don't ship**, even though CTR is better. The next step is to change the model's objective (e.g., optimize "clicked and stayed more than X seconds," or predict long-term value directly) and run another experiment.

**General principle:** decide the primary and guardrail metrics in advance, along with which one wins when they conflict, to avoid cherry-picking favorable metrics after the fact.

**Short version:** First verify both effects are real and significant. CTR up with retention down suggests clickbait: content that earns clicks but disappoints. I'd check post-click engagement and diversity. Since retention is closer to the true objective than CTR, I wouldn't ship; I'd change the model's objective toward long-term value and retest.

</details>

**🗣 Say it out loud:** A model's overall accuracy is good, but it is clearly worse on a small user group. What do you do?

<details>
<summary><b>Reference answer</b></summary>

**First confirm:** how large is this group's sample? Is the gap beyond random fluctuation (give a confidence interval)?

**Find the cause:** ① too few samples from this group in the training data; ② this group's labels are lower quality or defined differently; ③ the features are less informative or more often missing for them; ④ their X–Y relationship is genuinely different, and a single model fits them poorly.

**Remedies:** collect more data from this group; reweight or oversample; add features or interaction terms that distinguish the group; use a separate model or separate threshold if needed; evaluate by group and make the worst group's performance one of the launch criteria.

**The bigger issue:** if it involves a protected attribute (sex, race, age), this is a fairness problem with possible legal and reputational risk, and the standard should be decided together with product and legal. Never let an overall metric paper over a subgroup's failure.

**Short version:** First check the subgroup's sample size and whether the gap is beyond noise. Then diagnose: under-representation, worse labels, less informative features, or a genuinely different relationship. Remedies include more data, reweighting, group-specific features or thresholds, and reporting metrics by group. If it involves a protected attribute, it's a fairness issue to raise explicitly.

</details>

## Type 4: Estimation and open-ended questions

Questions like "How many orders does city X get per day?" or "Is this feature worth building?" test decomposition and assumptions, not the accuracy of the answer.

Approach: break the big number into a product of estimable factors → state your assumption for each → compute the result → sanity-check it (compare with known numbers) → say which assumption is most uncertain and how it would affect the result.

**🗣 Say it out loud:** How would you evaluate a campaign that "gives new users a $10 coupon"?

<details>
<summary><b>Reference answer</b></summary>

**The question to answer:** how much **incremental** value the coupon creates, and whether it exceeds the cost. The key word is "incremental": many people who redeem the coupon would have ordered anyway.

**Ideal approach:** a randomized experiment. Randomly give the coupon to some new users and not others, and compare the two groups.

**Metrics:** focus on **net** revenue or profit per user (after the coupon cost), not just order rate; also look at first-order conversion and 30-day / 90-day repeat purchase and retention (the coupon may attract one-time bargain hunters).

**Calculation:** incremental profit = treatment profit per user − control profit per user, times the number of users; compare it with the campaign cost to get ROI.

**Caveats:** ① long-term effects: will users learn to wait for discounts before buying; ② heterogeneous effects: which users show the largest lift — you might target only them (uplift modeling); ③ if the coupon already went to everyone and there is no control group, you can only use quasi-experimental methods (e.g., comparing with pre-campaign cohorts, regression discontinuity, difference-in-differences), and you should state the assumptions.

**Short version:** The question is incremental value versus cost, since many recipients would have purchased anyway. I'd run a randomized holdout, measure net revenue or profit per user rather than just conversion, and track repeat purchase and retention to catch one-time bargain hunters. Then I'd look at heterogeneous effects to target the coupon where the uplift is largest.

</details>

## Checkpoint

Given a random "metric dropped" question, you can lay out a structured investigation within 5 minutes; you can design a primary metric and guardrail metrics for a new feature and point out how the metrics could be gamed; and when metrics conflict, you can say which one wins and why.
