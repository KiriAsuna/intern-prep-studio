### Problem 1 · Conversion rate by country

A user "converts" if they have at least one completed order. For each country, compute the number of users, the number converted, and the conversion rate (ignore users with a missing country).

```python
buyers = orders[orders["status"] == "completed"]["user_id"].unique()
u = users.copy()
u["converted"] = u["user_id"].isin(buyers)
(
    u.groupby("country")
    .agg(n_users=("user_id", "count"), n_converted=("converted", "sum"), rate=("converted", "mean"))
    .reset_index()
)
```

**Result:**

|   | country | n_users | n_converted | rate |
|---|---|---|---|---|
| 0 | CA | 2 | 1 | 0.5 |
| 1 | CN | 3 | 3 | 1.0 |
| 2 | US | 4 | 4 | 1.0 |

*3 rows × 4 columns*

> Starting from `users` makes the denominator all users. The sum of a boolean column is a count; its mean is a proportion.

### Problem 3 · ARPU

Average revenue per user: total amount of completed orders divided by the number of **all** users.

```python
orders[orders["status"] == "completed"]["amount"].sum() / len(users)
```

**Result:**

```text
66.5
```

> The denominator is all users (10), not users who ordered (8). In an interview, clarify the denominator first.

### Problem 4 · Repeat purchase rate

Among users with a completed order, the share who have 2 or more.

```python
cnt = orders[orders["status"] == "completed"].groupby("user_id")["order_id"].count()
(cnt >= 2).mean()
```

**Result:**

```text
0.625
```

> `cnt` is a Series with one number per user; compare it and take the mean to get a proportion.

### Problem 11 · Day-1 retention

Treat the first login as day 0; a user is retained on day 1 if they also logged in on day 1. Compute day-1 retention across all users.

```python
logins = events[events["event_type"] == "login"][["user_id", "event_date"]].drop_duplicates().copy()
logins["event_date"] = pd.to_datetime(logins["event_date"])
first = logins.groupby("user_id")["event_date"].min().reset_index().rename(columns={"event_date": "first_date"})
m = logins.merge(first, on="user_id")
m["day_n"] = (m["event_date"] - m["first_date"]).dt.days
m[m["day_n"] == 1]["user_id"].nunique() / first["user_id"].nunique()
```

**Result:**

```text
0.7
```

> Start date → merge back onto the activity table → compute the day number → filter on the day number and count users.

### Problem 12 · Funnel

Number of users who viewed (view), added to cart (cart) and purchased (purchase), and the conversion rate of each step relative to the previous one.

```python
f = (
    events.groupby("event_type")["user_id"].nunique()
    .reset_index()
    .rename(columns={"event_type": "step", "user_id": "users"})
)
steps = pd.DataFrame({"step": ["view", "cart", "purchase"]}).merge(f, on="step", how="left")
steps["step_rate"] = steps["users"] / steps["users"].shift(1)
steps
```

**Result:**

|   | step | users | step_rate |
|---|---|---|---|
| 0 | view | 10 | NaN |
| 1 | cart | 7 | 0.7 |
| 2 | purchase | 5 | 0.7143 |

*3 rows × 3 columns*

> Count users per event type, then left-join onto a table that lists the steps in funnel order; this fixes the order (and drops `login` along the way). `shift(1)` gets the previous step's count.

```python
daily = (
    orders[orders["status"] == "completed"]
    .groupby("order_date")["amount"].sum()
    .reset_index()
    .sort_values("order_date")
)
daily["roll3"] = daily["amount"].rolling(3).mean()
daily.head(6)
```

### Problem 15 · Longest login streak

The longest run of consecutive login days for each user.

```python
lg = events[events["event_type"] == "login"][["user_id", "event_date"]].drop_duplicates().copy()
lg["d"] = (pd.to_datetime(lg["event_date"]) - pd.Timestamp("2026-01-01")).dt.days
lg = lg.sort_values(["user_id", "d"])
lg["key"] = lg["d"] - lg.groupby("user_id").cumcount()
runs = lg.groupby(["user_id", "key"])["d"].count().reset_index()
runs.groupby("user_id")["d"].max().reset_index().rename(columns={"d": "max_streak"})
```

**Result:**

|   | user_id | max_streak |
|---|---|---|
| 0 | 1 | 3 |
| 1 | 2 | 2 |
| 2 | 3 | 4 |
| 3 | 4 | 1 |
| 4 | 5 | 2 |
| 5 | 6 | 3 |
| 6 | 7 | 1 |
| 7 | 8 | 2 |
| 8 | 9 | 2 |
| 9 | 10 | 1 |

*10 rows × 2 columns*

> Classic trick: subtract the within-group row number from the day number, and consecutive days get the same `key`. For example, days 1, 2, 3 minus 0, 1, 2 all give 1. Then count days per (user, key) and take the maximum.

### Problem 17 · Repeat purchase within 7 days of the first order

Among users who ordered, which ones ordered again within 7 days after their first order (not counting the day of the first order).

```python
o = orders.copy()
o["order_date"] = pd.to_datetime(o["order_date"])
first = o.groupby("user_id")["order_date"].min().reset_index().rename(columns={"order_date": "first_date"})
m = o.merge(first, on="user_id")
m["gap"] = (m["order_date"] - m["first_date"]).dt.days
m[(m["gap"] >= 1) & (m["gap"] <= 7)]["user_id"].unique().tolist()
```

```text
[1, 2]
```

### Problem 18 · Average spend per user by channel (including zeros)

Average completed amount per user for each channel, counting users with no orders as 0.

### Problem 20 · Department summary table

For each department: headcount, average salary, and number of people above the department average.

```python
e = employees.copy()
e["above"] = e["salary"] > e.groupby("dept")["salary"].transform("mean")
(
    e.groupby("dept")
    .agg(n=("emp_id", "count"), avg_salary=("salary", "mean"), n_above=("above", "sum"))
    .reset_index()
)
```

**Result:**

|   | dept | n | avg_salary | n_above |
|---|---|---|---|---|
| 0 | Eng | 4 | 121.25 | 1 |
| 1 | Mkt | 2 | 72.5 | 1 |
| 2 | Sales | 3 | 76.6667 | 1 |

*3 rows × 4 columns*

> First use `transform` to build a boolean column on every row, then summarize with `agg`.

## A fixed recipe for mixed problems

1. **What is one row of the result?** (a user / a day / a channel)
2. **Aggregate the detail rows to "one row per unit" first**: `groupby(...).agg(...)`, or sort and then `drop_duplicates`.
3. **Merge**: start from the denominator table (usually `users`), left join, fill missing values with 0.
4. **Compute the metric**: `.mean()` of a boolean column is a proportion, `shift` / `pct_change` compare with the previous period, `rank` ranks.
