## 1. Indentation is the code block

R uses curly braces `{ }` to say "these lines belong together." Python **uses indentation instead of braces**:

```r
# R
if (x > 0) {
  print("positive")
}
```

```python
# Python
if x > 0:
    print("positive")
```

Two rules:

1. The condition is followed by a **colon** `:`;
2. Code that belongs to the block is **indented 4 spaces** (one Tab). When the indentation ends, the block ends.

The condition doesn't need parentheses.

````python
x = 5
if x > 0:
    print("x is positive")
    print("this line is also inside the if")
print("this line is outside the if and always runs")
````

## 2. if / elif / else

`elif` is R's `else if`.

Comparison operators are the same as in R: `==`, `!=`, `<`, `<=`, `>`, `>=`.
Logical operators are **English words**: `and`, `or`, `not` (R's `&`, `|`, `!`).

````python
score = 83
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "C"
print(grade)

age = 25
if age >= 18 and age < 65:
    print("working age")
````

## 3. for loops

Same idea as R's `for (x in xs)`; just drop the parentheses and add a colon and indentation.

````python
xs = [10, 20, 30]
for x in xs:
    print(x * 2)
````

### range: generating a sequence of integers

| R | Python | Gives |
|---|---|---|
| `0:4` | `range(5)` | 0, 1, 2, 3, 4 |
| `1:5` | `range(1, 6)` | 1, 2, 3, 4, 5 |

Like slicing, **the right end is excluded**.

````python
for i in range(5):
    print(i)
````

## 4. Pattern 1: accumulate

Set up a variable first, then keep adding to it inside the loop. `total += x` is shorthand for `total = total + x`.

````python
nums = [3, 8, 5, 12, 7, 10]
total = 0
for x in nums:
    total += x
print(total)
````

## 5. Pattern 2: count

Almost identical to accumulating, except you add 1 each time.

````python
nums = [3, 8, 5, 12, 7, 10]
count = 0
for x in nums:
    if x > 6:
        count += 1
print(count)
````

## 6. Pattern 3: collect results

Start with an empty list `[]`, and `append` to it inside the loop.

````python
nums = [3, -1, 4, 0, 8]
positives = []
for x in nums:
    if x > 0:
        positives.append(x)
print(positives)
````

## 7. Index and element together: enumerate

In R you'd often write `for (i in seq_along(xs))`. Python uses `enumerate`, which gives you the index `i` and the element `x` at once:

````python
names = ["Ann", "Bob", "Cy"]
for i, name in enumerate(names):
    print(i, name)
````

## 8. while and break

`while` means "keep repeating as long as the condition holds"; `break` means "exit the loop immediately." They're used less than `for`; just know they exist.

````python
n = 100
steps = 0
while n > 1:
    n = n // 2
    steps += 1
print(steps)

for x in [4, 7, 9, 12]:
    if x > 8:
        print("The first number greater than 8 is", x)
        break
````
