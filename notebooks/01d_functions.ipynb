## 1. Defining and calling

```r
# R
square <- function(x) {
  x^2
}
```

```python
# Python
def square(x):
    return x ** 2
```

Start with `def`, end the line with a colon, and indent the function body.

````python
def square(x):
    return x ** 2

print(square(4))
print(square(1.5))
````

## 2. You must write return ⚠️

R functions automatically return the value of the last line. **Python does not.** Without `return`, the function returns `None`.

This is, hands down, the most common mistake R users make in Python.

````python
def square_wrong(x):
    x ** 2          # computed, but no return

result = square_wrong(4)
print(result)       # None
````

## 3. Default arguments

As in R, parameters can have default values, and you can name arguments when calling.

````python
def scale(x, factor=10):
    return x * factor

print(scale(3))             # uses the default 10
print(scale(3, 2))          # passed by position
print(scale(3, factor=5))   # passed by name
````

## 4. Edge cases and early return

What happens with `mean([])`? `len([])` is 0, and dividing by 0 raises an error.

Interviewers care a lot about whether you proactively consider **edge cases** like this (empty input, 0, negative numbers, a single element). The usual approach is to handle them at the top of the function:

```python
if len(xs) == 0:
    return None
```

As soon as `return` runs, the function ends immediately and the code after it doesn't run.

Change the function so that an empty list returns `None`, and otherwise it returns the mean.

## 5. Returning multiple values

Just separate them with commas, and unpack them with commas too.

````python
def min_max(xs):
    return min(xs), max(xs)

lo, hi = min_max([3, 9, 1, 7])
print(lo, hi)
````
