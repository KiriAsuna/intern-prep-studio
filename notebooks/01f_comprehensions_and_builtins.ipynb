## 1. List comprehensions

The "collect" pattern from 01c takes 4 lines:

```python
out = []
for x in xs:
    if x > 0:
        out.append(x ** 2)
```

A comprehension squeezes it into one line:

```python
out = [x ** 2 for x in xs if x > 0]
```

Read it as: **"for each x in xs, if x > 0, take the square of x."**

The R equivalent is the vectorized `xs[xs > 0]^2`. Python lists aren't vectorized; comprehensions are the substitute.

The `if` part is optional: `[x * 2 for x in xs]`.

````python
xs = [3, -1, 4, 0, 8]
print([x * 2 for x in xs])
print([x for x in xs if x > 0])
print([x ** 2 for x in xs if x > 0])
````

## 2. String methods

Strings have methods too (`variable.method()`). They **return a new string** and leave the original unchanged.

| Method | What it does | R equivalent |
|---|---|---|
| `s.lower()` / `s.upper()` | To lowercase / uppercase | `tolower()` / `toupper()` |
| `s.strip()` | Remove leading and trailing spaces | `trimws()` |
| `s.split(",")` | Split on commas into a list | `strsplit()` |
| `s.replace("a", "b")` | Replace | `gsub()` |
| `",".join(xs)` | Join a list into a string with commas | `paste(collapse = ",")` |

````python
s = "  Hello World  "
print(s.strip())
print(s.strip().lower())
print("a,b,c".split(","))
print("-".join(["2026", "01", "15"]))
````

## 3. The key argument of sorted

`sorted(xs, key=function)` means "sort by the result of this function."

````python
words = ["banana", "fig", "apple"]
print(sorted(words))                 # alphabetical
print(sorted(words, key=len))        # by length
print(sorted(words, key=len, reverse=True))
````

## 4. zip: walking two lists side by side

`zip(xs, ys)` yields one pair `(x, y)` at a time, lining up the two lists like a zipper.

````python
names = ["Ann", "Bob", "Cy"]
ages = [28, 35, 41]
for name, age in zip(names, ages):
    print(name, age)

print(dict(zip(names, ages)))    # turn them into a dict while we're at it
````

## 5. Dict comprehensions

Swap the square brackets for curly braces and write `key: value`. Just know they exist.

````python
words = ["banana", "fig", "apple"]
print({w: len(w) for w in words})
````
