# 13 | Talking About Your Research, and Behavioral Questions

For PhD students, "Tell me about your research" is almost guaranteed to come up, and it is often the first question. It shapes the interviewer's first impression: can this person explain something complex clearly, and do their skills transfer to our problems?

Being technically strong but unable to explain it clearly costs you points directly.

About 50 minutes. **The output of this lesson is a few paragraphs you write yourself.** Write them in "Notes for this lesson" at the bottom of this page; they are saved automatically.

## 1. The most common mistakes

- Starting with the literature review and still not having said what you did after two minutes;
- Piling on jargon (identifiability, posterior, shrinkage) that the interviewer doesn't understand and doesn't dare ask about;
- Only describing the method, not **why it matters** or **what the results were**;
- Blurring "we" and "I," so the interviewer can't tell what you personally did;
- Giving the same talk to every audience.

## 2. The 30-second version

Five sentences, one point each:

1. **Problem:** what real problem it solves (in language the listener cares about);
2. **Data:** what data, at what scale;
3. **Method:** what you did (one sentence, no jargon pile-up);
4. **Result:** what you found (ideally with a number);
5. **Your contribution:** what you personally were responsible for.

Template:

> "My research is about **[the problem, in plain words]**. I work with **[the data and its scale]**. I developed **[the method, in one sentence]**, which **[the result / how it beats existing methods]**. I designed the model, implemented it in **[language / tools]**, and ran the experiments."

## 3. The two-minute version

**Problem → Why it's hard → Data → Method → Validation → Result → Limitation → Impact**

What it adds beyond the 30-second version:

- **Why it's hard:** why existing methods fall short and where the difficulty lies. This is where you show your level;
- **Validation:** how you know the method is right (simulations, comparisons with baselines, checks on real data);
- **Limitation:** volunteer one limitation; it comes across as mature and credible;
- **Impact / transfer:** where else this method or skill set applies. **Connect it to the role you're interviewing for.**

Translate your research skills into industry language:

| What you did | What industry hears |
|---|---|
| Built hierarchical models for repeated-measures data | Modeling users with repeated observations; borrowing strength across segments |
| Analyzed model identifiability | Figuring out what the data can and cannot answer |
| Simulation studies | Validating methods with synthetic data; running A/A tests |
| EM / MCMC / variational inference | Large-scale optimization and approximate inference |
| Ran large batches of experiments on a cluster | Managing compute pipelines, handling failures and reruns |
| Wrote papers and responded to reviewers | Explaining technical decisions to a skeptical audience |

**🗣 Say it out loud:** How would you present the same research to a statistician, a data scientist, and a product manager?

<details>
<summary><b>Reference answer</b></summary>

The three versions differ in **what they emphasize**; it's not the same paragraph with a few words swapped.

**Statistician:** you can discuss the model structure, assumptions, identifiability, estimation method, and theoretical properties; they will press on whether the assumptions are reasonable.

**Data scientist:** describe the problem and what the data look like, the rough structure of the features and model, how you validated it, which baselines you compared against, how well it generalizes, and where it fails. They care whether "this method actually works in practice."

**Product manager:** describe what business question it answers, what decisions it helps make, and how reliable the conclusions are. Skip method details entirely; use examples and analogies. For example: "By pooling information across many small groups, we can give reliable estimates even for groups with very little data, so we know which ones really need attention first."

Practice: write down all three versions and record yourself saying each one.

**Short version:** I'd change the emphasis, not just the vocabulary. For a statistician: assumptions, identifiability, and estimation. For a data scientist: the data, validation, baselines, and failure modes. For a product manager: what question it answers and what decision it supports, with a concrete example and no method details.

</details>

## 4. Follow-up questions you will definitely get

Prepare an answer for each in advance and write down the key points:

- Why this model? Why not something simpler?
- What baselines did you compare against?
- How did you validate it? How do you know it's right?
- What didn't work?
- What was the biggest technical challenge?
- What assumption worries you most?
- How would this scale to 100× the data?
- What did **you** personally do?
- If you had one more month, what would you do?
- How does this relate to what we do here?

For "why not a simpler method," the best answer is that you **actually tried** the simpler method and can say where it fell short.

## 5. A failure story

Prepare a real example of "the method didn't work." Structure:

**Hypothesis → Evidence → Diagnosis → Adjustment → What you learned**

> "I initially thought …, so I tried …. It turned out that … (the specific symptom). By … (how you investigated), I traced the cause to …. I then switched to …, and the result was …. This taught me …, and now when I do something similar I first …"

What the interviewer wants to see: how you noticed the problem, how you diagnosed it, and whether it changed how you work. Don't pick a fake failure like "I'm too much of a perfectionist," and don't shift the blame to others.

## 6. Behavioral: STAR

- **S**ituation: the background (a sentence or two);
- **T**ask: the problem you needed to solve;
- **A**ction: what **you** did (the bulk of the story; say "I," not "we");
- **R**esult: the outcome (quantify where possible) + what you learned.

Keep each story to 1.5–2 minutes.

### Prepare 8 stories

Most behavioral questions can be answered with one of these 8 stories. Write down 4–6 keywords for each; don't memorize full scripts.

1. The hardest technical problem;
2. A failure or mistake;
3. A disagreement with a collaborator (or advisor);
4. Making progress when the goal was unclear;
5. A task under a tight deadline;
6. Initiating or leading something;
7. Learning something new quickly (**"learning Python and SQL from scratch to prepare for interviews" is a ready-made example**);
8. Explaining technical content to someone without a technical background.

**🗣 Say it out loud:** How do you answer "Tell me about yourself"?

<details>
<summary><b>Reference answer</b></summary>

60 to 90 seconds, three parts:

**Present:** who you are and what you're working on. "I'm a PhD student in statistics at …, working on … (your research in one sentence)."

**Past / skills:** two or three highlights relevant to the role. "My work involves building and validating hierarchical models for repeated-measures data, which means a lot of statistical modeling, simulation, and large-scale computation in R and Python."

**Future / why here:** "I'm looking for an internship where I can apply this to real product questions, and I'm particularly interested in this team because … (a specific reason)."

Don't recite your résumé chronologically; don't go over 2 minutes; end on "why I want to be here."

</details>

**🗣 Say it out loud:** How do you answer "Why industry?" / "Why this role?"

<details>
<summary><b>Reference answer</b></summary>

**Don't say:** academic jobs are hard to find, industry pays more, or you just want to try it.

**Directions you can take:** ① you want your work to have real impact faster, affecting actual decisions and users; ② you enjoy collaborating with people from different backgrounds and working on real problems with constraints; ③ industry has data at a scale and problems that academia doesn't offer; ④ specifics about this company / team: their product, their data, a technical blog post or paper of theirs, or an issue you noticed while using their product.

**Be specific.** Research what the team works on beforehand and prepare a sentence like "I noticed that you …, which is closely related to my work on …."

</details>

## 7. Questions to ask your interviewer

Always have questions for "Do you have any questions for me?" Prepare three or four:

- What does a typical project for an intern on this team look like?
- How does the team decide what to work on?
- What does success look like for this internship?
- What's the balance between modeling work and data / engineering work?
- What's the most challenging data problem the team is facing right now?

Don't ask things you could find with a quick web search, and don't ask about salary or vacation in a technical interview.
