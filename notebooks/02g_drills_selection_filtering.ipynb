### Problem 17 · Add a column

On a copy of `orders`, add a column `amount_cny` equal to the amount times 7 (a rough USD-to-CNY rate).

|   | order_id | amount | amount_cny |

### Problem 18 · Conditional column: np.where

Add a column `size`: `"big"` if the amount is ≥ 50, otherwise `"small"`.

```python
o = orders.copy()
o["size"] = np.where(o["amount"] >= 50, "big", "small")
o[["order_id", "amount", "size"]].head(5)
```

**Result:**

|   | order_id | amount | size |
|---|---|---|---|
| 0 | 1001 | 30.0 | small |
| 1 | 1002 | 50.0 | big |
| 2 | 1003 | 20.0 | small |
| 3 | 1004 | 15.0 | small |
| 4 | 1005 | 15.0 | small |

*5 rows × 3 columns*

> R: `ifelse()`.

### Problem 19 · Count each value: value_counts

How many orders have each status, and how many distinct users have placed an order.

```python
print(orders["user_id"].nunique())
orders["status"].value_counts()
```

**Result:**

```text
8
```

| status | count |
|---|---|
| completed | 14 |
| refunded | 2 |

*2 rows × 1 column*

> `value_counts()` is R's `table()`; `nunique()` is `length(unique())`.

### Problem 20 · Proportion = mean of a boolean column

The share of orders that are completed, and the list of names of CN users.

```python
print((orders["status"] == "completed").mean())
print(users[users["country"] == "CN"]["name"].tolist())
```

**Result:**

```text
0.875
['Chen', 'Fay', 'Jia']
```

> True counts as 1 and False as 0, so `.mean()` is the proportion. `.tolist()` turns a column into a plain list.

## Recap

| Task | Code |
|---|---|
| Select columns | `df["a"]`, `df[["a", "b"]]` |
| Filter rows | `df[df["a"] > 1]`; multiple conditions `(cond1) & (cond2)`, `|`, `~` |
| In a set of values | `df["a"].isin([...])` |
| New column | `df["c"] = ...`, `np.where(cond, if_true, if_false)` |
| Sort | `sort_values(["a", "b"], ascending=[True, False])` |
| Count | `value_counts()`, `nunique()`; `.mean()` of a boolean column is a proportion |
