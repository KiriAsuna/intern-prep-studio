```python
def count_sessions(timestamps, gap):
    if not timestamps:
        return 0
    ts = sorted(timestamps)
    sessions = 1
    for i in range(1, len(ts)):
        if ts[i] - ts[i - 1] > gap:
            sessions += 1
    return sessions
```

What it tests: **sort first**; "more than" means strictly greater; handle empty input separately. O(n log n).

<summary><b>Problem 2 answer: Longest run of 1s</b></summary>

```python
def longest_ones(nums, k):
    left = 0
    zeros = 0
    best = 0
    for right in range(len(nums)):
        if nums[right] == 0:
            zeros += 1
        while zeros > k:
            if nums[left] == 0:
                zeros -= 1
            left += 1
        best = max(best, right - left + 1)
    return best
```

What it tests: restate the problem as "**the longest window containing at most k zeros**". It's the variable-length sliding window template (04c): when the number of zeros in the window exceeds k, shrink from the left. O(n).

<summary><b>Problem 3 answer: Minimum meeting rooms</b></summary>

```python
import heapq

def min_meeting_rooms(intervals):
    ends = []                                  # heap: end times of the rooms currently in use
    for start, end in sorted(intervals):
        if ends and ends[0] <= start:          # the room that frees up earliest is now empty; reuse it
            heapq.heappop(ends)
        heapq.heappush(ends, end)
    return len(ends)
```

What it tests: sort by start time + a min-heap (04f). The heap stores each room's end time, and the top of the heap is the room that frees up first.

An alternative without a heap: record every start time as +1 and every end time as −1, sort by time (at the same moment, ends come before starts), and the maximum running total is the answer.

```python
from collections import deque

def shortest_path_grid(grid):
    rows, cols = len(grid), len(grid[0])
    if grid[0][0] == 1 or grid[rows - 1][cols - 1] == 1:
        return -1
    visited = {(0, 0)}
    q = deque([(0, 0, 0)])                     # (row, col, steps taken so far)
    while q:
        r, c, steps = q.popleft()
        if r == rows - 1 and c == cols - 1:
            return steps
        for dr, dc in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
            nr, nc = r + dr, c + dc
            if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] == 0 and (nr, nc) not in visited:
                visited.add((nr, nc))
                q.append((nr, nc, steps + 1))
    return -1
```

What it tests: **minimum steps → BFS** (04h). Carry the step count in the queue, use `visited` to avoid revisiting, and loop over the four directions with `(dr, dc)`. O(rows × cols).
