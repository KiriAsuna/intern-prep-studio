# 07a | Linear regression: spoken review

"What are the assumptions of linear regression?" is one of the most frequently asked questions in statistics and DS interviews. Goal: answer it straight away and hold up under three or four layers of follow-up questions.

About 45 minutes.

> **How to use this page:** Answer each question **out loud** first (or write it in your notes for this page), then open the reference answer and compare. Statistics is your home turf, so the point here is not whether you understand it but whether you can explain it clearly in 30–60 seconds. Each reference answer ends with a **Short version**: the spoken answer in a few sentences. Practice it until it comes out naturally.

## 1. Model and estimation

$$Y = X\beta + \varepsilon, \qquad \hat\beta = \arg\min_\beta \|Y - X\beta\|^2 = (X^\top X)^{-1} X^\top Y$$

## 2. Assumptions: layer them by "what they buy you"

This is the key to the answer. Don't recite five or six assumptions as a flat list; explain **what conclusion each layer of assumptions buys**.

**For unbiasedness / consistency:**
- The model is linear in the parameters;
- X has full column rank (no perfect collinearity);
- Zero conditional mean: $E[\varepsilon \mid X] = 0$ (exogeneity).

**For Gauss–Markov (OLS is the best linear unbiased estimator, BLUE), add:**
- Homoskedastic, uncorrelated errors: $\mathrm{Var}(\varepsilon \mid X) = \sigma^2 I$.

**For exact t / F tests in finite samples, add:**
- Normally distributed errors.

⚠️ Normality is **not** required for OLS to be unbiased, and inference in large samples does not need it either (the CLT takes care of it). Calling it the most important assumption is a common mistake.

**🗣 Say it out loud:** Answer without notes: What are the assumptions of OLS?

<details>
<summary><b>Reference answer</b></summary>

I'd layer them by purpose. For OLS to be **unbiased** you need linearity in the parameters, full rank X, and errors with zero conditional mean. For the **Gauss–Markov** result (OLS is the best linear unbiased estimator) you also need homoskedastic, uncorrelated errors. **Normality** is only needed for exact t and F tests in small samples; in large samples the central limit theorem covers it. In practice the one I care about most is zero conditional mean, because violating it biases the coefficients, whereas violating the others usually only affects the standard errors, which robust standard errors can fix.

**Short version:** I'd group them by what they buy you. For unbiasedness: linearity in parameters, no perfect multicollinearity, and zero conditional mean of the errors. For Gauss–Markov efficiency, add homoskedastic, uncorrelated errors. Normality is only needed for exact small-sample t and F tests. In practice the exogeneity assumption matters most, because violating it biases the coefficients, while the others mainly affect standard errors.

</details>

## 3. Why zero conditional mean matters most

$$\hat\beta = \beta + (X^\top X)^{-1} X^\top \varepsilon \quad\Rightarrow\quad E[\hat\beta \mid X] = \beta + (X^\top X)^{-1} X^\top E[\varepsilon \mid X]$$

When $E[\varepsilon \mid X] = 0$ the second term vanishes and the estimator is unbiased.

**🗣 Say it out loud:** When does omitted variable bias occur? How do you tell the direction of the bias?

<details>
<summary><b>Reference answer</b></summary>

The omitted variable must satisfy **both**: (1) it affects Y; (2) it is correlated with some X in the model. It then gets absorbed into the error term, making the error correlated with X and breaking zero conditional mean.

Direction: with true model $Y = \beta X + \gamma Z + \varepsilon$, after omitting Z the expected coefficient on X is $\beta + \gamma\delta$, where $\delta$ is the coefficient from regressing Z on X. Sign of the bias = (direction of Z's effect on Y) × (direction of Z's correlation with X). Example: studying the effect of education on earnings while omitting "ability"; ability has a positive effect on earnings and is positively correlated with education, so the education coefficient is overstated.

**Short version:** When the omitted variable both affects the outcome and is correlated with an included regressor. The bias equals the omitted variable's effect on Y times its association with X, so its sign is the product of those two signs.

</details>

## 4. Heteroskedasticity

**🗣 Say it out loud:** Under heteroskedasticity, are the OLS coefficients still unbiased? What goes wrong? What do you do?

<details>
<summary><b>Reference answer</b></summary>

As long as zero conditional mean holds, the coefficients are still unbiased and consistent. The problems: (1) the classical standard-error formula is no longer correct, so t-tests and confidence intervals are invalid; (2) OLS is no longer the most efficient estimator.

Remedies: use heteroskedasticity-robust standard errors (HC / White / sandwich), the default in practice; or transform Y (e.g. take logs); if the variance structure is known, use weighted least squares (WLS). Diagnosis: plot residuals against fitted values and look for a "funnel"; or run a Breusch–Pagan test.

**Short version:** The coefficients stay unbiased as long as exogeneity holds. What breaks is the usual standard-error formula, so inference is wrong, and OLS is no longer efficient. I'd use heteroskedasticity-robust standard errors by default, or transform the outcome, or use weighted least squares.

</details>

## 5. Multicollinearity

- **Perfect collinearity:** $X^\top X$ is not invertible and the coefficients are not uniquely determined (e.g. the dummy variable trap);
- **High but imperfect collinearity:** coefficients are still unbiased, but their **variance is inflated**; estimates are unstable and signs can be counterintuitive.

**🗣 Say it out loud:** How do you detect multicollinearity? How does it affect prediction vs. interpretation? How do you handle it?

<details>
<summary><b>Reference answer</b></summary>

**Detection:** the variance inflation factor $\mathrm{VIF}_j = 1/(1 - R_j^2)$, where $R_j^2$ is the R² from regressing the j-th predictor on the other predictors; as a rule of thumb, a VIF above 5 or 10 deserves attention. Other signs: no individual coefficient is significant but the overall F-test is; coefficients change drastically when a variable is added or removed.

**Effect:** little effect on prediction (as long as the correlation structure among the variables stays the same in new data); a large effect on interpreting individual coefficients.

**Handling:** drop or combine highly correlated variables, use ridge, do PCA, or accept it (if the goal is only prediction).

**Short version:** I'd check variance inflation factors. Multicollinearity doesn't bias the coefficients or hurt prediction much, but it inflates their variance so individual effects are unstable. Options are dropping or combining variables, ridge regression, or simply accepting it if the goal is prediction.

</details>

## 6. R²

**🗣 Say it out loud:** What is the difference between R² and adjusted R²? Does a high R² mean the model is good?

<details>
<summary><b>Reference answer</b></summary>

$R^2 = 1 - \mathrm{SSE}/\mathrm{SST}$ is the share of variance explained by the model. On the training set, R² never decreases when you add a variable, even pure noise. Adjusted R² penalizes the number of variables and decreases when a new variable's improvement does not make up for the degree of freedom it uses.

A high R² does not mean a good model: it may be overfitting; it says nothing about causation; it does not mean the model is correctly specified (in time series, two trending variables can have a very high R² with no relationship at all). Conversely, a low R² does not mean the model is useless; the coefficients can still be precisely estimated and meaningful. To judge predictive ability, look at performance on a **test set**.

**Short version:** R-squared never decreases when you add predictors, so it can't be used for model selection; adjusted R-squared penalizes extra parameters. A high R-squared doesn't imply a good or causal model, and out-of-sample performance is what matters for prediction.

</details>

## 7. Interpreting coefficients

| Model | Meaning of β |
|---|---|
| $y \sim x$ | A 1-unit increase in x changes y by β units on average |
| $\log y \sim x$ | A 1-unit increase in x changes y by about 100·β% |
| $y \sim \log x$ | A 1% increase in x changes y by about β/100 units |
| $\log y \sim \log x$ | A 1% increase in x changes y by about β% (elasticity) |

Always add "**holding the other variables constant**".

**Interaction terms:** in $Y = \beta_0 + \beta_1 X + \beta_2 Z + \beta_3 XZ + \varepsilon$, the marginal effect of X on Y is $\beta_1 + \beta_3 Z$, which varies with Z. Here $\beta_1$ is only the effect of X when Z = 0.

**Categorical variables:** k categories use k − 1 dummy variables; each coefficient is the difference relative to the reference category.

**🗣 Say it out loud:** When can a regression coefficient be interpreted as a causal effect?

<details>
<summary><b>Reference answer</b></summary>

Regression by itself only gives conditional associations. A causal interpretation requires additional identification assumptions: no omitted confounders (given the controls, treatment is "as good as random"), no reverse causality, and no bias introduced by sample selection or by controlling for a mediator or collider. The most reliable source is a randomized experiment; with observational data you rely on a research design (instrumental variables, difference-in-differences, regression discontinuity, etc.) plus explicitly stated assumptions.

**Short version:** Only under additional identification assumptions: no unmeasured confounding given the controls, no reverse causality, and no selection or bad-control bias. Randomized experiments give that by design; with observational data you need a design such as IV, difference-in-differences, or regression discontinuity.

</details>

## 8. Common follow-up questions

**🗣 Say it out loud:** If normality isn't required, why do people keep drawing QQ plots?

<details>
<summary><b>Reference answer</b></summary>

(1) In small samples, the exactness of t / F tests depends on normality; (2) a QQ plot exposes heavy tails and outliers, which make estimates unstable; (3) clear non-normality is often a sign of misspecification (a missing nonlinearity, a log that should have been taken); (4) if you need **prediction intervals**, the shape of the error distribution matters directly.

</details>

**🗣 Say it out loud:** What is the difference between an outlier, a high-leverage point and an influential point?

<details>
<summary><b>Reference answer</b></summary>

Outlier: far from the fitted line **in the Y direction** (large residual). High-leverage point: far from the other points **in X space**, with a large potential "pull" on the fitted line. Influential point: removing it noticeably changes the fit; usually a point with both high leverage and a sizable residual, measured by Cook's distance.

**Short version:** An outlier is unusual in Y given X; a high-leverage point is unusual in X; an influential point actually changes the fit when removed, typically high leverage combined with a large residual, and is measured by Cook's distance.

</details>

**🗣 Say it out loud:** Why is a prediction interval wider than a confidence interval for the mean response?

<details>
<summary><b>Reference answer</b></summary>

The confidence interval only reflects uncertainty in the estimated **position of the regression line** (coming from $\hat\beta$). A prediction interval is for a **new individual observation**, so it also adds that individual's own random error $\varepsilon$. However large the sample, the former shrinks to 0 while the latter always leaves at least $\sigma$.

**Short version:** The confidence interval covers uncertainty in the estimated mean; the prediction interval also includes the irreducible noise of a new observation, so it never shrinks below the error standard deviation.

</details>
