### Problem 1 · Assignment doesn't copy

Guess the output first: after `b = a`, if you add an element to `b`, does `a` change?

```python
a = [1, 2]
b = a
b.append(3)
print(a, b)
```

**Result:**

```text
[1, 2, 3] [1, 2, 3]
```

> `b = a` just gives the same list a second name. This is the number one mistake for R users.

### Problem 2 · If you want a copy, use copy

Make it so that changing `b` leaves `a` unchanged.

```python
a = [1, 2]
b = a.copy()
b.append(3)
print(a, b)
```

**Result:**

```text
[1, 2] [1, 2, 3]
```

### Problem 3 · A function can modify the list outside it

Guess the output first: after the `append` inside the function, has `data` outside changed?

```python
def add_zero(xs):
    xs.append(0)

data = [5, 6]
add_zero(data)
data
```

**Result:**

```text
[5, 6, 0]
```

> So unless the problem asks for in-place modification, it's safer to build a new list inside the function and return it.

### Problem 4 · A shallow copy copies only one level

Guess the output first: modify an inner list after `copy()`.

```python
a = [[1], [2]]
b = a.copy()
b[0].append(9)
print(a, b)
```

**Result:**

```text
[[1, 9], [2]] [[1, 9], [2]]
```

> The outer list is new, but the inner `[1]` is still the same object. To copy the inner levels too, use `copy.deepcopy(a)`.

### Problem 5 · == vs. is

For two lists with the same contents, what do `==` and `is` give?

```python
a = [1, 2]
b = [1, 2]
c = a
print(a == b, a is b, a is c)
```

**Result:**

```text
True False True
```

> `==` compares contents; `is` checks whether it's the same object. Use `==` for everyday comparisons.

### Problem 6 · Checking for None

Write a function `describe(x)`: return `"missing"` when `x` is `None`, otherwise return `"present"`.

```python
def describe(x):
    if x is None:
        return "missing"
    return "present"

print(describe(None), describe(0))
```

**Result:**

```text
missing present
```

> The convention for checking `None` is `is None`. Note that 0 is not `None`.

### Problem 7 · try / except: keep only what converts

`raw = ["3", "x", "10", "", "7"]`: convert whatever can become an integer and skip the rest.

```python
raw = ["3", "x", "10", "", "7"]
nums = []
for t in raw:
    try:
        nums.append(int(t))
    except ValueError:
        pass
nums
```

**Result:**

```text
[3, 10, 7]
```

> `pass` means "do nothing." Both `int("")` and `int("x")` raise `ValueError`.

### Problem 8 · try / except: division by 0

Write a function `ratio(a, b)` that returns `None` when `b` is 0, using `try / except`.

```python
def ratio(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return None

print(ratio(6, 3), ratio(1, 0))
```

**Result:**

```text
2.0 None
```

> The early return with `if b == 0` from 01d works too; you should recognize both styles.

### Problem 9 · FizzBuzz

For 1 through 15: replace multiples of 3 with `Fizz`, multiples of 5 with `Buzz`, and multiples of both with `FizzBuzz`.

```python
def fizzbuzz(n):
    out = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            out.append("FizzBuzz")
        elif i % 3 == 0:
            out.append("Fizz")
        elif i % 5 == 0:
            out.append("Buzz")
        else:
            out.append(i)
    return out

fizzbuzz(15)
```

**Result:**

```text
[1, 2, 'Fizz', 4, 'Buzz', 'Fizz', 7, 8, 'Fizz', 'Buzz', 11, 'Fizz', 13, 14, 'FizzBuzz']
```

> `% 15` must be checked first; otherwise 15 is treated as a multiple of 3 before you get there.

### Problem 10 · Remove duplicates while keeping order

`[3, 1, 3, 2, 1, 5]` → `[3, 1, 2, 5]`.

```python
def dedupe(xs):
    seen = set()
    out = []
    for x in xs:
        if x not in seen:
            seen.add(x)
            out.append(x)
    return out

dedupe([3, 1, 3, 2, 1, 5])
```

**Result:**

```text
[3, 1, 2, 5]
```

> `set(xs)` also removes duplicates, but doesn't guarantee order.

### Problem 11 · Are there duplicates?

Write a function `has_duplicate(xs)`.

```python
def has_duplicate(xs):
    return len(set(xs)) != len(xs)

print(has_duplicate([1, 2, 3]), has_duplicate([1, 2, 1]))
```

**Result:**

```text
False True
```

> If deduplicating makes it shorter, there were duplicates.

### Problem 12 · The most frequent element

Write a function `most_common(xs)`.

```python
def most_common(xs):
    counts = {}
    for x in xs:
        counts[x] = counts.get(x, 0) + 1
    best = None
    best_count = 0
    for x, c in counts.items():
        if c > best_count:
            best = x
            best_count = c
    return best

most_common([3, 1, 3, 2, 1, 3])
```

**Result:**

```text
3
```

> Two steps: count, then find the max.

### Problem 13 · The first non-repeating character

The first character in `"swiss"` that appears only once.

```python
def first_unique(s):
    counts = {}
    for ch in s:
        counts[ch] = counts.get(ch, 0) + 1
    for ch in s:
        if counts[ch] == 1:
            return ch
    return None

first_unique("swiss")
```

**Result:**

```text
'w'
```

> The second pass follows the original string's order; the first character with a count of 1 is the answer.

### Problem 14 · Anagrams

Are `"listen"` and `"silent"` made of the same letters?

```python
def char_counts(s):
    counts = {}
    for ch in s:
        counts[ch] = counts.get(ch, 0) + 1
    return counts

print(char_counts("listen") == char_counts("silent"))
print(sorted("listen") == sorted("silent"))
```

**Result:**

```text
True
True
```

> Two approaches: compare count dicts (O(n)), or sort the letters and compare (O(n log n)). `sorted("listen")` splits the string into single letters and sorts them, giving a list. Comparing two dicts with `==` ignores key order.

### Problem 15 · Palindromes

Write a function `is_palindrome(s)` that returns `True` if the string reads the same forward and backward.

```python
def is_palindrome(s):
    n = len(s)
    for i in range(n // 2):
        if s[i] != s[n - 1 - i]:
            return False
    return True

print(is_palindrome("level"), is_palindrome("stats"), is_palindrome("data"))
```

**Result:**

```text
True True False
```

> Compare the `i`-th character with the `(i + 1)`-th from the end (index `n - 1 - i`); you only need to check the first half.

### Problem 16 · Two Sum

`nums = [2, 7, 11, 15]`: find the indices of two numbers that sum to 9.

```python
def two_sum(nums, target):
    seen = {}                      # number -> index
    for i, x in enumerate(nums):
        need = target - x
        if need in seen:
            return seen[need], i
        seen[x] = i
    return None

print(two_sum([2, 7, 11, 15], 9))
print(two_sum([3, 2, 4], 6))
```

**Result:**

```text
(0, 1)
(1, 2)
```

> Record "number → index" as you go; each dict lookup is O(1), so the total is O(n). Nested loops would be O(n²).

### Problem 17 · Find the missing number

`[3, 0, 1]` is 0 through 3 with one number missing. Find it.

```python
def missing_number(nums):
    n = len(nums)
    return n * (n + 1) // 2 - sum(nums)

missing_number([3, 0, 1])
```

**Result:**

```text
2
```

> The sum of 0 through n is `n(n+1)/2`; subtract the actual sum to get the missing number.

### Problem 18 · Longest run of consecutive 1s

How long is the longest run of consecutive 1s in `[1, 1, 0, 1, 1, 1, 0, 1]`?

```python
def longest_ones(xs):
    best = 0
    run = 0
    for x in xs:
        if x == 1:
            run += 1
            if run > best:
                best = run
        else:
            run = 0
    return best

longest_ones([1, 1, 0, 1, 1, 1, 0, 1])
```

**Result:**

```text
3
```

> `run` tracks the length of the current run and resets to zero on a 0; `best` tracks the longest seen so far.

### Problem 19 · Best time to buy and sell a stock

Given daily prices `[7, 1, 5, 3, 6, 4]`, buy once and sell once: what's the maximum profit?

```python
def max_profit(prices):
    if len(prices) == 0:
        return 0
    lowest = prices[0]
    best = 0
    for p in prices:
        if p < lowest:
            lowest = p
        if p - lowest > best:
            best = p - lowest
    return best

max_profit([7, 1, 5, 3, 6, 4])
```

**Result:**

```text
5
```

> Walk through once, remembering "the lowest price so far"; selling today earns `p - lowest`. Buy at 1, sell at 6, profit 5. Handle the empty list first, or `prices[0]` raises an error.

### Problem 20 · Merge two sorted lists

Merge `[1, 4, 7]` and `[2, 3, 8, 9]` into one sorted list without using `sorted`.

```python
def merge_sorted(a, b):
    i = 0
    j = 0
    out = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            out.append(a[i])
            i += 1
        else:
            out.append(b[j])
            j += 1
    out = out + a[i:] + b[j:]
    return out

merge_sorted([1, 4, 7], [2, 3, 8, 9])
```

**Result:**

```text
[1, 2, 3, 4, 7, 8, 9]
```

> Two pointers `i` and `j`, one per list; take the smaller one each time. Once one side runs out, append whatever is left on the other side.

## Recap of this lesson's syntax

| Task | Code |
|---|---|
| Copy a list | `b = a.copy()` (`b = a` is not a copy) |
